1. 클래스 변수(정적변수), 인스턴스 변수
- 정적변수(=클래스 변수) 동일시 되는 용어, 클래스명, 변수명으로 클래스명을 정의합니다. 
- [개념 핵심!]
    - 정적 변수(클래스 변수) = 클래스가 가지는 공용 값
    - 인스턴스 변수 = 각각의 객체가 가지는 개별 값

```python
class User:
    pass 

user = User()
user.name = "김철수"
user.age = 20

print(user.name,user.age)
```

```python
class User:
    count = 0
```

```python
user1 = User()
user2 = User()
```

```python
user1.count = 10
print(user1.count,user2.count)
```

```python
User.count = 100
print(User.count)
```

2. @classmethod, @staticmetod
- 객체를 생성하지 않아도 접근할 수 있는 메서드를 정의할 때

```python
class Calculator:
    count = 0

    # clz -> calculator 클래스 객체 참조 변수
    # clz -> 클래스 변수, 다른 클래스 메서드를 접근하기 위한 목적
    @classmethod
    def add(clz, num1, num2): # 첫번째 매개변수가 class를 참고함
        clz.count += 1
        print("호출 횟수:", clz.count)
        return num1 + num2 

    @staticmethod
    def minus(num1, num2): # 객체를 생성안하는 예제 / 첫번째 매개변수가 없음
        return num1 - num2 

```

```python
Calculator.add(10,20)
```
