## 1. 객체지향과 클래스, 인스턴스의 관계

<aside>
💡

우리는 지금까지 코드를 위에서 아래로 순서대로 실행하는 절차적인 방식에 익숙해져 있습니다. 하지만 프로그램의 규모가 커지면 데이터와 함수가 복잡하게 얽혀 관리하기가 매우 까다로워집니다.

이를 해결하기 위해 서로 관련 있는 데이터와 기능을 하나로 묶어 관리하는 '객체지향(OOP)' 방식이 등장했습니다. 먼저 설계도와 실제 물건의 비유를 통해 클래스의 기본 개념을 알아보겠습니다.

### 1.1 객체지향과 절차지향

<aside>

**절차지향과 객체지향의 차이**

---

절차지향은 프로그램을 '실행할 명령의 순서'로 바라봅니다. 그래서 데이터와 행동(함수)을 따로 취급합니다.

**절차지향 - 데이터와 함수가 서로 분리되어 있습니다.**

```python
user_name = "스파르타"
def print_user_name(name):
    print(f"회원 이름: {name}")
print_user_name(user_name)
```

반면 객체지향은 프로그램을 **'고유한 상태(데이터)와 행동(메서드)을 가진 독립적인 객체들의 모임'**으로 바라봅니다. 관련된 상태와 행동을 객체 단위로 묶으면 역할을 분리하고 코드를 재사용하기 쉬워질 수 있습니다. 다만 적절한 책임 분리가 함께 이루어져야 합니다.

**객체지향 - 데이터와 함수가 하나의 객체 안에 묶여 있습니다.**

</aside>

### 1.2 클래스(Class)와 인스턴스(Instance)

<aside>

**설계도와 실제 물건의 관계**

---

**클래스**는 객체를 만들기 위해 속성과 행동을 미리 정의해 둔 '설계도' 또는 '틀'입니다. 

**인스턴스**는 이 설계도를 바탕으로 메모리에 실제로 만들어진 '구체적인 물건(객체)'을 뜻합니다. 붕어빵 틀(클래스)을 하나 만들어 두면, 이를 이용해 서로 다른 재료를 넣은 붕어빵(인스턴스)을 원하는 만큼 찍어낼 수 있는 것과 같은 원리입니다.

```python
class Car:
    pass  # 빈 설계도를 만듭니다.

# 설계도를 바탕으로 실제 차량 객체들을 생성합니다.
car_a = Car()
car_b = Car()

# 두 차량이 메모리에서 서로 다른 객체인지 확인해 봅니다.
print(type(car_a))  # <class '__main__.Car'>
print(car_a is car_b)  # False (둘은 완전히 독립된 별개의 물건입니다.)
```

</aside>

## 2. 생성자와 self 키워드로 객체 제어하기

<aside>
💡 클래스라는 설계도를 정의했다면, 이제 이 설계도를 바탕으로 실제 물건(객체)을 만들 때 고유한 값들을 채워 넣을 수 있어야 합니다. 

파이썬은 객체가 생성되는 시점에 자동으로 특수한 함수를 실행하여 초기화 작업을 수행합니다. 이때 사용되는 초기화 함수(생성자)와 자기 자신을 가리키는 self 키워드의 역할을 자세히 살펴보겠습니다.

</aside>

### 2.1 초기화 함수 __init**__** (생성자)

<aside>

**객체의 초기 상태 설정**

---

**__init__** 메서드는 객체가 생성될 때 파이썬이 자동으로 가장 먼저 호출하는 특수한 함수입니다.

객체가 처음 만들어지는 순간, 그 객체가 가져야 할 고유한 데이터(속성)들을 선언하고 최초의 값을 채워 넣어주는 역할을 수행합니다.

```python
class Character:
    def __init__(self, nickname):
        self.nickname = nickname  # 캐릭터가 생성될 때 이름을 지어줍니다.

# 인스턴스를 만드는 순간 괄호 안의 값이 생성자로 전달됩니다.
player = Character("스파르타")
print(player.nickname)  # 스파르타
```

</aside>

### 2.2 self 키워드

<aside>

**객체 자기 자신을 가리키는 참조 변수**

---

self는 현재 메모리에 생성되어 동작 중인 '**인스턴스 자신**'을 의미합니다.

클래스 내부에서 메서드를 정의할 때는 **첫 번째 매개변수로 반드시 self를 지정해야 합니다.** 그래야 파이썬이 "아, 지금 이 특정 객체의 데이터를 다루는 중이구나" 하고 정확하게 인지할 수 있습니다.

</aside>

### 2.3 속성(Attribute)과 인스턴스 메서드(Method)

<aside>

**변수와 함수의 결합**

---

클래스 내부에 선언되어 객체의 상태를 나타내는 변수를 '**속성(Attribute)**'이라 하고, self를 매개변수로 받아 객체의 데이터를 다루는 함수를 '**인스턴스 메서드(Instance Method)**'라고 부릅니다.

반드시 **“self.속성명”** 형태로 접근해야만 객체별로 독립된 고유의 저장 공간에 데이터가 안전하게 기록되고 보존됩니다.

```python
class Smartphone:
    def __init__(self, model):
        self.model = model  # 속성
        self.battery = 100  # 속성 기본값

    def charge(self):  # 인스턴스 메서드
        self.battery = 100
        print(f"{self.model} 충전 완료!")

phone = Smartphone("Galaxy")
phone.charge()
```

</aside>

## 3. 은행 계좌(BankAccount) 클래스 만들기

<aside>
💡

앞서 배운 객체지향 개념을 바탕으로 실제 코드를 구현해 보겠습니다.

은행 계좌를 관리하는 BankAccount 클래스를 직접 설계한 뒤, 철수와 영희의 인스턴스를 각각 독립적으로 생성하여 데이터가 서로 섞이지 않고 개별 관리되는 모습을 주피터 노트북 환경에서 확인해 보겠습니다.

</aside>

### 3.1 BankAccount 클래스 설계하기

<aside>

**클래스 설계하기**

---

예금주 이름과 초기 잔액을 입력받아 계좌 객체의 속성을 초기화하고, 입출금을 처리하는 메서드(기능)들을 정의합니다.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner          # 예금주
        self.balance = balance      # 잔액

    # print()를 쓸 때 친숙하게 출력되도록 돕습니다.
    def __str__(self):
        return f"{self.owner}님의 계좌 (현재 잔액: {self.balance}원)"

    # 디버거 창이나 객체 단독 조회 시 명확하게 상태를 보여줍니다.
    def __repr__(self):
        return f"BankAccount(owner='{self.owner}', balance={self.balance})"

    def deposit(self, amount):
        self.balance += amount      # 입금 처리
        print(f"[{self.owner} 계좌] {amount}원 입금 / 현재 잔액: {self.balance}원")
        return self.balance

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount  # 출금 처리
            print(f"[{self.owner} 계좌] {amount}원 출금 / 현재 잔액: {self.balance}원")
        else:
            print(f"[{self.owner} 계좌] 출금 실패: 잔액이 부족합니다. (잔액: {self.balance}원)")
        return self.balance
```

- 주피터 노트북에서 위 코드를 실행하는 순간, 파이썬 인터프리터는 BankAccount라는 클래스 구조를 메모리에 올려 인식할 수 있는 상태로 만듭니다.
- 이 단계에서는 아직 실제 잔액이 계산되거나 계좌 주인이 지정되지 않은 상태이며, 나중에 객체(인스턴스)를 생성할 때 사용할 기능들의 틀만 먼저 정의해 둔 것입니다.

<aside>
❓ **__str__의 역할**

---

print(객체)를 했을 때 사용자가 보기 편하도록 예쁘게 정돈된 요약 문장을 보여줍니다.

**__repr__의 역할**

---

주피터 노트북 셀에서 객체 이름을 그냥 입력하거나 디버거 창에서 변수를 볼 때, 개발자가 데이터 상태를 명확히 파악할 수 있도록 공식 명세를 보여줍니다.

</aside>

### 3.2 서로 다른 계좌 객체 만들고 사용하기

<aside>

**독립된 다중 객체 생성**

---

똑같은 클래스로부터 생성되었더라도 철수의 계좌와 영희의 계좌는 메모리 상에서 완전히 분리되어 작동합니다. 한쪽 계좌에서 입출금을 진행하더라도 다른 쪽 계좌에는 아무런 영향을 주지 않음을 직접 확인해 보겠습니다.

앞서 정의한 **클래스를 기반으로 2개의 독립된 계좌 인스턴스를 생성합니다.**

```python
account_chulsoo = BankAccount("김철수", 5000)
account_younghee = BankAccount("이영희", 10000)
```

**매직 메서드 적용으로 객체를 직접 출력해도 정돈된 문장이 출력됩니다.**

```python
print(account_chulsoo)   # __str__ 메서드가 실행되어 "김철수님의 계좌 (현재 잔액: 5000원)" 출력
account_younghee         # 주피터 노트북 셀에서 단독 조회 시 __repr__이 실행되어 명세 텍스트가 출력됨
```

**각 계좌 객체의 메서드를 개별적으로 실행하여 잔액 변화를 확인합니다.**

```python
account_chulsoo.deposit(2000)      # 철수 계좌에 2000원 입금
account_younghee.withdraw(3000)    # 영희 계좌에서 3000원 출금
```

**두 인스턴스가 서로 독립된 메모리 영역을 유지하고 있는지 비교하여 확인합니다.**

```python
print(f"철수 잔액 최종 확인: {account_chulsoo.balance}원")
print(f"영희 잔액 최종 확인: {account_younghee.balance}원")
print(f"메모리 주소 비교: {id(account_chulsoo)} vs {id(account_younghee)}")
```

- BankAccount("김철수", 5000)를 실행하는 순간, 메모리의 새로운 방에 예금주 "김철수"와 잔액 5000이라는 값이 채워진 독립된 객체가 만들어집니다.
- account_chulsoo.deposit(2000)을 호출하면 파이썬은 내부적으로 철수 계좌의 주소값을 찾아가서 그 방의 잔액만 바꾸어 줍니다. 따라서 영희 계좌의 잔액은 아무런 간섭을 받지 않고 그대로 유지됩니다.
</aside>

### 4.2 개념 요약 키워드

- **객체지향 설계 관점:** 프로그램을 순차적 명령어의 흐름으로만 보지 않고, 고유한 상태를 지닌 정형 데이터(속성)와 이를 다루는 행동(메서드)을 하나의 단위로 묶어 객체의 역할과 책임을 중심으로 구성하는 소프트웨어 설계 방식입니다.
- **클래스와 인스턴스:** 객체가 가져야 할 속성과 기능을 정의한 추상적인 틀(클래스)과, 이를 기반으로 실제 메모리 공간을 할당받아 생성된 개별 실체(인스턴스)의 관계입니다.
- **__init__ 초기화 함수:** 인스턴스가 생성 완료되는 시점에 자동 실행되어 객체에 필요한 속성 데이터들의 초깃값을 바인딩해 주는 기본 초기화 함수입니다.
- **self 키워드:** 여러 개의 다중 객체 인스턴스가 실행될 때, 지금 연산을 수행 중인 대상이 정확히 어떤 객체인지를 파이썬 인터프리터에 명시하여 주소를 추적 연결하는 자기 참조형 매개변수입니다.
- **속성과 인스턴스 메서드:** 객체가 가지는 고유의 상태 데이터인 '속성(변수)'과, self를 통해 해당 속성에 접근하여 데이터를 제어하는 '인스턴스 메서드(함수)'가 유기적으로 결합된 구조입니다.

### 4.3 용어 사전
| 용어 | 영어 표기 | 설명 |
|---|---|---|
| **객체 지향 프로그래밍** | Object-Oriented Programming (OOP) | 프로그램을 단순히 순서대로 실행되는 명령어들의 나열로 보지 않고, 데이터와 기능을 `객체`라는 독립된 단위로 묶어 이들의 조립과 상호작용으로 구현하는 개발 방식입니다. |
| **클래스** | Class | 객체를 컴퓨터 메모리에 만들어내기 위해 정의한 `설계도(틀)`입니다. 객체가 가져야 할 상태(데이터)와 행동(메서드)을 한곳에 정의합니다. |
| **인스턴스 (객체)** | Instance | 클래스라는 설계도를 기반으로 실제 메모리 공간에 생성된 객체를 뜻합니다. 같은 클래스로 여러 개의 서로 다른 인스턴스를 만들 수 있습니다. |
| **인스턴스화** | Instantiation | 클래스를 사용하여 실제 인스턴스 객체를 생성하는 과정을 뜻합니다. 예를 들어 `person = Person()`처럼 클래스를 호출하여 객체를 만드는 과정입니다. |
| **생성자** | Constructor (`__init__`) | 인스턴스가 생성될 때 초기 상태를 설정하는 데 사용하는 특수 메서드입니다. 주로 객체가 처음 가져야 할 속성을 만들고 초기값을 저장하는 역할을 합니다. |
| **self 키워드** | self | 클래스의 인스턴스 메서드 안에서 `현재 그 메서드를 사용하고 있는 인스턴스 자신`을 가리키는 이름입니다. 인스턴스 메서드의 첫 번째 매개변수로 관례적으로 `self`를 사용합니다. |
| **속성** | Attribute | 객체가 가지고 있는 데이터나 상태를 뜻합니다. 주로 `self.변수명` 형태로 값을 저장하며, 각 인스턴스가 서로 다른 값을 가질 수 있습니다. |
| **메서드** | Method | 클래스 내부에 정의되어 객체가 수행할 행동이나 기능을 나타내는 함수입니다. 인스턴스 메서드는 첫 번째 매개변수로 현재 인스턴스를 전달받으며 관례적으로 `self`라는 이름을 사용합니다. |

# 수업 내용

1. 객체(Object)란 무엇인가요?
- 컴퓨터 과학에서 객체(Object)란 "속성(데이터)과 행동(함수)을 함께 묶어놓은 하나의 덩어리"입니다.
- 현실 세계의 모든 사물을 생각해 보면 이해하기 쉽습니다.

2. 객체 지향 프로그래밍(OOP)이란?
- 프로그램을 만들 때 데이터(변수)와 기능(함수)을 무작정 늘어놓는 것이 아니라, 현실 세계의 '사물(객체)' 단위로 나누어서 상호작용하도록 프로그램을 설계하는 방식입니다.

- 절차 지향 프로그래밍: 순서대로 위에서부터 아래로 코드를 실행하는 방식 (요리가 일직선으로 진행되는 과정)
- 객체 지향 프로그래밍: 부품(객체)들을 먼저 만들고, 이 부품들을 조립해서 완성된 프로그램을 만드는 방식 (자동차 공장에서 타이어, 엔진, 핸들 부품을 만들어 조립하는 과정) 즉, 객체와 객체들을 만들어서 영향을 주고 받는(인터렉션) 프로그래밍 방식입니다. 
💡 객체 지향의 장점: 코드를 재사용하기 쉽고, 수정이 간편하며, 대규모 프로젝트를 협업해서 만들 때 매우 유리합니다.

3. 클래스(Class)와 객체의 관계
- 클래스(Class): 객체를 만들기 위한 '설계도' 또는 '틀'입니다. (예: 붕어빵 틀, 자동차 설계도)
- 인스턴스/객체(Instance/Object): 설계도를 바탕으로 실제로 만들어진 '제품'입니다. (예: 슈크림 붕어빵, 실제로 도로를 달리는 내 아반떼 자동차) 클래스를 가지고 동작할 수 있는 거

- 인스턴스(Instance): 클래스라는 설계도를 바탕으로 컴퓨터 메모리 위에 실제로 만들어진 실물(객체)

##### 객체 vs 인스턴스, 차이가 뭔가요?
- 객체 (Object): 포괄적인 개념입니다. 현실의 사물이나 개념을 프로그램으로 표현한 모든 것을 통틀어 부를 때 씁니다. ("파이썬의 모든 것은 객체야")

- 인스턴스 (Instance): "특정 클래스로부터 만들어졌다"는 관계에 초점을 맞춘 표현입니다.

##### 첫 class를 생성할 때 __init__과 self가 들어가는 이유?
- 처음 클래스를 만들 때 __init__과 self가 세트로 쏙 들어가는 이유는 "메모리 위에 객체를 만들 때, 초기 데이터(속성)를 어디에 어떻게 담을지 정해주는 핵심 장치"이기 때문

1. 왜 __init__이 들어갈까요? (생성자 역할)
비유하자면: 새 스마트폰 공장에서 폰을 갓 찍어냈을 때, 화면에 기본 배경화면을 깔고 초기 세팅을 싹 해주는 '공장 초기 세팅 로봇'입니다.

이유: 클래스라는 설계도로 실제 객체(실물)를 만들 때, "만들어지자마자 자동으로 이 세팅 함수를 실행해라!"라고 파이썬에게 명령하는 약속된 이름이기 때문입니다. 이걸 안 쓰면 객체가 만들어질 때 속성(데이터)을 자동으로 채워 넣을 길이 없습니다.

2. 왜 self가 들어갈까요? (객체 자신을 가리키는 이름표)
비유하자면: 여러 대의 로봇(객체)이 공장에서 동시에 만들어지고 있을 때, 각 로봇이 쥐고 있는 "이건 내 거야!"라고 외치는 내 몸뚱이 이름표입니다.

이유:

만약 bus1과 bus2라는 버스 두 대를 만들었다고 해봅시다. ride() 함수가 실행될 때, 파이썬은 "도대체 누구의 승객 수(count)를 1명 늘려줘야 하지?"를 알아야 합니다.

이때 self가 "지금 막 명령을 내린 바로 그 객체(예: bus1)"를 콕 집어 가리키는 역할을 합니다.

💡 한 줄 요약
__init__: 객체가 만들어질 때 자동으로 실행되면서 속성을 세팅하려고

self: 방금 만들어진 그 객체 자신의 변수와 함수에 접근하려고 (self.count는 "나(self)의 승객 수"라는 뜻)

이 두 가지는 파이썬에서 객체를 만들 때 뗄래야 뗄 수 없는 가장 강력하고 필수적인 짝꿍입니다!

````python
class Bus:
    def __init__(self):
        self.count = 0  # 승객 수
        self.passengers = []  # 승객 객체들을 담을 리스트 추가

    def ride(self, passenger):  # 승객을 매개변수로 받음
        self.count += 1
        self.passengers.append(passenger)  # 리스트에 승객 추가
        print(f"{passenger.name} 탑승")

    def off(self, passenger):
        if self.count > 0 and passenger in self.passengers:
            self.count -= 1
            self.passengers.remove(passenger)
            print(f"{passenger.name} 하차")

    def show_count(self):
        print("승객수:", self.count)

    def print_self(self):
        print(id(self))

    def __str__(self):
        return f"승객수: {self.count}, 승객: {self.passengers}"

class Passenger:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"이름: {self.name}({self.age}세)"

    def __repr__(self):
        return f"passenger(name='{self.name}, age={self.age})"

# --- 실행부 ---
bus = Bus()

p1 = Passenger("김철수", 20)
p2 = Passenger("김영희", 18)

# 올바른 메서드 호출 방식으로 수정
bus.ride(p1)
bus.ride(p2)

print(bus)

print(p1)
print(repr(p1))
````

````python
# 1. 클래스(설계도) 만들기
class Car:
    # 생성자 (속성을 초기화하는 함수)
    def __init__(self, color, brand):
        self.color = color  # 속성
        self.brand = brand  # 속성

    # 행동(메서드) 정의
    def horn(self):
        print(f"{self.brand} 자동차가 경주를 울립니다! 빵빵!")


# 2. 객체(실제 제품) 만들기
my_car = Car("빨간색", "현대")

# 3. 객체의 속성과 행동 사용하기
print(my_car.color)  # 출력: 빨간색
my_car.horn()  # 출력: 현대 자동차가 경주를 울립니다! 빵빵!
````

````python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance  # __를 붙여서 외부 접근을 숨김 (정보 은닉)

    # 잔액을 안전하게 조회하는 메서드
    def get_balance(self):
        return self.__balance

    # 잔액을 안전하게 입금하는 메서드
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount


# 사용하기
account = BankAccount(10000)
# account.__balance = 50000  # 오류 발생! 외부에서 직접 수정 불가

account.deposit(5000)  # 메서드를 통한 안전한 접근
print(account.get_balance())  # 출력: 15000
````
