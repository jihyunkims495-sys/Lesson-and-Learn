# 1. 상속과 메서드 오버라이딩

우리는 지난 시간에 클래스라는 설계도를 정의하고, 이를 통해 독립적인 객체인 **인스턴스(Instance)**를 생성하는 방법을 배웠습니다.

그런데 이미 검증된 설계도에 약간의 기능만 추가하거나 수정하고 싶을 때마다 새로운 클래스를 처음부터 다시 작성하는 것은 매우 비효율적입니다.

이번 장에서는 기존 클래스의 기능을 그대로 이어받아 확장하는 **상속(Inheritance)**과 물려받은 기능을 필요에 맞게 재정의하는 **오버라이딩(Overriding)**의 원리를 알아보겠습니다.

---

## 1.1 상속(Inheritance)

### 부모 클래스의 재사용

상속은 이미 만들어진 클래스인 **부모 클래스(Parent Class)**의 속성과 메서드를 그대로 물려받아 새로운 **자식 클래스(Child Class)**를 정의하는 기술입니다.

동일한 코드를 여러 클래스에 중복해서 작성할 필요가 없으므로 개발 시간이 단축되고 코드를 유지보수하기도 훨씬 수월해집니다.

```python
class Animal:
    def breathe(self):
        print("숨을 쉽니다.")


# Animal 클래스의 기능을 그대로 물려받습니다.
class Dog(Animal):
    def bark(self):
        print("멍멍 짖습니다.")


baduk = Dog()

baduk.breathe()  # 부모에게 물려받은 기능을 사용합니다.
baduk.bark()     # 자식 클래스만의 고유 기능을 사용합니다.
```

### 실행 결과

```text
숨을 쉽니다.
멍멍 짖습니다.
```

### 코드 구조

```python
class 자식클래스(부모클래스):
    자식 클래스의 코드
```

`Dog` 클래스에는 `breathe()` 메서드가 직접 작성되어 있지 않습니다.

하지만 `Dog` 클래스가 `Animal` 클래스를 상속받았기 때문에 `Animal` 클래스의 `breathe()` 메서드를 사용할 수 있습니다.

---

## 1.2 메서드 오버라이딩(Overriding)

### 물려받은 메서드의 재정의

부모 클래스로부터 물려받은 메서드를 자식 클래스의 요구사항에 맞게 새롭게 재정의하는 것을 **메서드 오버라이딩(Method Overriding)**이라고 합니다.

자식 객체에서 해당 메서드를 호출하면 부모 클래스의 원본 메서드 대신 자식 클래스에서 새롭게 정의한 메서드가 우선적으로 실행됩니다.

```python
class Animal:
    def sound(self):
        print("소리를 냅니다.")


class Cat(Animal):
    # 부모의 sound 메서드를 자식 고유의 행동으로 덮어씁니다.
    def sound(self):
        print("야옹 하고 웁니다.")


nabi = Cat()
nabi.sound()
```

### 실행 결과

```text
야옹 하고 웁니다.
```

`Cat` 클래스는 `Animal` 클래스의 `sound()` 메서드를 물려받았습니다.

그러나 `Cat` 클래스 내부에서 같은 이름의 `sound()` 메서드를 다시 정의했기 때문에 부모 클래스의 메서드가 아니라 자식 클래스의 메서드가 실행됩니다.

---

## 1.3 부모 클래스를 호출하는 `super()`

### 부모 클래스의 기능 확장하기

`super()`는 자식 클래스 내부에서 부모 클래스의 메서드나 초기화 함수인 `__init__()`을 호출할 때 사용합니다.

특히 자식 클래스만의 새로운 속성을 추가하면서 부모 클래스의 기존 초기화 로직도 그대로 재사용하고 싶을 때 사용합니다.

코드를 중복해서 작성하지 않고 부모 클래스의 자원을 안전하게 이어받는 연결고리 역할을 합니다.

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, major):
        # 부모 클래스의 생성자를 실행하여 name을 먼저 저장합니다.
        super().__init__(name)

        # Student 클래스만의 고유 속성을 추가합니다.
        self.major = major


student = Student("김철수", "컴퓨터공학")

print(student.name)
print(student.major)
```

### 실행 결과

```text
김철수
컴퓨터공학
```

### 실행 순서

1. `Student("김철수", "컴퓨터공학")`을 호출합니다.
2. `Student` 클래스의 `__init__()`이 실행됩니다.
3. `super().__init__(name)`이 부모 클래스인 `Person`의 `__init__()`을 호출합니다.
4. 부모 클래스에서 `self.name`을 저장합니다.
5. 다시 자식 클래스의 `__init__()`으로 돌아옵니다.
6. 자식 클래스에서 `self.major`를 저장합니다.

# 2. 캡슐화와 정보 은닉, Getter와 Setter

상속을 통해 프로그램의 구조를 유연하게 확장했다면 이제는 객체 내부의 핵심 데이터를 안전하게 보호해야 합니다.

객체지향 프로그래밍에서는 데이터와 데이터를 다루는 기능을 하나의 단위로 묶는 **캡슐화(Encapsulation)**와 불필요한 내부 구현을 외부에 노출하지 않는 **정보 은닉(Information Hiding)**을 중요하게 다룹니다.

파이썬에서 데이터를 은닉하고 이를 제어할 수 있는 올바른 접근 통로를 설계하는 방법을 알아보겠습니다.

---

## 2.1 정보 은닉과 캡슐화(Encapsulation)

### 데이터와 기능의 보호

캡슐화는 서로 관련된 데이터와 기능을 클래스라는 하나의 캡슐 안에 모으는 것을 의미합니다.

정보 은닉은 외부에서 객체 내부의 민감한 데이터를 함부로 확인하거나 수정하지 못하도록 제한하는 개념입니다.

물건의 복잡하거나 위험한 내부 구조는 감추고 겉면에 안전한 버튼만 노출하는 것과 비슷합니다.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age  # 외부에서 언제든지 수정할 수 있습니다.


person = Person("김철수", 20)

print(person.age)

# 외부에서 나이를 비정상적인 값으로 직접 변경할 수 있습니다.
person.age = -100

print(person.age)
```

### 실행 결과

```text
20
-100
```

현재 `age` 속성은 외부에서 직접 수정할 수 있기 때문에 잘못된 값이 저장될 위험이 있습니다.

---

## 2.2 파이썬 프라이빗 변수와 이름 변형 규칙

### 언더바 두 개(`__`)를 통한 접근 제한

파이썬에는 외부 접근을 물리적으로 완전히 차단하는 별도의 `private` 키워드가 없습니다.

대신 변수 이름 앞에 언더바 두 개(`__`)를 붙이면 외부에서 원래 이름으로 직접 접근하기 어렵도록 이름이 변경되는 **이름 변형 규칙(Name Mangling)**이 적용됩니다.

이를 통해 다음과 같은 문제를 줄일 수 있습니다.

- 외부에서 원래 이름으로 실수로 접근하는 문제
- 외부에서 속성을 함부로 수정하는 문제
- 상속 관계에서 동일한 속성 이름이 충돌하는 문제

다만 이름 변형은 접근 자체를 보안 수준으로 완전히 차단하는 기능은 아닙니다.

```python
class Secret:
    def __init__(self):
        self.__password = "1234"


my_secret = Secret()

# 원래 속성 이름으로 접근하면 AttributeError가 발생합니다.
# print(my_secret.__password)

# 내부적으로 변형된 이름을 사용하면 접근할 수 있습니다.
print(my_secret._Secret__password)
```

### 실행 결과

```text
1234
```

`self.__password`는 내부적으로 다음과 같은 이름으로 변형됩니다.

```python
self._Secret__password
```

이름이 변형되는 기본 구조는 다음과 같습니다.

```text
_클래스이름__속성이름
```

따라서 `__`를 사용했다고 해서 데이터에 절대 접근할 수 없는 것은 아닙니다.

## 2.3 Getter와 Setter 메서드

### 안전한 접근 통로

객체 내부의 비공개 속성인 **프라이빗 속성(Private Attribute)**을 안전하게 조회하고 수정할 수 있도록 중간에서 관리하는 메서드를 **Getter**와 **Setter**라고 합니다.

- Getter: 숨겨진 값을 외부에 반환합니다.
- Setter: 전달받은 값이 올바른지 검사한 후 값을 변경합니다.

데이터를 아무런 제약 없이 반환하거나 변경하는 것이 아니라 값의 유효성을 검증한 후 처리하는 안전장치 역할을 합니다.

```python
class Wallet:
    def __init__(self):
        self.__money = 1000

    # Getter: 숨겨진 값을 반환합니다.
    def get_money(self):
        return self.__money

    # Setter: 전달받은 값을 검사한 후 변경합니다.
    def set_money(self, amount):
        if amount >= 0:
            self.__money = amount
        else:
            print("경고: 지갑의 돈은 음수가 될 수 없습니다.")
```

### Getter 사용하기

```python
wallet = Wallet()

money = wallet.get_money()
print(money)
```

### 실행 결과

```text
1000
```

### Setter 사용하기

```python
wallet.set_money(5000)
print(wallet.get_money())
```

### 실행 결과

```text
5000
```

### 잘못된 값 전달하기

```python
wallet.set_money(-1000)
print(wallet.get_money())
```

### 실행 결과

```text
경고: 지갑의 돈은 음수가 될 수 없습니다.
5000
```

Setter가 음수 값을 차단했기 때문에 기존 값인 `5000`이 유지됩니다.

# 3. MinusAccount 클래스 구현과 정보 은닉 적용하기

앞에서 배운 객체지향 프로그래밍의 상속과 정보 은닉 개념을 활용해 실제 코드를 작성해 보겠습니다.

이전 시간에 구현한 `BankAccount` 클래스를 상속받아 잔액이 마이너스가 되어도 출금할 수 있는 `MinusAccount` 클래스를 정의합니다.

이어서 잔액 속성을 프라이빗 속성으로 은닉하고 검증된 접근 통로를 통해서만 제어하는 구조를 확인하겠습니다.

---

## 3.1 BankAccount 클래스 상속 및 출금 로직 오버라이딩

### 클래스 정의

부모 클래스의 초기화 로직인 예금주와 잔액 설정을 그대로 이어받으면서 출금 기능은 마이너스 한도까지 허용하도록 재정의합니다.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return f"{self.owner}님의 계좌 (현재 잔액: {self.balance}원)"


# BankAccount를 상속받는 MinusAccount 자식 클래스를 정의합니다.
class MinusAccount(BankAccount):
    def __init__(self, owner, balance, limit):
        # 부모 클래스의 생성자를 호출합니다.
        super().__init__(owner, balance)

        # 자식 클래스만의 마이너스 한도 속성을 추가합니다.
        self.limit = limit

    # 출금 기능을 마이너스 통장 규칙에 맞게 오버라이딩합니다.
    def withdraw(self, amount):
        # 현재 잔액과 마이너스 한도를 합친 범위까지 출금할 수 있습니다.
        if (self.balance + self.limit) >= amount:
            self.balance -= amount

            print(
                f"[{self.owner} 마이너스 계좌] "
                f"{amount}원 출금 완료 / 현재 잔액: {self.balance}원"
            )
        else:
            print(
                f"[{self.owner} 마이너스 계좌] "
                f"출금 실패: 마이너스 한도({self.limit}원)를 초과했습니다."
            )

        return self.balance
```

### 인스턴스 생성하기

```python
account = MinusAccount("홍길동", 5000, 10000)

print(account)
print(f"마이너스 한도: {account.limit}원")
```

### 실행 결과

```text
홍길동님의 계좌 (현재 잔액: 5000원)
마이너스 한도: 10000원
```

`MinusAccount` 인스턴스를 생성하면 `super().__init__()`을 통해 부모 클래스의 초기화 로직이 먼저 실행됩니다.

부모 클래스에서 다음 속성이 저장됩니다.

```python
self.owner = owner
self.balance = balance
```

이후 자식 클래스에서 다음 속성이 추가됩니다.

```python
self.limit = limit
```

### 마이너스 한도 안에서 출금하기

```python
account.withdraw(12000)
```

### 실행 결과

```text
[홍길동 마이너스 계좌] 12000원 출금 완료 / 현재 잔액: -7000원
```

현재 잔액은 `5000원`이고 마이너스 한도는 `10000원`입니다.

따라서 최대 `15000원`까지 출금할 수 있습니다.

```python
현재 출금 가능 금액 = 현재 잔액 + 마이너스 한도
```

```text
5000 + 10000 = 15000원
```

### 마이너스 한도를 초과하여 출금하기

```python
account.withdraw(9000)
```

### 실행 결과

```text
[홍길동 마이너스 계좌] 출금 실패: 마이너스 한도(10000원)를 초과했습니다.
```

첫 번째 출금 후 잔액은 `-7000원`입니다.

따라서 추가로 출금할 수 있는 금액은 `3000원`입니다.

```text
-7000 + 10000 = 3000원
```

`9000원`은 출금 가능 금액인 `3000원`보다 크기 때문에 출금이 거절됩니다.

## 3.2 `__balance` 프라이빗 변수와 Getter/Setter 설계

### 데이터 직접 접근 제한 및 유효성 검증

잔액 속성을 외부에서 직접 변경할 수 없도록 프라이빗 속성으로 전환하고 올바른 조건에서만 데이터 수정을 허용하는 Setter를 구현합니다.

```python
class ProtectedAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    # Getter: 숨겨진 잔액을 반환합니다.
    def get_balance(self):
        return self.__balance

    # Setter: 전달받은 값을 검사한 후 잔액을 변경합니다.
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
            print(f"[시스템] 잔액이 {amount}원으로 정상 변경되었습니다.")
        else:
            print(
                "[시스템 경고] 잘못된 수치 입력: "
                "잔액은 음수가 될 수 없습니다."
            )
```

### 직접 수정과 Setter를 통한 수정 비교

```python
account = ProtectedAccount("이몽룡", 3000)

print(f"최초 잔액: {account.get_balance()}원")
```

### 실행 결과

```text
최초 잔액: 3000원
```

### 외부에서 직접 수정 시도하기

```python
account.__balance = -5000

print(f"직접 수정 시도 후 조회 결과: {account.get_balance()}원")
```

### 실행 결과

```text
직접 수정 시도 후 조회 결과: 3000원
```

다음 코드는 에러가 발생하지 않습니다.

```python
account.__balance = -5000
```

하지만 기존의 프라이빗 속성인 `__balance`를 변경한 것도 아닙니다.

객체 외부에서 새로운 `__balance` 속성을 추가했을 뿐입니다.

객체 내부에는 다음 두 속성이 별도로 존재하게 됩니다.

```text
_ProtectedAccount__balance = 3000
__balance = -5000
```

따라서 Getter가 반환하는 실제 잔액은 여전히 `3000원`입니다.

### Setter를 통해 잘못된 값 전달하기

```python
account.set_balance(-1000)
```

### 실행 결과

```text
[시스템 경고] 잘못된 수치 입력: 잔액은 음수가 될 수 없습니다.
```

### Setter를 통해 올바른 값 전달하기

```python
account.set_balance(5000)

print(f"최종 안전 조회 결과: {account.get_balance()}원")
```

### 실행 결과

```text
[시스템] 잔액이 5000원으로 정상 변경되었습니다.
최종 안전 조회 결과: 5000원
```

Setter를 사용하면 전달받은 값이 올바른지 검사한 후에만 실제 잔액을 변경할 수 있습니다.

## 4.2 개념 요약 키워드

### 클래스 상속

부모 클래스의 속성과 메서드를 그대로 물려받아 공통 코드를 중복으로 작성하지 않고 새로운 자식 클래스를 정의하는 확장 방식입니다.

### 메서드 오버라이딩

상속 관계에서 부모 클래스의 메서드 대신 자식 클래스 고유의 기능이 실행되도록 같은 이름의 메서드를 자식 클래스에 새롭게 정의하는 방식입니다.

### `super()`

자식 클래스 내부에서 부모 클래스의 생성자 또는 메서드를 호출할 때 사용하는 함수입니다.

### 캡슐화와 정보 은닉

연관된 데이터와 기능을 클래스라는 하나의 단위로 묶고 외부의 무분별한 직접 접근을 제한하여 내부 데이터를 보호하는 설계 원칙입니다.

### 이름 변형 규칙(Name Mangling)

속성 이름 앞에 언더바 두 개(`__`)를 붙였을 때 파이썬이 속성 이름 앞에 클래스 이름을 결합하여 내부 이름을 변경하는 기능입니다.

### Getter와 Setter

은닉된 객체 내부 데이터에 접근할 수 있도록 공개된 전용 접근 통로입니다.

- Getter: 값을 조회하고 반환합니다.
- Setter: 값의 유효성을 검사한 후 수정합니다.

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def __str__(self):
        return (
            f"{self.owner}님의 계좌 "
            f"(현재 잔액: {self.__balance}원)"
        )

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
            print(f"[시스템] 잔액이 {amount}원으로 변경되었습니다.")
        else:
            print("[시스템 경고] 잔액은 음수가 될 수 없습니다.")

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(
                f"[{self.owner} 계좌] "
                f"{amount}원 입금 완료 / "
                f"현재 잔액: {self.__balance}원"
            )
        else:
            print("[입금 실패] 입금액은 0원보다 커야 합니다.")

        return self.__balance


class MinusAccount(BankAccount):
    def __init__(self, owner, balance, limit):
        super().__init__(owner, balance)
        self.limit = limit

    def withdraw(self, amount):
        current_balance = self.get_balance()

        if amount <= 0:
            print("[출금 실패] 출금액은 0원보다 커야 합니다.")
        elif current_balance + self.limit >= amount:
            new_balance = current_balance - amount

            # 마이너스 통장은 음수 잔액을 허용하므로
            # 부모 클래스의 set_balance() 대신 별도의 처리가 필요합니다.
            self._BankAccount__balance = new_balance

            print(
                f"[{self.owner} 마이너스 계좌] "
                f"{amount}원 출금 완료 / "
                f"현재 잔액: {new_balance}원"
            )
        else:
            print(
                f"[{self.owner} 마이너스 계좌] "
                f"출금 실패: 마이너스 한도 "
                f"({self.limit}원)를 초과했습니다."
            )

        return self.get_balance()

2. 메서드 오버라이딩 (Overriding)
- 메서드를 재정의
- 부모쪽 클래스와 동일한 이름으로 정의하여 기능을 수정하는 경우 or 기능을 재정의
- 자식쪽 클래스에 정의된 동일한 이름의 함수가 호출 우선순위를 가진다.

````python
class Animal:

    def __init__(self,kind):   # 매직 메서드
        self.kind = kind # 동물 종류 

    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") # 부모쪽 객체(인스턴스 객체)/ 인스턴스 변수인 kind에 초기화


    def move(self):
        print("강아지 걸음으로 움직인다.")


class Cat(Animal): 
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print("고양이 걸음으로 움직인다.")
````

````python
dog=Dog()
cat=Cat()
dog.move()
cat.move()
````

````python
# 동물 종류 받아오기
# 자식쪽에서 부모쪽에 동일한 이름의 메서드를 호출해야하는 경우는 self로 접근할 수 없음
## super().부모쪽 메서드(...)
class Animal:

    def __init__(self,kind):   # 매직 메서드
        self.kind = kind # 동물 종류 

    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") # 부모쪽 객체(인스턴스 객체)/ 인스턴스 변수인 kind에 초기화


    def move(self):
        super.move() # 부모쪽 객체의 move를 호출 
        print(self.kind +"걸음으로 움직인다.")


class Cat(Animal): 
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print(self.kind +"걸음으로 움직인다.")
````

````python
dog=Dog()
cat=Cat()
dog.move()
cat.move()
````

## [중요!] 파이썬은 다중 상속이 가능!
---
- 여러 부모 클래스를 상속 관계로 정의가 가능하다!
- 다중 상속 시, 같은 메서드가 있을 때 호출 우선순위가 있다.(\_\_mro\_\_)
- 자식 클래스의 메서드 -> 상속 관계에서 왼쪽에 있는 클래스

````python
class ParentA:
    def a(self):
        print("ParentA - a()")
    def b(self):
        print("ParentA - b()")

class ParentB:
    def a(self):
        print("ParentB - a()")
    def c(self):
        print("ParentB - c()")

class Child(ParentA,ParentB): # 상속관계는 ()소괄호로 명시하며, 형태는 옆에 참고!
    pass
````

4. 정보 은닉과 캡슐화(Encapsulation)
- "캡슐화는 데이터와 기능을 묶고, 필요한 부분만 외부에 공개하는 설계 방식"
- 캡슐화(Encapsulation)는 클래스 안에 다음 두 가지를 하나로 묶어 관리하는 것
    - 데이터: 객체가 가진 정보인 속성
    - 기능: 데이터를 다루는 메서드
- 객체 내부의 중요한 데이터를 외부에서 함부로 변경하지 못하게 보호하는 것을 정보 은닉

- 왜 필요한가?


````python

class schedule:
    def __init__(self,year,month,day):
        self.year = year
        self.month = month
        self.day = day

    def __str__(self):
        return f"year={self.year},month={self.month},day{self.day}"
````

````python
s1 = schedule(2026,8,18)

print(s1)
````

````python
# 인스턴스 변수에 직접 대입이 가능하므로 값에 대한 통제는 불가!
# 통제 불가능하면 -> 직접 접근을 막아야 한다! -> 인스턴스 변수를 외부에서 접근하지 못하게 막는다.
s1.year = 2026
s1.month = 2
s1.day = 31

print(s1)
````

5. 파이썬 프라이빗(Private) 변수와 이름 변형 규칙(NameMangling)
- \_\_변수명 : 클래스 내부에서만 접근 가능한 인스턴스 변수가 된다.

````python
class Schedule : 
    def __init__(self,year,month,day):
        self.__year = year ### 프라이빗 변수로 선언
        self.__month = month
        self.__day = day

    def __str__(self):
        return f"year={self.__year},month={self.__month},day{self.__day}"
````

````python
s1 = Schedule(2026,8,18)
print(s1)
````

````python
s1.__year = 2026
s1.__month = 8
s1.__day = 18
print(s1)
````

6. Getter와 Setter 메서드
- Getter와 Setter는 클래스 안에 숨겨둔 속성을 외부에서 조회하거나 변경하기 위한 메서드
- Getter: 속성값을 가져오는 메서드 / get: 가져오다
- Setter: 속성값을 설정하거나 변경하는 메서드 / set: 설정하다

````python
class Schedule : 
    def __init__(self,year,month,day):
        self.__year = year ### 프라이빗 변수로 선언
        self.__month = month
        self.__day = day


#get 반복적으로 사용 예시
    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month

    def get_day(self):
        return self.__day


#set 반복적으로 사용 예시
    def set_year(self,year):
        self.__year = year

    def set_month(self,month):
            self.__month = month

    def set_day(self,day):
        if self.__month == 2 and day > 28 :  # 값에 대한 통제가 가능
            day = 28
    

    def __str__(self):
        return f"year={self.__year},month={self.__month},day{self.__day}"
````

````python
s1 = Schedule(2026,8,18)

s1.set_month(2)
s1.set_day(31)

print(s1)
````
