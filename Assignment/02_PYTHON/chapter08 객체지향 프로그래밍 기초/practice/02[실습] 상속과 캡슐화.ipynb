### 필수 1 : 클래스 상속과 스마트폰 기능의 확장
---
#### 문제 1-1 :  기존 휴대폰을 상속받은 스마트폰 클래스 구현하기

```python
class phone :  #부모 클래스
    def call(self):
        print("통화를 시작합니다.")

class SmartPhone(phone): #자식 클래스
    def call(self):
            print("영상 통화를 시작합니다.") #오버라이딩

    def search_web(self):
        print("인터넷 검색을 시작합니다.") #자식의 새 기능

smartphone = SmartPhone()
smartphone.call()
smartphone.search_web()

```

### 필수 2 : 정보 은닉과 안전한 스마트홈 도어락
---
#### 문제 2-1 :  프라이빗 변수와 Getter 및 Setter를 활용한 도어락 시스템

```python
class DoorLock:
    def __init__(self, password):
        self.__password = password

    def get_password(self):
        return self.__password

    def set_password(self, new_password):
        if len(new_password) >= 4:
            self.__password = new_password
            print("정상 변경되었습니다")
        else:
            print("경고: 비밀번호는 4자리 이상이어야 합니다")


door = DoorLock("1234")

print(door.get_password())

door.set_password("12")
print(door.get_password())

door.set_password("5678")
print(door.get_password())
```

```python
# 강사님 풀이
class DoorLock:
    def __init__(self, password):
        if not self.validate(password):
            return

        self.__password = password

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if not self.validate(password):
            return

        self.__password = password

    def validate(self, password):
        result = len(password) >= 4

        if not result:
            print("경고: 비밀번호는 4자리 이상이어야 합니다.")

        return result


door = DoorLock("0000")

door.__password = "abcdef"
print(door.get_password())

door._DoorLock__password = "abcdef"
print(door.get_password())
```

### 심화 1 : 부모 클래스를 호출하는 super()와 회원 관리 확장
---
#### 문제 3-1 : super()를 활용한 우수 회원 등급 클래스 설계하기

```python
class User:
    def __init__(self, name):
        self.name = name


class VIPUser(User):
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point


vip = VIPUser("지현", 1000)

print(vip.name)
print(vip.point)
```

```python
# 강사님 풀이
class User:
    def __init__(self, name):
        self.name = name


class VIPUser(User):
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point


user = VIPUser("김철수", 1000)

print(user.name, user.point)
```
