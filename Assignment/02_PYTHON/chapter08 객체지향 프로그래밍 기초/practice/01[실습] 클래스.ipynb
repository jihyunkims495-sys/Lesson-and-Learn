### 필수 1 : 도서 정보 관리 클래스 설계
---
#### 문제 1-1 : 도서 정보 출력 프로그램

```python
class Book:
    def __init__(self,title,author):
        self.title = title
        self.author = author

    def info(self):
        print(f"책 제목:{self.title}, 저자:{self.author}")



book1 = Book("The Ant","Bernard Werber")
book2 = Book("Demian", "Hermann Hesse")
book1.info()
book2.info()
```

### 필수 2 : 스마트 조명 제어 클래스 설계
---
#### 문제 2-1 :  스마트 조명 전원 스위치 구현

```python
class SmartLamp:
    def __init__(self,model):
        self.model = model
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print(f"[{self.model}명] 조명이 켜졌습니다.")

    def turn_off(self):
        self.is_on = False
        print(f"[{self.model}명] 조명이 꺼졌습니다.")

smartlamp = SmartLamp("000-0000")
smartlamp.turn_on()
smartlamp.turn_off()
```

### 심화 1 : 게임 캐릭터 전투 및 체력 관리 시스템 설계
---
#### 문제 3-1 : 영웅 캐릭터의 피해량 처리 기능

```python
class Hero:
    def  __init__(self,name):
        self.name = name
        self.hp = 100

    def take_damage(self,damage):
        self.hp = self.hp - damage
        print(f"{self.name}의 남은 체력:{self.hp}")

        if self.hp <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌다.")

hero = Hero("전사")

hero.take_damage(30)
hero.take_damage(40)
hero.take_damage(50)
```

```python
# 강사님 풀이

class Hero:
    def __init__(self, name):
        self.name = name 
        self.hp = 100

    def take_damage(self,damage):
        minus_damage = damage if self.hp >= damage else self.hp # 3항 조건식
        self.hp -= minus_damage
        print("남은 체력:",self.hp)

        if self.hp <= 0 :
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")

hero = Hero("피카츄")
hero.take_damage(90)
hero.take_damage(20)
```
