### 필수 1 : 클로저를 활용한 나만의 비밀 주머니 만들기
---
#### 문제 1-1 :  텍스트 메시지 저장소 구현하기

```python
def message_store():
    storage = ""

    def add_message(message):
        nonlocal storage
        storage += message + "\n"
        return storage

    return add_message


my_storage = message_store()

print(my_storage("새로운 알림이 도착했습니다"))
print(my_storage("출석 체크가 완료되었습니다"))
```

### 필수 2 : 데코레이터를 활용한 환영 인사말 꾸미기
---
#### 문제 2-1 :  데코레이터를 활용한 알림 테두리 장식하기

```python
def deco_border(func):
    def wrapper():
        print("==========")
        func()
        print("==========")

    return wrapper


@deco_border
def show_welcome():
    print("환영합니다 스파르타님")


show_welcome()
```

### 심화 1 : 가변 인자를 활용한 자동 로깅 시스템 구축
---
#### 문제 3-1 : 함수 호출 정보 가동 상태 추적하기

```python
def auto_tracker(func):
    def wrapper(*args, **kwargs):
        print(f"{func.__name__} 함수 실행 시작")
        print("args:", args)
        print("kwargs:", kwargs)

        result = func(*args, **kwargs)

        print(f"{func.__name__} 함수 실행 종료")

        return result

    return wrapper


@auto_tracker
def add_numbers(a, b):
    return a + b


result = add_numbers(10, 20)
print("최종 결과:", result)
```
