## 필수 2 : 온라인 쇼핑몰 결제 금액 및 상태 반환 시스템 구현
---
### 문제 2-1 :  배송비 자동 계산 및 다중 데이터 반환

```python
def  calculate_total( price, delivery_fee = 3000 ) :
    total =  price  + delivery_fee
    return total, "결제가 완료되었습니다."

price = 25000
print(calculate_total(price))
```

```python
a,b = calculate_total(price)  # 변수 a,b를 선언해서 언패킹
print(a,b)
```

```python
a = calculate_total(price) # 변수 a를 선언해서 튜플 상태로 그대로 출력
print(a)
```
