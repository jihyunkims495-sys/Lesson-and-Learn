### 필수 1 : 전역 변수와 지역 변수의 유효 범위와 은닉 법칙
---
#### 문제 1-1 :  게임 머니 변수 격리를 통한 은닉 법칙 검증

```python
game_money = 7000 # 전역 변수


def item_shop():
    game_money = 300 # 지역 변수
    print(game_money)

item_shop()  # 요게 은닉! 지역변수가 함수 안에서 동작해서, 파이썬이 먼저 출력함!

print(game_money)
```

### 필수 2 : global 키워드를 활용한 전역 데이터 수정
---
#### 문제 2-1 :  전역 점수 누적 시스템 구현

```python
total_score = 0 #전역 변수

def add_score():
    global total_score  #global 키워드 사용해서 전역변수를 함수 안에 끌어오기!
    total_score += 10 

add_score()
print(total_score)
add_score()
print(total_score)
```

```python
total_score = 0 #전역 변수

def add_score():
    global total_score  
    total_score += 10 
    print(total_score)

add_score()
```

### 심화 1 : 람다 함수와 내장 함수의 연동을 통한 데이터 일괄 가공
---
#### 문제 3-1 : 정수 리스트 일괄 곱셈 변환 및 기준 값 초과 필터링

```python
base_numbers = [1,2,3,4,5,6]

multiplied_numbers = list(map(lambda x:x*3,base_numbers))

print(multiplied_numbers)

final_filtered_numbers = list(filter(lambda x: x > 10, multiplied_numbers))

print(final_filtered_numbers)
```

```python
counter = 0 

def increase(): 
    global counter 
    counter += 1 


def local_increase(): 
    counter = 100 
    increase() 
    increase() 
    local_increase() 


print(counter)
```
