### 1. 전역 변수(Global)와 지역 변수(Local)
---
- 전역 변수(Global) : 함수 외부에 정의되는 변수로, 코드 전역에서 접근 가능
    - 규모가 큰 프로젝트에서는 어떤 함수가 이 값을 바꿨는지 추적하기 힘든 단점
- 지역 변수(Local) : 함수 내부에 정의되는 변수로, 함수가 실행될 때만 유효한 변수
    - 일반적으로 함수 실행이 끝나면 해당 호출의 지역 변수에 직접 접근할 수 없습니다. 
    - 다만 클로저처럼 외부에서 계속 참조되는 값은 유지될 수 있습니다.

- 전역 변수 오염 방지 원리
    - 파이썬은 함수 내부에서 외부 전역 변수와 똑같은 이름으로 변수를 선언하더라도, 전역 변수의 값을 덮어쓰지 않고 새로운 독립된 지역 변수를 생성하는 구조를 가집니다. 이를 통해 함수 외부의 중요한 정형 데이터가 예기치 않게 변경되어 오염되는 상황을 완벽히 식별하여 차단합니다.

```python
# 예시 - Nameerror가 뜹니다! 왜? / b변수가 '지역변수'로 함수안에 있기 때문에, 
a = 10 #전역 변수

def func() : 
    b = 20 #지역변수

print(a,b)
```

```python
# 예시 - 전역 변수 출력
a = 10 #전역 변수

def func() : 
    b = 20 #지역변수

print(a)
```

```python
# 예시 - 지역 변수 출력
a = 10 #전역 변수

def func() : 
    b = 20 #지역변수
    print(b)

func()
```

```python
# 예시 
a = 10 #전역 변수

def func() : 
    b = 20 #지역변수
    print(f'{a=},{b=}') #{a=}는 변수 이름과 값을 같이 보여주는 f-string 문법

func()
```

```python
# 파이썬은 함수안에 함수도 정의 할 수 있어요~
# 함수가 중첩이 되어 있을 때, 안쪽이 먼저 실행 된 후에 바깥 함수가 실행 됩니다~
def outer():
    a = 10

    
    def inner():
        b = 20
        print(f'{a=},{b=}')
    inner()

outer()
```

### 2. 파이썬의 변수 검색 규칙 : LEGB  스코프 
---
- LEGB는 파이썬이 변수 이름을 찾는 순서
- 파이썬이 name 같은 변수명을 만나면 무작정 전체 코드를 찾지 않고, 가까운 곳부터 다음 순서로 찾는다.
- L → E → G → B


| 순서 | 영어 | 변수 이름 | 확인하는 곳 |
| --- | --- | --- | --- |
| 1 | Local | 지역 변수 | 현재 실행 중인 함수 안 |
| 2 | Enclosing | 외부 함수 변수 | 현재 함수를 감싸고 있는 바깥 함수 |
| 3 | Global | 전역 변수 | 모든 함수 바깥의 전역 공간 |
| 4 | Built-in | 내장 변수 | 파이썬이 기본으로 제공하는 이름 |

- 안쪽부터 바깥쪽으로 찾는 탐색 규칙
    - 파이썬 인터프리터는 코드에서 특정 변수명을 마주했을 때, 현재 속한 가장 안쪽 영역부터 단계적으로 바깥 영역을 순차적으로 스캔하며 변수의 존재를 식별합니다.
- LEGB 네 가지 계층 구조
    - 변수를 찾아내는 우선순위는 다음의 4단계 범위를 따르며, 조건 만족 여부에 따라 첫 번째로 매핑되는 변수의 값을 도출합니다.
        1. **L (Local):** 현재 작동 중인 함수 내부 공간 (가장 높은 우선순위)
        2. **E (Enclosing):** 중첩 함수 구조에서 내부 함수를 감싸고 있는 외부 함수의 공간
        3. **G (Global):** 함수 바깥쪽이자 프로그램 소스코드 파일 전체의 전역 공간
        4. **B (Built-in):** 파이썬이 태생적으로 내장하고 있는 print(), len() 같은 기본 함수 공간 (가장 낮은 우선순위)

```python
# 예시 1
## 동일한 'a' 변수를 정의 했을 때, 
a = 'global'

def outer():
    a = 'enclosing'
    print("outer:",a)
    def inner():
        a = 'local'
        print("inner:",a)

    inner()

outer()

```

```python
# 예시 2
a = 'global'

def outer():
    a = 'enclosing'
    print("outer:",a)
    def inner():
        # a = 'local' <<예시 1에서 inner()의 변수 a를 없앤다면, 
        print("inner:",a)

    inner()

outer()
```

### 3. Global 키워드
---

```python
# 예시 3 : 만약 전역 변수를 사용하고 싶다면, 
a = 'global'

def outer():
    a = 'enclosing'
    print("outer:",a)
    def inner():
        global a    # a라는 변수는 전역변수로 찾겠다!
        print("inner:",a)

    inner()

outer()
```

```python
# 예시 4 : global 키워드의 부작용 / 전역 변수의 오염
a = True

def outer():
    a = 'enclosing'
    print("outer:",a)
    def inner():
        global a        # a라는 변수는 전역변수로 찾겠다!
        a = not a       
        print("inner:",a)

    inner()

outer()
```

-  global 키워드의 작동 문법과 스코프 꼬임 현상 예방법
    - 전역 변수 수정 통로 개방
        - 원래 파이썬 함수 "내부"에서는 전역 변수의 값을 읽을 수만 있고 수정하는 명령은 차단됩니다. 하지만 함수 내에서 global 변수명을 명시적으로 선언하면, 인터프리터에게 "지금부터 수정할 변수는 전역 공간에 있는 저 변수다"라고 식별 신호를 주어 값을 바꿀 수 있게 허용합니다.
    - 스코프 꼬임 현상 예방 가이드
        - global 키워드를 무분별하게 남용하면 여러 함수가 하나의 전역 변수를 두고 서로 수치를 바꾸는 난잡한 구조가 됩니다. 이는 코드 실행 순서에 따라 결과가 계속 바뀌어 버그 진단이 불가능해지는 '스코프 꼬임 현상'을 유발하므로,  함수의 인수를 통해 데이터를 입력받고 return으로 결과값을 돌려받는 독립적인 흐름 설계를 취해야 합니다.

### 4. 일급 함수(First-class Function)
---
- 파이썬에서 함수도 숫자나 문자열처럼 하나의 값으로 취급할 수 있다는 뜻. 즉, 파이썬에서는 함수를 다음과 같이 사용할 수 있다
- "변수 = 함수"
    - 변수에 저장할 수 있다.
    - 다른 함수의 인수로 전달할 수 있다.
    - 함수의 반환값으로 돌려줄 수 있다.
    - 리스트나 딕셔너리 등에 저장할 수 있다.
    - [중요!] 함수의 매개변수로 함수를 사용할 수 있고, 반환값으로도 함수를 사용할 수 있습니다!!!!!
    
| 코드 | 의미 |
| --- | --- |
| greeting | 함수 자체 |
|  greeting()  | 함수를 지금 실행한 결과 |

```python
# 일급 함수 예시!
def greeting():
    print("안녕하세요")


hello = greeting
hello()
```

```python
# 콜백 함수 : 함수를 매개변수로 사용

def callback2():
    print("callback()")

def func(callback):
    print("func()")
    callback()

func(callback2)
```

```python
# 팩토리 함수 : 함수를 반환값으로 사용
## 
def outer(num1):
    def inner(num2):
        return num1 + num2

    return inner 
```

```python
a = outer(10) # 항상 10을 더해주는 새로운 함수
a
```

```python
a(20) # 10+20
```

```python
b = outer(20) # 항상 20을 더해주는 새로운 함수
b(20)  # 20+20
```

```python
b(30)  #20+30
```

#### [별첨.] 팩토리 함수 / 콜백 함수 / 클로저
- [중요!] 함수의 매개변수로 함수를 사용할 수 있고, 반환값으로도 함수를 사용할 수 있습니다!!!!!
    - 함수를 매개변수로 사용: 콜백 함수
    - 함수를 반환값으로 사용: 팩토리 함수, 클로저(Closure)
### 클로저
- 반환된 안쪽 함수가 바깥 함수의 값을 계속 기억하면 -> "클로저"라고 한다. 
- 바깥 함수의 실행이 끝난 뒤에도, 안쪽 함수가 바깥 함수의 값을 기억하고 사용하는 구조다.

#### 팩토리 함수와 클로저 예제

```python
# 다시 리뷰해보자!!

def outer(num1):    # 함수를 만들어 반환하는 함수 팩토리
    def inner(num2): # outer 안에서 만들어지는 함수 # num1 =10 은 inner가 기억하는 바깥 함수의 값 # 클로저는 num1 = 10을 기억하고 있는 inner 함수
        return num1 + num2

    return inner   # inner 함수 자체, inner()는 함수 실행!!!! 이므로, outer()함수의 return 값은 inner함수 자체를 반환한다!!!! 


a = outer(10)  # outer(10) 함수를 실행해서 a 변수에 넣는다. / 반환된 inner 함수를 저장한 변수
c = a(20) # a(20)은 저장된 inner 함수 실행한다.

print(c)

# 전체를 읽어 보면, outer(10)이면, outer함수의 return 값인 inner 함수 자체를  a변수에 저장!
# outer 함수를 실행하면, 하위 inner 함수자체가 return 값이 된 inner함수를 만나는데, 이때 inner함수는 10 + num2가 된다.
## inner 함수가 num1 + num2라는 코드를 가지고 있고, 바깥 함수에서 받은 num1 = 10을 기억한다.  이 inner 함수가 "클로저"
# a(20)이라면, 결국 10+num2 수식에 10+20이 되면서 a(20) = 30 으로 계산이 된 뒤 return이 된다. 여기서 이 값을 c로 넣어서 출력한다.
```

#### 콜백 함수 예제
- 콜백 함수란?
    - 콜백 함수(Callback Function)는 다른 함수에 전달되어, 그 함수 안에서 호출되는 함수
        쉽게 말하면, “이 함수를 너에게 맡길 테니, 필요한 시점에 대신 실행해 줘.”

```python
# 예시 1
def hello():    # 콜백 함수
    print("안녕하세요")


def execute(func):  # 콜백 함수를 전달받아 실행하는 함수
    func()          # 전달받은 hello 함수가 저장되는 매개변수


execute(hello)
```

```python
# 예시 2
def change_numbers(*args,callback):
    numbers = list(args) # *args가 튜플형태로 넘어오므로, list로 변환

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers
```

```python
def square(num):
    return num ** 2
```

```python
change_numbers(10,20,30,callback=square)
```

```python
def plus(num):
    return num + 10

change_numbers(10,20,30, callback=plus)
```

### 5. lamda 함수
---
- 간단하게 일회용으로 함수를 사용하는 경우, 한줄의 식으로 간단하게 함수를 정의할 때
- 일회용으로 사용되므로 매개 변수 이름도 최대한 간결하게 "한 자"로 표현합니다. (e.g) x

```python
# 함수의 구조
lambda 매개변수, 매개변수 ... : 처리 부분 # 처리부분은 return 부분
```

```python
# 예시 1
square = lambda num : num ** 2 
square(10)
```

```python
# 예시 2
def change_numbers(callback, *args):
    numbers = list(args) 

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers
```

```python
change_numbers(lambda num: num ** 2, 10, 20, 30)
```

```python
# lambda 함수의 변수는 최대한 짧게
change_numbers(lambda x : x ** 2, 10, 20, 30)
```

### 7. lamda와 map(),filter()를 활용한 데이터 일괄 가공
---
- map() : 변환 작업
- filter() : 특정 조건에서 데이터를 선별

```python
# map 함수 예시
fruits=['사과','오렌지','망고']
map(lambda x: f"**{x}**",fruits)
```

```python
fruits=['사과','오렌지','망고']
list(map(lambda x: f"**{x}**",fruits))
```

```python
# filter 함수 예시
numbers = [*range(1,11)]  # 1 ~ 10까지
numbers
```

```python
list(filter(lambda x: x % 2 == 1, numbers))
```
