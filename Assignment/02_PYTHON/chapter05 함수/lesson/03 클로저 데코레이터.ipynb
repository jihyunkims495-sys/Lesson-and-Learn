### 2. 클로저(Closure)
---
- 고차 함수(Higher Order Function)
    - 고차 함수는 두 가지 조건 중 하나 이상을 만족하는 함수를 의미합니다. 

1. 다른 함수를 인수로 전달받는 구조 
2. 둘째는 연산의 결과로 또 다른 함수를 반환하는 구조
파이썬에서는 함수를 실행하여 **새로운 함수를 만들어내고 반환하는 방식**을 통해 클로저와 데코레이터의 고도화된 제어를 구현합니다.

```python
# 예시 1 : 클로저 예시
def outer():
    num1 = 10

    def inner():
        num2 = 20
        print(num1 + num2)

    return inner

inner = outer() # num1 은 inner에 의해 사용되므로, 힙 영역에 생성되어 있음
inner() # 10 + 20
```

```python
# 예시 2 - 고차 함수
def add(num1):

    def _add(num2):

        def __add(num3):
            return num1 + num2 + num3

        return __add
        

    return _add
```

```python
result1 = add(10)(20)(30)

result1
```

```python
add10 = add(10)

add10(30)(40)
```

```python
add10_20 = add(10)(20)

add10_20(30)
```

### 3. 데코레이터(Decorator)
---
- 디자인 패턴 중 한가지
- 기존 함수의 코드는 안 건드리고, 앞뒤에 기능을 덧붙이는 함수
- 원래 함수를 포장해서 기능을 추가하는 것

```python
# 예시 1 : # 10부터 1까지 곱해지는 팩토리 연산
def factorial1(num):
    result = 1

    for i in range(1, num +1):
        result *= i

    return result
```

```python
factorial1(10) 
```

```python
import time


start = time.time()

result = factorial1(10)
print(result)

end = time.time()
print("걸린 시간:", end - start)
```

- 재귀 함수 (Recursive)

```python
# 예시 2 : 같은 함수 내에서 같은 함수를 반복해서 호출하는 방법 / 재귀 함수 (Recursive)
def factorial2(num):
    if num < 1:
        return 1

    return num * factorial2(num - 1)
```

```python
factorial2(10)
```

```python
import time

start = time.time()  ## 공통 기능

result = factorial2(10)  ## 핵심 기능
print(result)

end = time.time()   ## 공통 기능
print("걸린 시간:", end - start)   ## 핵심 기능
```

```python
import time

def time_checker(callback, num):
    start = time.time() # 공통 기능

    result = callback(num) # 핵심 기능 - 팩토리얼 연산 함수 (factorial1, factorial2)

    end = time.time() # 공통 기능
    print("걸린 시간:", end - start)

    return result
```

```python
time_checker(factorial2, 10)
```

```python
# 데코레이터 예시 - wrapper
def time_check2(callback):
    def wrapper(num):

        
        result = callback(num) # 핵심 기능

        return result 
```

```python
import time

def time_check2(callback):
    def wrapper(num):

        start = time.time() # 공통기능

        result = callback(num) # 핵심 기능

        end = time.time() # 공통 기능
        print("걸린 시간:", end - start)

        return result

    return wrapper
```

```python
@time_check2   # "@" 붙이면 데코레이터 기능이 붙었다!!!!
def factorial2(num):
    if (num < 1):
        return 1

    return num * factorial2(num - 1)
```

```python
factorial2(10)
```

```python
@time_check2   # 데코레이터를 호출해서 하단 함수의 성능 측정을 하고 싶어! 
def add(num1, num2):
    return num1 + num2
```

```python
add(10,20)
```
