### 2. 매개변수(parameter)와 인수(argument)
---
- 매개변수 (Parameter)
    - 함수 호출 시, 입력 부분에서 값을 담아주는 변수
    - 함수를 정의할 때 "외부에서 이런 데이터를 입력받을 거야"라고 미리 설정해두는 '입력 통로의 이름'입니다. 함수의 '선언부' 공간에 위치합니다.
- 인수 (Argument)
    - 호출시 사용된 값
    - 함수를 실제로 작동시킬 때 해당 통로를 통해 전달하는 '실제 데이터 값'입니다. 함수의 '호출부' 공간에서 주입됩니다.

```python
# 정의할 때의 name은 '매개변수'
def welcome(name):
    print(f"{name}님, 환영합니다!")

# 호출할 때의 "민수"는 '인수'
welcome("민수")
welcome("영희")
```

### 3. 위치 인수(positional argument) 와 키워드 인수(keyword arguments)
---

- 위치 인수(Positional Arguments)와 키워드 인수(Keyword Arguments)
    - 함수를 호출할 때 정의된 매개변수의 "순서대로" 인수를 차례대로 주입하는 가장 기본적인 방식
    - 인터프리터가 오직 '위치(순서)'만을 기준으로 데이터를 식별하여 매핑하므로, 순서가 바뀌면 의도치 않은 변수에 값이 대입되는 오작동을 유발할 수 있어 순서를 엄격히 준수해야 합니다.

- 키워드 인수를 활용한 자유로운 제어
    - 인수를 전달할 때 "매개변수명=값" 형태로 이름을 명시하여 전달하는 방식입니다. 이 구조를 사용하면 함수 정의부의 매개변수 배치와 상관없이 원하는 순서로 자유롭게 바꾸어 호출할 수 있습니다. 다만 위치 인수와 함께 사용할 때는 반드시 위치 인수를 먼저 작성해야 합니다. 이 방식을 활용하면 인수의 의미가 명확해져 코드의 가독성이 대폭 향상됩니다.

```python
def introduce_self(name, age):
    print(f"이름: {name}, 나이: {age}")

# 1. 위치 인수를 사용한 호출: 정의된 순서대로 정확히 주입해야 함
introduce_self("홍길동", 30)  # 정상 작동
introduce_self(30, "홍길동")  # 오작동 발생 (이름 공간에 수치 30이 식별됨)

# 2. 키워드 인수를 사용한 호출: 매개변수 이름을 특정하므로 순서가 상관없음
introduce_self(age=25, name="이영희")  # 순서를 뒤바꿔도 안전하게 매핑되어 정상 작동
```

```python
# 예시
def add(num1,num2) :
    result = num1 + num2

    return result 
```

```python
# 위에 예시에서, 변수가 3개(num1,num2,result)여서 메모리를 많이 차지하므로
# 변수를 아끼기 위해
def add(num1,num2) :
    return num1 + num2
# 이렇게 쓰기도 해요
# 여기에서, 순서를 지키면서 순차적으로! 
print(add(10,20)) # num1 = 10, num2 = 20으로 순서대로 인수가 맵핑 # 위치 인수 

print(add(num1=10,num2=20)) #이 형태, 키워드 맵핑도 가능 # 키워드 인수
print(add(num2=20,num1=20)) #키워드 인수는 위치 인수처럼 순서대로가 아니니까, 위치가 바뀌어도 되!
```

```python
def add(num1,num2,name,mobile):
    return f"{num1+num2},{name=},{mobile=}"
print(add(10,20,mobile="010-0000-0000",name="홍길동")) #위치인수, 키워드 인수 섞어 쓰는게 많아요~
```

### 4. 매개변수의 확장과 결과값 반환 제어
---
- 기본값 매개변수(Default Parameter) : 값을 안 줬을 때 자동으로 들어가는 값!
    - 기본값 매개변수의 작동 방식
        - 함수를 정의할 때 매개변수에 특정 값을 미리 대입해 두면, 함수 호출 시 인수를 생략하더라도 설정해둔 기본값이 자동으로 적용되어 작동합니다.
    - 문법적 선언 순서의 규칙
        - 기본값을 가진 매개변수는 반드시 기본값이 없는 매개변수보다 뒤쪽에 배치해야 합니다. 파이썬 인터프리터가 인수를 앞에서부터 차례대로 매핑하여 식별하기 때문에, 순서가 섞이면 어떤 매개변수에 값을 넣어야 할지 판단할 수 없어 문법 에러가 발생합니다.

- 가변 인수(*args)와 키워드 가변 인수(**kwargs)
    - 가변 인수(*args)
        - 입력될 인수의 개수가 실행 전에는 정확히 정해지지 않았을 때 사용합니다. 매개변수 앞에 별표 하나(*)를 붙여 선언하며, 외부에서 전달된 여러 인수를 하나의 '튜플(Tuple)' 구조로 묶어 차례대로 순회할 수 있도록 식별합니다.
    - 키워드 가변 인수(**kwargs)
        - 인수의 이름을 지정하여 키=값 형태로 전달하는 유연한 방식입니다. 매개변수 앞에 별표 두 개(**)를 붙여 선언하며, 입력된 모든 쌍 데이터를 하나의 '딕셔너리(Dictionary)' 구조로 변환하여 처리합니다.

-  return 문의 역할과 복수 데이터 반환
    - 결과값 반환과 즉시 종료
        - return 키워드는 함수 내부에서 연산된 최종 결과값을 함수 밖의 호출부로 내보내는 통로 역할을 합니다. 또한 인터프리터가 작동 중 return을 만나는 순간, 하위에 코드가 얼마나 남아있든 상관하지 않고 함수를 즉시 종료하고 탈출하는 제어 특성을 가집니다.
    - 복수 데이터 반환
        - 파이썬은 여러 개의 값을 쉼표로 구분하여 동시에 반환할 수 있습니다. 이때 실제로는 데이터들이 하나의 '튜플(Tuple)' 구조로 자동으로 묶여 내보내지며, 받는 쪽에서 여러 변수로 식별하여 활용할 수 있습니다.

1. 가변인수? (*args)
    - 함수가 받을 값의 개수를 미리 정할 수 없을 때 사용하는 문법
    - 함수가 위치 인수를 "개수 제한 없이 받도록" 만들어줍니다.
    - 가변인수는 위치 인수 기반! (당연히 키워드 인수는 안되겠죠?)
    - 매개변수의 갯수는 상관 없습니다!
    - 결과가 튜플 형태로 나옵니다!!!
## 'args' 라는 이름이 반드시 필요한가? -> "NO!" 가변인수는 "*" 별표가 핵심이에요!
## 하지만, 개발자들이 관례상 *args를 많이 사용합니다. *args = 여러 개의 위치 인수를 튜플로 받기

```python
# 왜 가변 인수가 필요할까?
## 일반적인 함수는 받을 값의 개수를 미리 정합니다.
 def add(a, b):
    return a + b
# 이 함수는 값 2개만 받도록 설계되었습니다. 그래서
add(10, 20)      # 정상이고
add(10, 20, 30)  # 오류입니다.
# 그런데 숫자를 몇 개 받을지 미리 알 수 없다면 어떻게 해야 할까요?
add_all(10, 20)
add_all(10, 20, 30)
add_all(10, 20, 30, 40, 50)
# 이때 *args를 사용합니다.
```

```python
def add_all(*numbers):
    print(numbers)
#이 함수를 호출한다면,
add_all(1, 2, 3)
# 함수 안에서는 다음과 같이 묶입니다.
numbers = (1, 2, 3)
# 즉, 전달된 값들이 튜플 하나로 묶이는 것입니다. 
# 1,2,3 -> (1,2,3) <- 튜플의 형태로! 
```

```python
# 응용 예시!
def add_all(*numbers):
    total = 0

    for number in numbers:
        total += number

    return total

print(add_all(10, 20, 30))

# 위 예시의 실행과정을 보면, 
numbers = (10, 20, 30)

total = 0
total = 0 + 10  → 10
total = 10 + 20 → 30
total = 30 + 30 → 60
```

2. 키워드 가변 인수 (**kwargs)
    - **kwargs는 키=값 형태의 인수를 "개수 제한 없이" 받습니다.
    - 결과가 딕셔너리 형태로 나옵니다!!!!
## kwargs라는 이름이 반드시 필요한가? -> "NO!" 가변인수는 "**" 별표2개가 핵심이에요!
## **kwargs = 여러 개의 키워드 인수를 딕셔너리로 받기

```python
def user_info(**data):
    print(data)

user_info(name="지현", level=5, active=True)
# 다음과 같이 호출하면, 함수 내부에서는 딕셔너리로 묶입니다.
data = {
    "name": "지현",
    "level": 5,
    "active": True
}
# name="지현", level=5, active=True -> {"name": "지현", "level": 5, "active": True}
# 호출할 때 작성한 이름이 딕셔너리의 key가 되고, 실제 값이 딕셔너리의 value가 됩니다.

# 코드 실행 과정
def user_info(**data):
    print(data)


user_info(name="지현", level=5, active=True)
# 위에 것을 실행하면, 
{'name': '지현', 'level': 5, 'active': True} # 딕셔너리 형태로 결과가 나옵니다. 
```

```python
def user_info(**data):
    for key, value in data.items():
        print(f"{key}: {value}")


user_info(name="지현", level=5, active=True)
```

## `*args`와 `**kwargs`의 차이

| **구분** | **`*args`** | **`**kwargs`** |
|---|---|---|
| **전달 방법** | 값만 전달 | `키=값` 형태로 전달 |
| **호출 예시** | `func(10, 20, 30)` | `func(name="지현", age=38)` |

### 5. 함수를 활용한 기능 분리와 집계 시스템 구현
---
- 인자 개수 제한 없는 평균값 계산 함수 구현
    - 가변 인수를 활용한 유연한 통계 집계
        - 입력 데이터의 개수가 유동적인 상황에서 평균을 산출하는 구조입니다. 전달된 인수가 없을 때 발생할 수 있는 오류를 차단하기 위해 데이터의 개수 len()를 먼저 식별하는 방어 코드를 포함하여 작동시킵니다.

```python
# 여러 점수의 평균을 구하는 함수
def get_average(*scores):   # *scores는 여러 값을 받는 역할이고, 받은 값들을 튜플로 묶습니다.
    if len(scores) == 0:    # 점수의 갯수를 확인합니다. # len()
        return 0            # 점수가 하나도 없다면 나누기 계산을 하지 말고 바로 0을 반환하며 함수를 종료해라.! 오류 방어코드 !!

    total = sum(scores)     # 점수의 합계 계산
    avg = total / len(scores) # 점수의 평균 계산 = 합계 / 갯수
    return avg

result1 = get_average(80, 90, 100)
result2 = get_average(75, 60, 88, 92, 100, 45)

print(f"3과목 평균 점수: {result1:.2f}")
print(f"6과목 평균 점수: {result2:.2f}")
-----
# 이 함수의 흐름을 정리하면, 
# get_average(80, 90, 100) → scores = (80, 90, 100) → 점수 개수는 3개이므로 계속 실행 → total = 270 → avg = 270 / 3 → 90.0 반환 → result1 = 90.0 → 화면에 90.00 출력
```

### 7. [심화] 파이썬 함수의 메모리 생애주기 (Heap vs Call Stack)
---
- 메모리 영역별 역할 비교
    - 파이썬은 "함수의 정의(코드)는 힙(Heap)에 상주하고, 함수의 실행(상태)은 콜 스택(Call Stack)에 임시 생성되었다가 사라진다"는 구조를 따릅니다.
| **구분** | **힙(Heap) 영역** | **콜 스택(Call Stack) 영역** |
|---|---|---|
| **저장 대상** | **함수 정의 객체** (`PyCodeObject`, `PyFunctionObject`) | **함수 실행 프레임 객체** (`PyFrameObject`) |
| **생성 시점** | 모듈을 불러오고 `def` 문을 해석할 때 | 함수를 호출할 때 (`func()`) |
| **소멸 시점** | 프로그램 종료 또는 가비지 컬렉션(GC)으로 정리될 때 | 함수 실행이 끝날 때 (`return`) 스택에서 제거(POP) |
| **특징** | 함수를 여러 번 호출해도 **하나의 원본 함수 객체를 재사용** | 호출할 때마다 **새로운 독립 실행 프레임 생성**(재귀 호출 포함) |

- 단계별 메모리 적재 프로세스
    - 정의 단계 (def 문 해석)
        - 파이썬 인터프리터가 바이트코드를 읽어 힙(Heap)에 불변 데이터 구조체인 PyCodeObject와 이를 감싸는 PyFunctionObject를 올립니다.
    - 호출 단계 (Function Call)
        - 함수가 호출되면 콜 스택(Call Stack)에 전용 실행 공간인 **PyFrameObject**가 PUSH됩니다.
        - 매개변수와 지역 변수는 프레임 내부의 **f_locals** 영역에 실시간으로 매핑됩니다.
    - 종료 단계 (Return)
        - 함수가 값을 반환하고 종료되면 콜 스택에서 해당 프레임 객체가 POP 되며, 임시 할당되었던 지역 변수 공간이 소멸합니다.
        - 힙 영역에 존재하는 원본 함수 코드 객체는 그대로 유지되어 다음 호출 시 다시 활용됩니다.

- 파이썬 함수의 실행 문맥: 프레임 객체(Frame Object)
    - 파이썬 인터프리터(CPython)는 함수가 호출될 때마다 해당 함수의 실행 상태를 관리하기 위해 ‘프레임 객체(Frame Object, PyFrameObject)’라는 메모리 구조체를 생성하여 콜 스택(Call Stack)에 쌓습니다.
    - 함수가 호출되면 새 프레임 객체가 생성되어 스택에 PUSH됩니다.
    - 함수 실행이 완료(리턴)되면 해당 프레임 객체는 스택에서 POP되어 메모리에서 해제됩니다.

- 프레임 객체의 주요 구성 요소
    - 프레임 객체는 함수가 실행되는 동안 필요한 모든 내부 정보(지역 변수, 전역 변수, 호출자 정보 등)를 담고 있습니다.
| **속성명** | **역할 및 설명** |
|---|---|
| `f_locals` | 현재 함수 내부의 지역변수(Local Variables)와 매개변수를 저장하는 딕셔너리 |
| `f_globals` | 현재 함수가 속한 모듈의 전역변수(Global Variables)를 참조하는 딕셔너리 |
| `f_builtins` | `print()`, `len()` 등 파이썬 내장 함수와 예외 객체가 속한 스코프 딕셔너리 |
| `f_back` | 현재 프레임을 호출한 이전 프레임(부모 함수 또는 호출자)을 가리키는 참조 |
| `f_code` | 함수의 바이트코드, 상수 목록(`co_consts`), 변수명 목록(`co_varnames`) 등을 담고 있는 코드 객체 |

-  프레임 객체 직접 확인하기
    - 파이썬은 sys 모듈의 _getframe() 함수를 통해 실행 중인 프레임 객체에 직접 접근할 수 있습니다.


```python
import sys

def outer_func():
    x = 10
    
    def inner_func():
        y = 20
        
        # 1. 현재 실행 중인 프레임 객체 가져오기
        frame = sys._getframe()
        
        print(f"1. 현재 실행 함수: {frame.f_code.co_name}")
        print(f"2. 현재 지역 변수(f_locals): {frame.f_locals}")
        print(f"3. 전역 스코프 참조 여부: {'outer_func' in frame.f_globals}")
        
        # 2. f_back을 통해 부모 프레임(outer_func)으로 이동
        parent_frame = frame.f_back
        print(f"4. 부모 함수 이름: {parent_frame.f_code.co_name}")
        print(f"5. 부모 함수의 지역 변수: {parent_frame.f_locals}")

    inner_func()

outer_func()
```

### [별첨.] 메모리란?
- 메모리는 프로그램이 실행되는 동안 코드와 데이터를 잠시 올려놓는 작업 공간
    - 정적 메모리: 공용 자료 보관 공간

    - 스택 메모리: 현재 처리 중인 업무 서류를 쌓아두는 공간

#### 1. 정적 메모리
- 프로그램을 실행할 때 만들어지고, 프로그램이 종료될 때까지 유지되는 공간
    - 프로그램의 코드
    - 전역변수
    - 프로그램 전체에서 사용하는 데이터

#### 2. 동적 메모리
- 동적 메모리는 프로그램이 실행되는 중에 필요한 만큼 그때그때 만들어 사용하는 공간이에요. 일반적으로 "힙(Heap) 메모리"라고 부릅니다.
- 문자열, 리스트, 딕셔너리 같은 객체가 주로 이 공간에 만들어집니다.
- 파이썬은 해당 객체가 더 이상 필요하지 않으면 가비지 컬렉션이라는 관리 기능을 통해 메모리를 정리합니다.

#### 3. 스택 영역 메모리 (임시 메모리)
-  함수가 실행될 때 특히 중요하게 사용되는 공간
- 함수를 호출하면 그 함수의 작업 공간인 스택 프레임(Stack Frame)이 만들어집니다.
    - 함수에 전달받은 값
    - 함수 내부에서 만든 지역변수
    - 함수 실행이 끝난 후 돌아갈 위치
    - 현재 함수의 실행 상태

#### 변수 자체가 객체를 담는 상자라기보다 객체를 가리키는 이름
#### 함수는 호출될 때 자신만의 임시 작업 공간을 만들고, 작업이 끝나면 결과만 돌려준 뒤 그 공간을 정리한다.

##### [기타.]

1. 스택(Stack)과 큐(Queue)
- 스택은 마지막에 넣은 것부터, 큐는 먼저 넣은 것부터 꺼내는 자료구조
- “접시 = 스택, 대기 줄 = 큐”
- 스택 (Stack) : Last In, First Out
    - 쌓여있는 구조로, 마지막에 넣은 것을 먼저 꺼냄 (e.g) 컵 처럼, 세워진 원통의 구조로 바닥이 막혀있는 ...
    - 아래가 막혀있어서 가장 처음 값이 가장 나중에 꺼내지는 자료 구조
    - 실제 사용 사례
        - 웹브라우저의 뒤로 가기
        - Ctrl + Z 실행 취소
        - 함수가 실행된 순서를 기억하는 호출 스택
- 큐(Queue): First In, First Out
    - 대기열이라는 뜻 / 선착순!
    - 먼저 넣은 것을 먼저 꺼냄 (e.g) 가로 통로 처럼, 양 옆이 뚫린 구조...
    - 실제 사용 사례
        - 프린터 출력 대기
        - 은행 번호표 대기
        - 주문 처리
        - 서버 요청 처리

```python
# 스택 예시
## append() : 맨 뒤에 데이터 넣기
## pop() : 맨 뒤의 데이터 꺼내면서 삭제하기

stack = []

stack.append("접시 1")
stack.append("접시 2")
stack.append("접시 3")

print(stack.pop())  # 접시 3
print(stack.pop())  # 접시 2
print(stack.pop())  # 접시 1
```
