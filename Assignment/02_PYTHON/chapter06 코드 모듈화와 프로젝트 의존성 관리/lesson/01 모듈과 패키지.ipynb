### 1. 독립 소스 파일 모듈과 디렉터리형 패키지 구조
---
우리가 작성하는 코드의 규모가 커지면 하나의 파일에 모든 로직을 담는 것이 불가능해집니다. 가독성이 떨어질 뿐만 아니라, 특정 기능을 수정할 때 코드 전체를 찾아다녀야 하기 때문입니다. 

파이썬은 코드를 성격에 따라 여러 파일로 나누어 관리하는 '모듈(Module)'과, 이 모듈들을 폴더 형태로 계층화하여 묶는 '패키지(Package)' 구조를 제공합니다. 소스 코드를 효율적으로 분리하고 관리하는 독립적 파일 구조의 원리를 알아보겠습니다.

#### 1.1 독립 소스 파일 모듈(Module)
- 모듈의 본질
    - 모듈은 파이썬 코드가 저장된 하나의 독립적인 .py 파일입니다. 자주 사용되는 변수, 함수, 클래스들을 기능별로 하나의 파일에 모아둔 것을 의미하며, 이를 통해 코드의 중복을 줄이고 언제든 필요한 공간에서 호출하여 재사용할 수 있는 장점을 가집니다.
- 코드 관리의 효율성
    - 모듈을 사용하면 특정 기능에 문제가 생겼을 때 전체 코드를 건드리지 않고 해당 모듈 파일만 수정하면 되므로 안전하게 코드를 관리하는 유연한 구조를 구성할 수 있습니다.

#### 1.2 디렉터리형 패키지(Package)
- 패키지의 개념
    - 패키지는 여러 모듈 파일들을 체계적으로 묶어둔 '폴더(디렉터리)'입니다. 모듈의 개수가 많아지면 폴더를 통해 계층적으로 분류하여 보관해야 하는데, 이 폴더 구조 자체를 패키지라고 식별합니다.
- 계층 구조와 패키지
    - 패키지 내부에는 또 다른 하위 폴더(서브 패키지)나 다양한 모듈 파일들이 존재할 수 있습니다. 파이썬 폴더 구조 내에서 점(.) 기호를 사용하여 상위 폴더와 하위 폴더, 그리고 모듈 파일의 경로를 추적하여 접근하는 동작 방식을 취합니다.
- 상대 경로를 통한 계층 접근 방식
    - 패키지 구조 안에서 특정 모듈이 다른 모듈을 불러올 때, 고정된 전체 경로 대신 현재 위치를 기준으로 접근하는 상대 경로 표기법을 사용할 수 있습니다.
    - **단일 점 (.):** '현재 내가 속한 폴더 위치'를 식별합니다. (예: **from . import 현재폴더모듈**)
    - **더블 점 (..):** 계층 구조 상에서 '나를 감싸고 있는 상위 폴더 위치'를 식별합니다. (예: **from ..상위폴더 import 상위모듈**)


##### 상대 경로 사용 시 주의점
---
<aside>

### 2. 모듈 로드를 위한 키워드와 표준 라이브러리 활용
---
- 외부로 분리된 모듈이나 패키지를 현재 내가 작업 중인 파일로 불러와 사용하기 위해서는 전용 키워드가 필요합니다. 
- 파이썬이 기본적으로 제공하는 강력한 도구 모음인 '파이썬 표준 라이브러리'의 검색 방법과 함께, import, from, as 키워드를 활용해 원하는 요소를 정확히 식별하여 로드하는 방법을 배워보겠습니다.

#### 2.1 import, from, as 키워드
- import 키워드
    - 특정 모듈 전체를 현재 파일의 메모리에 로드할 때 사용하는 가장 기본적인 명령어입니다. 모듈 안의 변수나 함수를 호출할 때는 모듈이름.함수이름() 형태로 출처를 명시하여 식별합니다.
- from 키워드
    - 모듈 전체가 아닌, 그 내부의 특정한 함수나 클래스만을 선택하여 직접 가져오고 싶을 때 사용합니다. 이 구조를 활용하면 호출 시 앞에 모듈 이름을 붙이지 않고 함수 이름만으로 곧바로 동작시킬 수 있습니다.
- as 키워드
    - 가져올 모듈이나 함수의 이름이 너무 길거나 현재 파일의 다른 변수명과 충돌할 위험이 있을 때, 개발자가 원하는 짧은 '별칭(Alias)'을 부여하는 도구입니다.

```python
# import 키워드를 사용한 전체 모듈 로드 방식
import math
print(math.sqrt(16))  # 모듈명을 명시하여 식별

# from 키워드를 사용한 특정 기능 로드 방식
from math import sqrt
print(sqrt(25))       # 모듈명 없이 함수 이름만으로 곧바로 실행

# as 키워드를 사용한 별칭 부여 방식
import math as m
print(m.sqrt(36))     # 설정한 별칭 m을 사용하여 연산 수행
```

#### 2.2 파이썬 표준 라이브러리
- 파이썬 표준 라이브러리(Standard Library)
    - 파이썬을 설치할 때 별도의 추가 다운로드 없이 기본적으로 함께 탑재되어 제공되는 유용한 모듈들의 모음집입니다. 운영체제 제어, 수학 연산, 시간 가공 등 이미 검증된 고성능 모듈들이 준비되어 있습니다.
- 대표적인 파이썬 표준 라이브러리

| 모듈명 | 주요 기능 및 역할 | 핵심 동작 방식 및 특징 |
|---|---|---|
| `sys` | 시스템 및 인터프리터 제어 | 파이썬 인터프리터가 제공하는 변수와 함수를 직접 제어하며, 실행 경로(`path`) 확인이나 시스템 종료 등의 흐름을 식별합니다. |
| `random` | 무작위 난수 생성 | 지정된 범위 내에서 무작위로 값을 뽑거나, 시퀀스 데이터의 순서를 임의로 섞는 등 다양한 난수 생성 기능을 활용합니다. |
| `datetime` | 날짜 및 시간 데이터 가공 | 컴퓨터 시스템의 시계 수치를 조회하여 날짜와 시간 정보가 결합된 객체 구조를 생성하고, 표준 포맷 문자열로 변환합니다. |

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
```

```python
import calculator as calc

# 모듈 내부 함수를 식별하여 연산 결과 도출
res_add = calc.add(10, 5)
res_sub = calc.subtract(10, 5)

print(f"사용자 정의 모듈 더하기 결과값: {res_add}")
print(f"사용자 정의 모듈 빼기 결과값: {res_sub}")
```

#### 3.2 random 내장 모듈과 datetime 포맷팅 출력 구현

1. 무작위 번호 추출

```python
import random

# random 모듈의 메서드를 활용한 번호 추출기 구현 (1부터 45 사이의 무작위 수)
lucky_number = random.randint(1, 45)
print(f"추출된 무작위 번호: {lucky_number}")
```

2. datetime 포맷팅 출력

```python
import datetime

# datetime 모듈을 활용한 현재 시간 수집 및 포맷팅 문자열 출력
current_time = datetime.datetime.now()
print(f"원시 시간 데이터 구조: {current_time}")

# 원하는 양식의 문자열 형식으로 변환 출력
formatted_time = current_time.strftime("%Y-%m-%d %H:%M:%S")
print(f"포맷팅 완료된 표준 시간 출력: {formatted_time}")
```

## 개념 요약 키워드
- **모듈과 패키지:** 기능별로 소스코드를 분리해 담은 독립적인 하나의 .py 파일(모듈)과, 이러한 모듈들을 폴더 형태로 체계화하여 관리하는 계층적 디렉터리 구조(패키지)의 개념입니다.
- **import와 from:** 외부 모듈 전체를 현재 파일로 가져오는 기본 방식(import)과, 모듈 내에서 필요한 특정 변수나 함수만 선택해서 가져오는 방식(from ... import)의 차이입니다.
- **as 키워드:** 이름 충돌을 예방하고 소스코드 가독성을 향상시키기 위해 임의의 짧은 대체 명칭을 변수에 매핑하는 별칭 지정 구문입니다.
- **파이썬 표준 라이브러리:** 별도의 설치 없이 기본으로 탑재되어 시스템, 수학, 시간 연산 등을 즉시 제공하는 내장 모듈 집합입니다.
- **사용자 정의 모듈:** 직접 작성한 .py 파일을 import하여 다른 코드에서 함수, 변수, 클래스를 재사용할 수 있습니다.

## 용어 사전

| 용어 | 영어 표기 | 설명 |
|---|---|---|
| 모듈 | Module | 공통으로 재사용할 수 있는 함수, 변수, 클래스들을 따로 분리해서 담아놓은 하나의 파이썬 파일(`.py`)입니다. 코드의 비대화를 막고 가독성을 높이는 독립된 부품 역할을 합니다. |
| 패키지 | Package | 관련된 여러 개의 모듈 파일들을 효율적으로 계층화하여 관리하기 위해 하나의 폴더(디렉터리) 구조로 묶어놓은 모듈 주머니입니다. 폴더 구조를 통해 대규모 프로젝트의 경로를 체계적으로 관리합니다. |
| 표준 라이브러리 | Standard Library | 파이썬을 설치할 때 컴퓨터에 기본으로 제공되는 검증된 모듈과 패키지들의 집합입니다. 외부에서 따로 설치할 필요 없이 곧바로 불러와 사용할 수 있는 기본 도구 모음입니다. |
| `import` 키워드 | import Keyword | 외부 파일(모듈)이나 패키지에 준비되어 있는 기능을 현재 작업 중인 소스코드에서 사용할 수 있도록 불러오는 명령어입니다. |
| `from` 키워드 | from Keyword | 특정 모듈이나 패키지 안에서 필요한 함수, 클래스, 변수 등을 지정하여 가져올 때 사용하는 명령어입니다. |
| `as` 키워드 | as Keyword | 불러온 모듈이나 함수의 이름이 너무 길거나 가독성이 떨어질 때, 소스코드 안에서 편리하게 사용할 수 있도록 별칭을 붙여주는 명령어입니다. 예: `import pandas as pd` |
| 사용자 정의 모듈 | User-defined Module | 파이썬이 기본으로 제공하는 모듈이 아니라, 개발자가 필요한 기능을 직접 작성하여 별도의 `.py` 파일로 분리해 만든 모듈입니다. 예: `calculator.py` |
| `random` 모듈 | random Module | 파이썬 표준 라이브러리 중 하나로, 무작위 숫자(난수)를 생성하거나 데이터에서 임의의 요소를 추출하고 순서를 섞는 기능 등을 제공합니다. |
| `datetime` 모듈 | datetime Module | 파이썬 표준 라이브러리 중 하나로, 날짜와 시간 데이터를 생성·조회·계산하고 원하는 문자열 형식으로 변환할 수 있도록 지원합니다. |
| `sys` 모듈 | sys Module | 파이썬 표준 라이브러리 중 하나로, 파이썬 인터프리터와 관련된 시스템 정보를 조회하거나 모듈 탐색 경로인 `sys.path` 등을 확인하고 제어할 수 있도록 지원합니다. |

3. import,from,as 키워드
- import : 모듈 불러오기
- 불러온 묘듈에서 마침표(.)를 통해서 접근 합니다. 
- from 모듈명/패키지명 import 모듈에 있는 함수, 변수, 클래스 등등에 접근 가능
- 필요에 따라 .py파일을 만들어서 원하는 것들을 추출해서 자주 사용합니다.
- from 모듈명 import * : 모듈의 전체 함수, 변수, 클래스를 불러옵니다. / 자원이 많지 않을 때, "전체 불러오기!"

- as : alias의 별칭 / 별명을 붙이는 키워드로 "가져온 모듈/함수/클래스에 별명 붙이기"
    - import math as m # math 모듈을 가져오고, 이 파일에서는 m이라는 이름으로 부르겠다.

```python
import mod1    # 모듈명 = 파일 이름 / mod1(모듈 예시).py 파일 전체가 하나의 모듈
```

#### \_\_init\_\_.py
- 현재 디렉토리가 패키지임을 알려주는 파일
- 하위 패키지/모듈에서 공통적으로 처리해야 하는 코드를 작성할 수 있으며, 모듈을 불러올 때 최초 한번만 실행됩니다. 
- 요즘 Python에서는 __init__.py가 없어도 패키지처럼 동작할 수 있는 경우가 있습니다.

- 패키지 관련 초기 설정을 둘 수 있는 특별한 파일
- __init__.py는 그 패키지를 처음 사용할 때 기본적으로 어떤 걸 준비할지 정하는 안내데스크 같은 역할
- 언제 실행 되나?
    - import 폴더(=패키지)
        - Python이 utils 패키지를 불러오면서 utils/__init__.py 안의 코드를 실행할 수 있습니다.
- 패키지 안의 특별한 파일
- 패키지를 불러올 때 초기 설정을 넣을 수 있음
- 다른 모듈의 기능을 패키지 바깥으로 꺼내주는 역할도 가능

- 패키지 내에서 사용할 공통적인 함수, 변수, 클래스를 정의할 수 있습니다. 

- from 패키지 import * : __all__ = [....]: 에 정의된 모듈만 import
