### 필수 1 : random 모듈을 활용한 주사위 게임 만들기
---
#### 문제 1-1 :  무작위 주사위 눈 추출하기

```python
import random

a = random.randint(1, 6)
print("결과:", a)
```

### 필수 2 : datetime 모듈을 활용한 현재 날짜 출력하기
---
#### 문제 2-1 :  연도, 월, 일 형식 지정 출력하기

```python
import datetime

now = datetime.datetime.now()

today = now.strftime("%Y-%m-%d")

print(today)
```

### 심화 1 : as 키워드를 활용한 모듈 별칭 부여 및 시간 연산
----
#### 문제 3-1 : 별칭을 사용한 현재 시각 상세 출력

```python
import datetime as dt

now = dt.datetime.now()

now_time = now.strftime("%H:%M:%S")

print(now_time)
```
