### 필수 1 : 딕셔너리 생성 및 안전한 데이터 조회

#### 문제 1-1 :  회원 정보 수정 및 결측치 예외 처리

```python
dict = {
    'name':'Sparta',
    'age':20
}

dict['age'] = 21
print(dict)

dict['role'] = 'admin'
print(dict)

phone = dict.get('phone',None)
print(phone)
```

```python
phone = dict.get('phone','010-0000-0000')
print(phone)
```

### 필수 2 : 집합을 활용한 중복 데이터 정제 및 포함 여부 확인
#### 문제 2-1 :  이메일 로그 중복 제거 및 리스트 복원

```python
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"] 

raw_emails_2 = set(raw_emails)
print(raw_emails_2)

refined_emails = list(raw_emails_2)
print(refined_emails)
```
