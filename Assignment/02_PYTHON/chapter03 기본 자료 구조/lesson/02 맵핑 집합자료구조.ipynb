##### 1.1 딕셔너리의 정의와 해시 테이블 기반 작동 방식
- 키-값(Key-Value) 구조
    - 딕셔너리는 데이터를 저장할 때 고유한 식별자인 키(Key)와 실제 저장할 값(Value)을 한 쌍으로 묶어서 관리하는 중괄호({}) 기반 자료구조
- 해시 테이블(Hash Table) 작동 방식
    - 내부적으로 해시 함수를 사용하여 키를 고유한 메모리 주소로 실시간 변환하기 때문에, 시퀀스 자료구조처럼 처음부터 끝까지 데이터를 하나씩 훑어보지 않고도 원하는 데이터를 단번에 식별하고 조회할 수 있습니다.
    - 검색이 가능하다는 말!!!
- 키의 중복 불가능성
    - 하나의 딕셔너리에는 같은 키가 두 개의 독립된 항목으로 동시에 존재할 수 없습니다. 같은 키에 값을 다시 대입하면 기존 값이 새 값으로 덮어써집니다.

##### 1.2 딕셔너리의 생성 및 가변성(Mutability)
- 가변 자료형(Mutable)
    - 리스트와 마찬가지로 생성 이후에 요소를 자유롭게 추가, 수정, 삭제할 수 있는 특성을 가집니다.
- 동적 데이터 갱신
    - 존재하지 않는 새로운 키에 값을 대입하면 자동으로 새로운 키-값 쌍이 추가되며, 기존에 존재하는 키에 새로운 값을 대입하면 실시간으로 값이 수정됩니다.

```python
# 딕셔너리 선언 및 제어 예시
user_info = {"name": "Sparta", "age": 20}
user_info["age"] = 21       # 값 수정
user_info["role"] = "admin" # 새로운 키-값 쌍 추가
print(user_info)  # 출력: {'name': 'Sparta', 'age': 21, 'role': 'admin'}
```

- 파이썬은 딕셔너리 내부의 데이터를 유기적으로 가공할 수 있는 전용 내장 메서드를 지원하며, 존재하지 않는 키를 조회했을 때 프로그램이 강제 종료되는 치명적인 에러를 안전하게 방지하는 보완 도구도 제공합니다. 코드의 가독성과 안전성을 높이는 핵심 제어 기술을 정리합니다.


##### 2.1 주요 뷰(View) 메서드 활용
- keys() 메서드
    - 딕셔너리 내부에 존재하는 모든 고유 키(Key)들을 모아서 반환합니다. 데이터의 식별자 목록을 일괄 검사할 때 주로 활용됩니다

- items() 메서드
    - 키와 값을 하나의 튜플 묶음((Key, Value))으로 결합하여 전체 목록을 반환합니다. 반복문과 결합하여 데이터를 순회 정제할 때 핵심적으로 쓰이는 작동 방식입니다.

#### 2.3 자주 사용하는 메서드

```python
# ==========================================
# 딕셔너리 메서드 사용 예시
# ==========================================


# 1. get(key, default)
# 키가 있으면 해당 값을 반환
# 키가 없으면 None 또는 지정한 기본값을 반환

d = {"name": "김지현", "age": 38}

print(d.get("age"))        # 38
print(d.get("job"))        # None
print(d.get("job", "없음")) # 없음


# 2. keys()
# 딕셔너리의 모든 키를 반환

d = {"name": "김지현", "age": 38, "job": "MD"}

keys = d.keys()

print(keys)
# dict_keys(['name', 'age', 'job'])

for key in d.keys():
    print(key)


# 3. values()
# 딕셔너리의 모든 값을 반환

d = {"name": "김지현", "age": 38, "job": "MD"}

values = d.values()

print(values)
# dict_values(['김지현', 38, 'MD'])

for value in d.values():
    print(value)


# 4. items()
# 딕셔너리의 모든 (키, 값) 쌍을 반환

d = {"name": "김지현", "age": 38, "job": "MD"}

items = d.items()

print(items)
# dict_items([('name', '김지현'), ('age', 38), ('job', 'MD')])

for key, value in d.items():
    print(key, value)


# 5. update(other)
# 다른 딕셔너리의 내용을 추가하거나 기존 값을 수정

d = {"name": "김지현", "age": 38}

d.update({"age": 39, "job": "AI 서비스 기획자"})

print(d)
# {'name': '김지현', 'age': 39, 'job': 'AI 서비스 기획자'}

# 기존 키인 age의 값은 수정되고,
# 없었던 키인 job은 새로 추가됨


# 6. pop(key, default)
# 지정한 키와 값을 삭제하고, 삭제된 값을 반환

d = {"name": "김지현", "age": 38, "job": "MD"}

deleted_value = d.pop("age")

print(deleted_value)  # 38
print(d)
# {'name': '김지현', 'job': 'MD'}

# 키가 없을 때 기본값 사용
result = d.pop("city", "해당 키 없음")

print(result)  # 해당 키 없음
print(d)       # 딕셔너리는 그대로 유지됨


# 7. popitem()
# 마지막으로 추가된 (키, 값) 쌍을 삭제하고 튜플로 반환

d = {
    "name": "김지현",
    "age": 38,
    "job": "MD"
}

key, value = d.popitem()

print(key)    # job
print(value)  # MD
print(d)
# {'name': '김지현', 'age': 38}


# 8. setdefault(key, default)
# 키가 있으면 기존 값을 반환
# 키가 없으면 기본값을 저장한 후 반환

d = {"name": "김지현", "count": 3}

# count 키가 있으므로 기존 값 3을 반환
result = d.setdefault("count", 0)

print(result)  # 3
print(d)
# {'name': '김지현', 'count': 3}

# city 키가 없으므로 '서울'을 저장하고 반환
result = d.setdefault("city", "서울")

print(result)  # 서울
print(d)
# {'name': '김지현', 'count': 3, 'city': '서울'}


# 9. clear()
# 딕셔너리의 모든 항목 삭제

d = {"name": "김지현", "age": 38, "job": "MD"}

d.clear()

print(d)
# {}
```

#### 3. 중복 없는 고유한 데이터 관리: 집합(Set) 자료구조
---
- 수학의 집합 개념을 프로그래밍 세상으로 고스란히 가져온 자료구조가 바로 집합(Set)
- 집합 자료구조는 대량의 데이터 집합 속에서 중복된 요소를 단 한 줄의 코드로 완전히 걸러내고 고유한 알맹이만 남겨야 할 때 탁월한 성능을 발휘합니다. 
- 중복을 허용하지 않으며, 순서가 결정되지 않습니다.
- {값1,값2,값3....} 형태로 집합이 됩니다.

- 중복 불가능성
    - 집합 내의 모든 요소는 무조건 고유해야 합니다. 동일한 데이터를 여러 번 추가하더라도 내부적으로는 단 하나의 고유 데이터만 남도록 강제 필터링
- 순서 없는(Unordered) 특성
    - 집합 자료구조는 요소 간의 앞뒤 순서가 존재하지 않으므로, 인덱스를 활용한 위치 번호 조회(set_data[0])나 슬라이싱 구간 자르기가 불가능합니다. 딕셔너리의 키처럼 내부 해시 테이블 구조로 고유 요소를 실시간 식별하기 때문

```python
a = {10,10,10,20,30}  #10이란 요소가 3번 중복되어 있다면
print(a) #a는 중복값은 없어지고 출력 됩니다. 
```

##### 집합 연산
- 합집합 : a|b a.union(b)
- 교집합 : a&b a.intersection(b)
- 차집합 : a-b a.difference(b)
- 대칭차집합 : a^b a.symmetric.differnce(b)

```python
# ==========================================
# 집합(set) 연산 예시
# ==========================================

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}


# 1. 합집합
# a 또는 b에 들어 있는 모든 요소
# 중복되는 값은 한 번만 포함

result = a.union(b)
print(result)
# {1, 2, 3, 4, 5, 6}

# 연산자 사용
result = a | b
print(result)
# {1, 2, 3, 4, 5, 6}


# 2. 교집합
# a와 b에 공통으로 들어 있는 요소

result = a.intersection(b)
print(result)
# {3, 4}

# 연산자 사용
result = a & b
print(result)
# {3, 4}


# 3. 차집합
# a에는 있지만 b에는 없는 요소

result = a.difference(b)
print(result)
# {1, 2}

# 연산자 사용
result = a - b
print(result)
# {1, 2}


# 차집합은 순서가 중요함
# b에는 있지만 a에는 없는 요소

result = b.difference(a)
print(result)
# {5, 6}

result = b - a
print(result)
# {5, 6}


# 4. 대칭차집합
# a와 b 중 한쪽에만 있는 요소
# 두 집합의 공통 요소는 제외

result = a.symmetric_difference(b)
print(result)
# {1, 2, 5, 6}

# 연산자 사용
result = a ^ b
print(result)
# {1, 2, 5, 6}
```

```python
# ==========================================
# 요소 조작 메서드
# ==========================================


# 1. add(x)
# 집합에 요소 하나를 추가

s = {1, 2, 3}

s.add(4)

print(s)
# {1, 2, 3, 4}


# 이미 존재하는 값을 추가해도 중복 저장되지 않음

s.add(2)

print(s)
# {1, 2, 3, 4}


# ------------------------------------------


# 2. update(iterable)
# 리스트, 튜플, 집합 등에 들어 있는
# 여러 요소를 한꺼번에 추가

s = {1, 2, 3}

s.update([4, 5, 6])

print(s)
# {1, 2, 3, 4, 5, 6}


# 튜플도 추가할 수 있음

s.update((7, 8))

print(s)
# {1, 2, 3, 4, 5, 6, 7, 8}


# ------------------------------------------


# 3. remove(x)
# 지정한 요소를 삭제
# 요소가 존재하지 않으면 KeyError 발생

s = {1, 2, 3}

s.remove(2)

print(s)
# {1, 3}


# 아래 코드는 10이 없으므로 KeyError 발생
# s.remove(10)


# ------------------------------------------


# 4. discard(x)
# 지정한 요소를 삭제
# 요소가 존재하지 않아도 오류가 발생하지 않음

s = {1, 2, 3}

s.discard(2)

print(s)
# {1, 3}


# 10이 없어도 오류가 발생하지 않음

s.discard(10)

print(s)
# {1, 3}


# ------------------------------------------


# 5. pop()
# 집합에서 임의의 요소 하나를 꺼내면서 삭제
# 집합은 순서가 없으므로 어떤 값이 나올지 보장되지 않음

s = {10, 20, 30}

deleted_value = s.pop()

print(deleted_value)
# 10, 20, 30 중 하나

print(s)
# 꺼낸 요소를 제외한 나머지 집합


# 빈 집합에서 pop()을 사용하면 KeyError 발생

empty_set = set()

# empty_set.pop()


# ------------------------------------------


# 6. clear()
# 집합에 들어 있는 모든 요소를 삭제

s = {1, 2, 3}

s.clear()

print(s)
# set()
```

##### 3.2 집합 연산과 포함 여부 확인
- 집합 연산 기호 지원
    - 수학적 연산인 교집합(&), 합집합( | ), 차집합( - )을 특수 연산 기호로 직관적으로 수행하여 두 데이터 군의 공통점과 차이점을 신속하게 식별할 수 있습니다.
- in / not in 연산자의 최적화 효과 ## 키 값의 존재 여부를 확인합니다.
    - 특정 데이터가 집합에 존재하는지 대조하여 참/거짓(True/False) 결과값을 도출
    - 집합은 내부 해시 기반 구조 덕분에 리스트나 튜플에 비해 포함 여부를 확인하는 연산 속도가 압도적으로 빠르다는 장점이 있습니다.

    - set(리스트 or 튜플)

```python
a = {10,20,30,40}
b = {30,40,50,60}

print("합집합:",a|b)
print("교집합:",a&b)
print("차집합:",a-b)
print("대칭차집합:",a^b)

#리스트,튜플로 딕셔너리를 바꾼다면 각 집합 연산이 순서대로 출력 됩니다. 

a = a.set()
b = b.set()
print("합집합:",a|b)
print("교집합:",a&b)
```

#### 자료구조 간의 상호 변환과 데이터 정제
---
- 현실의 프로젝트에서는 하나의 자료구조만 고집해서 개발하지 않습니다. 
- 데이터 수집 단계에서는 리스트로 묶어둔 뒤, 중복을 제거할 때는 집합으로 변환하고, 최종 가공을 위해 다시 리스트나 딕셔너리로 변환하는 유기적인 연동 능력이 필수적입니다. 각 자료구조를 서로 교환하며 데이터를 정형화하는 변환 기술을 확인해 보겠습니다.

##### 4.1 상호 변환과 데이터 정제
- 생성 함수를 통한 변환
    - **list(), tuple(), set()** 등의 변환 내장 함수를 활용하면 시퀀스 자료구조와 집합 자료구조의 형태를 언제든 자유롭게 변환할 수 있습니다.
- 중복 제거 정제
    - 대량의 사용자 로그 데이터가 담긴 리스트에서 중복을 완전히 제거한 청정 고유 리스트를 추출하고 싶다면, **리스트 -> 집합(중복 자동 제거) -> 리스트** 형태의 파이프라인 작동 방식을 적용할 수 있습니다.
- [중요!!!] 자료형을 set으로 변환하면 중복은 제거되지만 순서가 보장되지 않습니다. 이 때문에 인덱스 접근이 불가능하여 슬라이싱이나 특정 요소 추출을 할 수 없습니다. 만약 중복을 제거한 뒤 인덱싱이나 슬라이싱 같은 가공을 해야 한다면, list() 함수를 이용해 다시 리스트로 변환해야 합니다.

```python
# 실무 데이터 정제 패턴
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]

# 1단계: 집합으로 변환하여 중복 제거
unique_set = set(raw_emails) 

# 2단계: 다시 리스트로 돌려놓아 슬라이싱 및 요소 접근을 위한 순서적 활용 보장
refined_emails = list(unique_set)
print(refined_emails)  # 출력: ['a@sparta.com', 'b@sparta.com']
```

[별첨.] 구조적 언패킹이란?

- 리스트나 튜플 등에 묶여 있는 여러 값을 자료구조의 순서에 맞춰 여러 변수에 한 번에 나눠 담는 문법
- 여러 값을 각각 할당할 수 없어서 쓰는 것이 아니라, 데이터의 구조를 드러내면서 여러 값을 더 간결하고 안전하게 변수에 나눠 담기 위해 사용하는 문법

왜 언패킹을 할까?
- 이유 1: 코드가 짧고 의미가 명확해짐, 언패킹을 사용하지 않으면 각 위치를 일일이 지정해야 함
- 이유 2: 인덱스를 반복해서 적는 실수를 줄여준다. 언패킹은 순서대로 자동 대응됨
- 이유 3: 함수가 반환한 여러 값을 편하게 받을 수 있다
- 이유 4: 반복문에서 키와 값을 동시에 꺼낼 수 있다
- 이유 5: 필요 없는 값이나 나머지를 쉽게 처리할 수 있다



```python
product = ("셔츠", 30000, "상의")
name, _, category = product
name, price, category = product #언패킹 예시 -> 언패킹을 하면, 한줄로 출력할 수 있고, 구조를 보기 좋다!!!!!

#언패킹 출력물
name = "셔츠"
price = 30000
category = "상의"
```
