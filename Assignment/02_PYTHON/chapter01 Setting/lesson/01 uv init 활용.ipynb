### 1. 시스템 전역 설치의 문제와 가상환경 격리의 필요성
---
<aside>

처음에는 하나의 프로그램만 만드니 아무런 문제가 없어 보입니다.
하지만 실무에서 여러 프로젝트를 진행하다 보면, A 프로젝트는 최신 버전의 패키지가 필요하고 B 프로젝트는 과거 버전에서만 작동하는 상충되는 상황을 마주하게 됩니다.

이때 전역 공간에 패키지를 새로 설치했다가 기존에 잘 돌아가던 다른 프로그램들이 도미노처럼 망가지는 '버전 지옥(Dependency Hell)'을 경험하게 되는데요.

실제로 수많은 입문자가 소스 코드 오류가 아니라, 바로 이 **환경 충돌** 때문에 첫날부터 지쳐 개발을 포기하곤 합니다.

우리는 편리해 보이는 시스템 전역 설치가 추후 배포와 협업에서 왜 시한폭탄이 되는지 그 이유를 명확히 확인하고, 프로젝트마다 독립된 가상환경을 격리해야 하는 필요성에 대해 제대로 짚고 넘어갈 예정입니다.

#### 1.1 시스템 전역 설치의 문제
- 파이썬 입문자의 흔한 실수: 시스템 전역 설치
    - 파이썬을 처음 배울 때 내 컴퓨터 전체(시스템 전역)에 환경을 구축하는 방식입니다.
    - 처음에는 하나의 프로그램만 만드므로 아무런 문제 없이 작동하는 것처럼 보입니다.
    - 많은 입문자가 소스 코드의 문법 오류가 아니라, 첫날부터 발생하는 **환경 충돌과 오류** 때문에 지쳐서 개발을 포기하곤 합니다.

- 전역 설치가 가져오는 시한폭탄: '버전 지옥(Dependency Hell)’
    - 여러 프로젝트를 동시에 진행할 때 발생하는 문제
    예) A 프로젝트는 최신 버전의 패키지가 필요하지만, B 프로젝트는 과거 버전에서만 정상 작동하는 상황
    - 전역 공간에 새 패키지를 설치하는 순간, 기존에 잘 돌아가던 다른 프로그램들이 도미노처럼 망가지는 현상을 겪게 됩니다.

#### 1.2 가상환경 격리의 필요성
- '가족과 사는 집' vs '나만의 자취방’
    - **시스템 전역 (가족과 다 함께 사는 집):** 거실 리모컨 하나, 냉장고 반찬 하나 바꾸는 것도 서로 눈치가 보이고 부딪히기 마련입니다. 내 마음대로 환경을 바꾸거나 실험하기가 어렵습니다.
    - **가상환경 (독립된 나만의 자취방):** 최신 스마트 가전으로 채우든 레트로 가구로 채우든 그 누구도 간섭하지 않습니다. 오롯이 나만의 규칙과 타이밍에 맞춰 공간을 자유롭게 제어할 수 있습니다.
- 가상환경 격리가 주는 핵심 가치
    - A 프로젝트 폴더에는 A가 필요한 패키지만, B 프로젝트 폴더에는 B가 필요로 하는 패키지만 독립적으로 담아둡니다. 컴퓨터 안에서 두 프로그램을 동시에 켜두고 작업해도 절대 충돌하거나 간섭하지 않습니다.
    - 가상환경을 사용하는 것은 단순히 폴더를 깔끔하게 정리하는 것을 넘어, 내가 만든 프로그램이 1년 뒤에도, 혹은 동료의 컴퓨터나 클라우드 서버로 옮겨갔을 때도 똑같이 안전하게 돌아가도록 보장하는 장치입니다.


### 2. 파이썬 개발 환경 구축
---
그럼 가상환경이 왜 필요한지 알았으니, 이제 진짜 안전하게 격리된 나만의 개발 공간을 구축해 볼까요?

본격적인 실습을 시작하기 전에, 기존 파이썬 생태계에서는 가상환경을 어떻게 만들어왔는지 그 방식을 가볍게 알아보려고 합니다. 기존 방식이 가졌던 숨겨진 문제점들은 무엇이었는지, 그리고 이를 완벽하게 해결해 주는 모던 툴체인 uv를 우리가 왜 첫날부터 도입해야 하는지 그 필연적인 이유를 명확히 짚고 넘어가겠습니다.

#### 2.1 기존 가상환경 제어 방식의 한계
- 도구의 파편화로 인한 복잡성 증가
    - 가상환경 하나를 쓰기 위해 pyenv(버전관리), venv(공간 격리), pip(패키지 설치), requirements.txt(명세 기록)등 너무 많은 도구를 제각각 따로 배워서 조합해야 했습니다.
    - 코드를 작성하기도 전에 복잡한 환경 설정 단계에서 많은 에너지를 소모하게 됩니다.

- 패키지 누적에 따른 속도 저하
    - 표준 패키지 관리자인 pip는 프로젝트 규모가 커지고 라이브러리가 늘어날수록 다운로드와 의존성 분석 속도가 급격히 느려집니다.

- 수동 관리로 인한 버전 불일치
    - A 프로젝트 폴더에는 A가 필요한 패키지만, B 프로젝트 폴더에는 B가 필요로 하는 패키지만 독립적으로 담아둡니다. 컴퓨터 안에서 두 프로그램을 동시에 켜두고 작업해도 충돌하거나 간섭하지 않습니다.
    - 가상환경을 사용하는 것은 단순히 폴더를 깔끔하게 정리하는 것을 넘어, 내가 만든 프로그램이 1년 뒤에도, 혹은 동료의 컴퓨터나 클라우드 서버로 옮겨갔을 때도 똑같이 안전하게 돌아가도록 보장하는 장치입니다.

#### 2.2 모던 툴체인 uv를 사용하는 이유
- Rust 기반의 압도적 초고속 매커니즘
    - 시스템 프로그래밍 언어인 Rust로 작성되어 기존 파이썬 도구들보다 최소 10배에서 최대 100배 이상 빠른 속도를 자랑합니다. 라이브러리를 설치할 때  지루한 대기 시간을 획기적으로 줄여줍니다.

- 자동화된 파이썬 버전 관리
    - 내 컴퓨터에 파이썬이 미리 깔려있지 않아도 괜찮습니다. 프로젝트 명세에 필요한 정확한 파이썬 버전을 uv가 알아서 다운로드해 가상환경에 자동으로 설치해 줍니다.

- 단 하나의 도구로 완벽한 통합 제어
    - 가상환경 생성, 파이썬 버전 관리, 패키지 설치 및 고정(Lock)까지 uv 하나로 모두 처리합니다. 복잡한 환경 설정 없이 오직 파이썬 코드 작성에만 집중할 수 있도록 쾌적한 환경을 제공합니다.

#### 2.4 uv init을 통한 프로젝트 초기화
- 실습용 폴더 생성 및 이동

- 프로젝트 초기화 명령어 실행

```python
# 이동한 폴더에서 프로젝트의 기본 뼈대를 구성하기 위해 다음 명령어를 실행합니다.
uv init
# 명령어가 정상적으로 수행되면 터미널 화면에 Initialized project ‘first-project’ 라는 완료 메세지가 출력됩니다.
## 이 단계에서는 파이썬이 바로 다운로드되거나 가상환경(.venv)이 생성되지 않고 오직 프로젝트 환경설정을 위한 기초 파일들만 만들어집니다.
```

- uv init 실행 시 자동으로 생성되는 파일 구성
    - 명령어 실행 후 폴더 내부에 생성되는 핵심 파일 구조는 다음과 같습니다.
        - `pyproject.toml`: 프로젝트 이름, 버전, 라이브러리 의존성을 선언적으로 기록하는 모던 파이썬 표준 명세서
        - `.python-version`: 이 프로젝트에서 사용할 타깃 파이썬 버전을 명시해 두는 텍스트 파일
        - `main.py`: 기본 샘플 코드가 담겨 있는 첫 번째 파이썬 실행 파일
- 실제 파이썬 자동 다운로드와 가상환경 구축 시점
    - 초기화 이후, 처음으로 프로젝트를 구동하거나 패키지를 추가하는 명령어를 실행하는 순간 동작합니다.
        예) uv run main.py 또는 uv add 라이브러리명
    - 이때 시스템에 명세서와 일치하는 파이썬이 없다면, uv가 자동으로 다운로드하여 프로젝트 루트 내 .venv 폴더를 만들고 저장합니다.

#### 2.5 특정 파이썬 버전 지정하기
- —python 플래그를 이용한 버전 명시
    - uv init 명령어를 실행할 때 뒤에 —-python 플래그와 원하는 버전을 명시하면, 시스템 기본 파이썬 대신 지정한 버전으로 프로젝트가 생성됩니다.

```python
uv init --python 3.13
# 이 명령어를 사용하면 자동 생성되는 .python-version 파일에 3.13이 기록되고, pyproject.toml 파일의 프로젝트 명세(requires-python)에도 해당 버전 제약 조건이 자동으로 반영됩니다.
```

- 이미 생성된 프로젝트의 버전을 변경하는 방법
    - 이미 프로젝트를 생성한 이후에 파이썬 버전을 바꾸고 싶다면 다음 명령어를 사용해 타깃 버전을 변경할 수 있습니다.

```python
uv python pin 3.12 # 현재 프로젝트의 파이썬 버전을 3.12로 변경
uv sync
```

### 3. pyproject.toml 명세서 표준 구조와 메타데이터의 이해
---
- 과거에는 패키지 목록을 requirements.txt에 적고, 프로젝트 설정은 setup.py나 setup.cfg 등 여러 파일에 파편화하여 기록하곤 했습니다. 하지만 현대 파이썬 생태계는 이 단 하나의 파일로 모든 설정을 통합하는 표준 규격을 채택했습니다.

- pyproject.toml 파일이 어떤 구조로 이루어져 있고, 왜 선언적 환경 관리가 강력한 무기가 되는지 알아보겠습니다.

#### 3.1 pyproject.toml
- 기본 구조
    - **[project]**
    파이썬 공식 표준 규격(PEP 621)을 따르는 프로젝트 고유의 메타데이터를 선언하는 가장 핵심적인 영역입니다.
    - **name 및 version**
    프로젝트의 이름과 현재 버전을 정의합니다. 추후 이 프로젝트를 패키지 형태로 배포하거나 라이브러리화할 때 식별자 역할을 합니다.
    - **requires-python = ">=3.13”**
        - 이 프로젝트가 정상적으로 동작하기 위해 필요한 **최소 파이썬 버전**을 지정합니다.
        - uv는 이 옵션을 확인하고 가상환경을 구성할 때 명세에 맞는 올바른 파이썬 인터프리터를 매핑해 줍니다.
    - **dependencies = []**
        - 프로젝트가 의존하는 외부 라이브러리(예 - requests, fastapi 등) 목록이 기록되는 배열입니다.
        - 지금은 비어있지만, 추후 uv add 명령어로 패키지를 설치하면 이 자리에 사용자가 요청한 라이브러리와 버전 제약 조건이 자동으로 기록됩니다.

```python
# uv init 명령어로 프로젝트를 초기화하면 생성되는 가장 표준적이고 직관적인 명세서 구조입니다.
[project]
name = "first-project"
version = "0.1.0"
description = "Add your description here"
readme = "README.md"
requires-python = ">=3.13"
dependencies = []
```

#### 3.2 선언적 환경 관리의 이점
- 단일 소스 관리를 통한 파편화 방지
    - 빌드 설정, 의존성 패키지, 개발 도구 설정을 여러 텍스트 파일에 나누어 적지 않고 pyproject.toml 파일 하나에서 중앙 집중식으로 관리하므로 프로젝트 구조가 명확해집니다.
- 어디서나 동일한 실행 환경 보장
    - "내가 개발할 때 사용한 패키지와 버전 구조"를 명확하게 문서화하여 선언해 두기 때문에, 동료 개발자가 프로젝트를 이어받거나 배포 서버에 올릴 때도 토씨 하나 틀리지 않고 100% 동일한 실행 환경을 재현해 낼 수 있습니다.
- 패키지 간의 뛰어난 호환성
    - uv 뿐만 아니라 Poetry, Hatch, Flit 등 현대적인 파이썬 빌드/패키지 관리 도구들이 모두 이 표준 규격을 해석할 수 있으므로, 향후 툴체인을 전환하거나 확장할 때도 호환성 리스크가 전혀 없습니다.

- uv가 구성해 준 첫 번째 프로젝트의 중심 파일(main.py)을 직접 실행해 보면서, 우리가 구축한 독립된 개발 환경이 안전하고 정상적으로 살아 움직이는지 최종 검증해 보겠습니다.

```python
# 앞서 uv init을 실행했을 때 폴더 내부에 자동으로 만들어진 main.py 파일을 열어보면 다음과 같은 코드가 기본적으로 작성되어 있습니다.
def main():
    print("Hello from first-project!")


if __name__ == "__main__":
    main()

# 이 코드는 단순히 한 줄짜리 명령이 아니라, 파이썬 진입점(main) 함수 구조를 갖춘 가장 정석적인 프로젝트 실행 파일의 형태를 띠고 있습니다.
```

#### 4.2 프로젝트 실행과 가상환경의 자동 빌드
- uv run 명령어 실행

```python
# 터미널 창에 아래 명령어를 입력하여 우리의 첫 파이썬 프로젝트를 구동합니다.
uv run main.py
```

- 첫 실행 시 터미널 뒤에서 일어나는 마법
    - 명령어를 입력하고 엔터를 누르는 순간, uv는 프로젝트 루트 폴더를 빠르게 스캔합니다.
    - 만약 이 단계에서 내 컴퓨터에 지정된 버전의 파이썬이 없거나 가상환경 폴더(.venv)가 존재하지 않는다면, uv가 해당 파이썬 인터프리터를 자동으로 다운로드하고 프로젝트 전용 격리 공간을 그 자리에서 즉시 구축합니다.
    - 복잡한 설치 경로 지정이나 환경 변수 세팅을 전혀 신경 쓰지 않아도, 명령어 단 한 줄로 이 모든 프로젝트 인프라가 원스톱으로 처리됩니다.

- **시스템 전역 오염:** 한 공간에 패키지를 몰아 깔면 버전이 충돌하는 시한폭탄이 됩니다.
- **가상환경 격리:** 상호 간섭이 없는 '나만의 독립된 자취방'을 만드는 확실한 보험입니다.

- **pyproject.toml:** 어디서나 동일한 실행 환경을 보장하는 파이썬 표준 개발 계약서입니다.
- **uv run:** 명령어 단 한 줄로 파이썬 다운로드와 가상환경 빌드, 스크립트 실행까지 원스톱으로 처리하는 진입점입니다.

| **패키지 관리 도구** | `Package Manager` | 외부 개발자들이 만들어 둔 코드 묶음인 패키지를 다운로드하거나 삭제해 주는 **설치 관리자**입니다. |
| **uv** | `uv` | 프로젝트 생성, 파이썬 버전 관리, 패키지 설치 등을 빠르고 깔끔하게 처리해 주는 **모던 패키지 관리 도구**입니다. |
| **pyproject.toml** | `pyproject.toml` | 프로젝트 정보, 사용할 파이썬 버전, 필요한 외부 패키지 등을 기록하는 **프로젝트 설계 명세서 파일**입니다. |
| **uv.lock** | `uv.lock` | 프로젝트에서 사용하는 패키지와 관련 의존성의 정확한 버전을 기록한 **버전 잠금 파일**입니다. |
