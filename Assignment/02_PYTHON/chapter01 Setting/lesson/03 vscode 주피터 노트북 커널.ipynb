# 1. REPL과 스크립트의 결합: 주피터 노트북

주피터 노트북은 데이터 분석, 데이터 사이언스, 머신러닝, 인공지능 개발 분야에서 널리 사용되는 도구입니다.

주피터 노트북이 널리 사용되는 이유는 REPL의 장점인 **즉각적인 피드백**과 스크립트 파일의 장점인 **영구 저장 기능**이 결합되어 있기 때문입니다.

이번 장에서는 주피터 노트북의 개념을 이해하고 현대 파이썬 프로젝트에서 주피터 노트북이 널리 사용되는 이유를 알아보겠습니다.

---

## 1.1 주피터 노트북의 도입 배경

### 기존 개발 방식의 한계

기존에는 파이썬 코드를 실행할 때 주로 다음 두 가지 방식을 사용했습니다.

| 방식 | 장점 | 단점 |
|---|---|---|
| 스크립트 방식 | 작성한 코드를 파일로 저장할 수 있다. | 코드 일부만 수정해도 파일 전체를 다시 실행해야 할 수 있다. |
| REPL 방식 | 코드를 한 줄씩 입력하고 결과를 바로 확인할 수 있다. | 터미널을 종료하면 작성한 코드와 실행 기록이 사라질 수 있다. |

### 데이터 과학과 AI 실험의 필요성

대용량 데이터를 다루는 데이터 과학과 AI 분야에서는 데이터를 메모리에 한 번 불러온 후 다양한 조건과 모델을 반복적으로 실험하고 수정하는 과정이 필요합니다.

스크립트 파일을 처음부터 다시 실행하면 다음 작업이 반복될 수 있습니다.

1. 데이터 파일을 다시 불러옵니다.
2. 데이터를 다시 전처리합니다.
3. 모델을 다시 생성합니다.
4. 결과를 다시 확인합니다.

주피터 노트북은 필요한 코드 영역만 선택하여 실행할 수 있기 때문에 이러한 반복 작업을 줄일 수 있습니다.

### 셀 단위의 동적 실행

주피터 노트북은 코드를 **셀(Cell)**이라는 독립된 블록으로 나누어 관리합니다.

따라서 전체 코드를 한 번에 실행하지 않고 필요한 셀만 원하는 시점에 실행할 수 있습니다.

```python
numbers = [10, 20, 30, 40, 50]
```

다른 셀에서 앞서 만든 변수를 사용할 수 있습니다.

```python
total = sum(numbers)
print(total)
```

### 실행 결과

```text
150
```

이처럼 데이터를 한 번 메모리에 저장해 놓고 필요한 연산만 반복해서 실행할 수 있습니다.

## 1.2 시각적 문서화의 장점

### 실행 결과 저장

주피터 노트북은 코드와 함께 다음과 같은 실행 결과를 문서 안에 저장할 수 있습니다.

- 텍스트 출력 결과
- 계산 결과
- 데이터 표
- 시각화 그래프
- 이미지
- 오류 메시지

### 실행 결과

```text
김지현님은 AI Agent 엔지니어 부트캠프 과정을 학습하고 있습니다.
```

### 살아 있는 문서 제작

주피터 노트북에서는 파이썬 코드 사이에 제목, 설명, 학습 기록 등을 작성할 수 있습니다.

하나의 파일 안에서 다음 내용을 함께 관리할 수 있습니다.

- 개념 설명
- 실행 가능한 파이썬 코드
- 실행 결과
- 표와 그래프
- 실험 과정
- 학습 내용 요약

즉, 주피터 노트북은 단순한 코드 파일이 아니라 코드와 설명, 실행 결과가 함께 들어 있는 **실행 가능한 문서**입니다.

# 2. VS Code Jupyter 확장 플러그인 설치

VS Code는 확장 플러그인을 설치하여 다양한 개발 기능을 추가할 수 있습니다.

VS Code에 Python과 Jupyter 확장을 설치하면 별도의 주피터 프로그램으로 이동하지 않고 VS Code 안에서 노트북 파일을 작성하고 실행할 수 있습니다.

---

## 2.1 확장 플러그인 설치

### 확장 플러그인 탭 열기

VS Code 왼쪽 사이드바에서 블록이 결합된 모양의 **Extensions** 아이콘을 클릭합니다.

다음 단축키를 사용할 수도 있습니다.

```text
Ctrl + Shift + X
```

### Python 확장 설치

1. Extensions 검색창에 `Python`을 입력합니다.
2. 게시자가 Microsoft인지 확인합니다.
3. **Install** 버튼을 클릭합니다.

Python 확장은 다음 기능을 제공합니다.

- 파이썬 문법 검사
- 코드 자동 완성
- 오류 표시

1. Extensions 검색창에 `Jupyter`를 입력합니다.
2. 게시자가 Microsoft인지 확인합니다.
3. **Install** 버튼을 클릭합니다.

Jupyter 확장은 다음 기능을 제공합니다.

- `.ipynb` 파일 열기
- 코드 셀과 마크다운 셀 작성
- 셀 단위 실행
- 실행 결과 출력
- 주피터 커널 선택
- 노트북 실행 상태 관리

---

## 2.2 주피터 통신 엔진 `ipykernel` 설치

### 주피터와 파이썬을 연결하는 엔진

주피터 노트북이 파이썬 코드를 실행하려면 파이썬 인터프리터와 통신해야 합니다.

이때 주피터의 실행 명령을 파이썬 인터프리터에 전달해 주는 핵심 패키지가 `ipykernel`입니다.

```text
Jupyter Notebook
        ↓
    ipykernel
        ↓
Python Interpreter
```

### 가상환경마다 설치해야 하는 이유

`uv init`으로 생성한 프로젝트의 `.venv`는 프로젝트 전용 파이썬 인터프리터와 패키지를 관리하는 독립된 가상환경입니다.

주피터 노트북이 운영체제의 전역 파이썬이 아니라 프로젝트의 `.venv`에 설치된 파이썬과 통신하려면 해당 프로젝트에 `ipykernel`이 설치되어 있어야 합니다.

### 프로젝트 폴더로 이동

PowerShell 또는 VS Code 터미널에서 프로젝트 폴더로 이동합니다.

```powershell
cd C:\Users\사용자이름\프로젝트폴더
```

예시는 다음과 같습니다.

```powershell
cd C:\Users\jihyu\student_data_project
```

### `ipykernel` 설치

프로젝트 폴더의 터미널에서 다음 명령어를 실행합니다.

```powershell
uv add ipykernel
```

### 설치 확인

다음 명령어로 설치된 패키지 목록을 확인할 수 있습니다.

```powershell
uv pip list
```

목록에서 `ipykernel`이 표시되는지 확인합니다.

### 의존성 복원

다른 사람이 프로젝트를 내려받았거나 기존 가상환경을 다시 구성할 때는 다음 명령어를 실행합니다.

```powershell
uv sync
```

`uv add ipykernel`을 실행하면 `pyproject.toml`의 의존성 목록에 `ipykernel`이 기록됩니다.

따라서 다른 개발자도 프로젝트 파일을 받은 후 `uv sync`를 실행하면 같은 패키지 환경을 구성할 수 있습니다.

### 명령어 비교

| 명령어 | 기능 |
|---|---|
| `uv add ipykernel` | 프로젝트 의존성에 `ipykernel`을 추가하고 설치합니다. |
| `uv sync` | `pyproject.toml`과 `uv.lock`을 기준으로 가상환경을 복원합니다. |
| `uv pip list` | 현재 가상환경에 설치된 패키지 목록을 확인합니다. |

# 3. 노트북 파일 생성 및 가상환경 커널 연결

주피터 노트북에서 코드를 실행하려면 어떤 파이썬 환경을 사용할 것인지 지정해야 합니다.

주피터 노트북의 코드를 실제로 실행하는 엔진을 **커널(Kernel)**이라고 합니다.

프로젝트에서 만든 `.venv` 가상환경을 커널로 연결하지 않으면 운영체제에 설치된 전역 파이썬이 선택될 수 있습니다.

이 경우 프로젝트에 설치한 패키지를 찾지 못하거나 서로 다른 패키지 버전이 충돌할 수 있습니다.

---

## 3.1 `.ipynb` 파일 생성하기

### 새로운 파일 생성

다음 방법 중 하나로 새 파일을 만듭니다.

#### 방법 1: 상단 메뉴 사용

```text
File → New File
```

#### 방법 2: 탐색기 사용

VS Code 탐색기의 빈 공간을 마우스 오른쪽 버튼으로 클릭한 후 **New File**을 선택합니다.

### 파일 이름 작성

파일 이름을 다음과 같이 작성합니다.

```text
chapter01_03.ipynb
```

파일 이름 뒤에 반드시 `.ipynb` 확장자를 입력해야 합니다.

```text
파일 이름.ipynb
```

`.ipynb` 확장자를 사용하면 VS Code가 해당 파일을 일반 텍스트 파일이 아니라 주피터 노트북 파일로 인식합니다.

---

## 3.2 주피터 커널 매핑

### 커널 선택하기

1. 생성한 `.ipynb` 파일을 엽니다.
2. 노트북 화면 오른쪽 위에 있는 **Select Kernel**을 클릭합니다.
3. **Python Environments...**를 선택합니다.
4. 현재 프로젝트의 `.venv` 가상환경을 선택합니다.

### Windows 가상환경 경로 예시

```text
프로젝트폴더\.venv\Scripts\python.exe
```

예시는 다음과 같습니다.

```text
C:\Users\jihyu\student_data_project\.venv\Scripts\python.exe
```

프로젝트 폴더 안의 `.venv` 경로인지 반드시 확인해야 합니다.

### 올바른 커널 확인 코드

노트북의 코드 셀에 다음 코드를 입력하고 실행합니다.

```python
import sys

print(sys.executable)
```

### 정상적인 실행 결과 예시

```text
C:\Users\jihyu\student_data_project\.venv\Scripts\python.exe
```

출력된 경로에 현재 프로젝트 폴더와 `.venv`가 포함되어 있으면 프로젝트 가상환경이 정상적으로 연결된 것입니다.

### 파이썬 버전 확인

```python
import sys

print(sys.version)
```

또는 다음 코드를 사용할 수 있습니다.

```python
from platform import python_version

print(python_version())
```

### 커널 연결 확인 코드

```python
import sys
from platform import python_version

print("파이썬 실행 경로:", sys.executable)
print("파이썬 버전:", python_version())
```

### 커널 연결이 중요한 이유

올바른 가상환경 커널을 선택하면 다음과 같은 장점이 있습니다.

- 프로젝트별 패키지를 독립적으로 관리할 수 있습니다.
- 다른 프로젝트의 패키지와 충돌하지 않습니다.
- 프로젝트에서 설치한 라이브러리를 노트북에서 사용할 수 있습니다.
- 팀원이 동일한 실행 환경을 구성할 수 있습니다.
- 패키지 버전 차이로 발생하는 오류를 줄일 수 있습니다.

# 4. 코드 셀과 마크다운 셀

주피터 노트북은 내용을 셀 단위로 작성합니다.

각 셀은 용도에 따라 코드 셀과 마크다운 셀로 구분할 수 있습니다.

---

## 4.1 코드 셀

코드 셀은 실제 파이썬 코드를 작성하고 실행하는 영역입니다.

파이썬 인터프리터가 코드 셀의 내용을 해석하고 실행 결과를 셀 아래에 출력합니다.

### 코드 셀 예제

```python
message = "주피터 노트북 코드 셀입니다."

print(message)
```

### 실행 결과

```text
주피터 노트북 코드 셀입니다.
```

계산 결과도 확인할 수 있습니다.

```python
price = 10000
quantity = 3
total_price = price * quantity

print(f"총금액: {total_price}원")
```

### 실행 결과

```text
총금액: 30000원
```

---

## 4.2 마크다운 셀

마크다운 셀은 실제 파이썬 코드를 실행하는 영역이 아닙니다.

제목, 설명, 학습 내용, 표, 링크 등을 작성하는 문서화 영역입니다.

### 마크다운 셀 작성 예시

```markdown
# 가장 큰 제목

## 두 번째 제목

### 세 번째 제목

- 첫 번째 항목
- 두 번째 항목

**굵은 글씨**

`인라인 코드`
```

마크다운 셀을 실행하면 마크다운 문법이 적용된 문서 형태로 표시됩니다.

---

## 4.3 코드 셀과 마크다운 셀 비교

| 구분 | 코드 셀 | 마크다운 셀 |
|---|---|---|
| 영어 표기 | Code Cell | Markdown Cell |
| 주요 역할 | 파이썬 코드 작성 및 실행 | 제목과 설명 작성 |
| 실행 결과 | 코드 실행 결과 출력 | 문서 형태로 변환 |
| 주요 내용 | 변수, 함수, 클래스, 연산 | 개념 설명, 학습 기록, 표 |
| 변환 단축키 | `Y` | `M` |

# 5. 주피터 노트북 핵심 단축키

주피터 노트북에는 셀을 빠르게 생성하고 실행할 수 있는 단축키가 있습니다.

단축키를 사용하려면 먼저 셀의 상태를 확인해야 합니다.

---

## 5.1 편집 모드와 명령 모드

### 편집 모드(Edit Mode)

셀 내부에 커서가 있고 코드나 글을 입력할 수 있는 상태입니다.

편집 모드로 들어가려면 셀을 더블 클릭하거나 `Enter` 키를 누릅니다.

```text
Enter → 편집 모드
```

### 명령 모드(Command Mode)

셀 내부의 내용을 입력하는 것이 아니라 셀 자체를 추가하거나 변환하는 상태입니다.

명령 모드로 전환하려면 `Esc` 키를 누릅니다.

```text
Esc → 명령 모드
```

`A`, `B`, `M`, `Y`와 같은 셀 제어 단축키는 일반적으로 명령 모드에서 사용합니다.

# 6. 실행 번호 `In [n]`과 비순차적 실행

일반적인 파이썬 스크립트는 위에서 아래로 순서대로 실행됩니다.

그러나 주피터 노트북에서는 원하는 셀을 선택하여 자유롭게 실행할 수 있습니다.

이러한 실행 방식은 편리하지만 실행 순서가 뒤섞이면 메모리에 저장된 변수 상태 때문에 예상하지 못한 오류가 발생할 수 있습니다.

---

## 6.1 실행 번호 `In [n]`

코드 셀을 실행하면 셀 왼쪽에 실행 번호가 표시됩니다.

```text
In [1]
In [2]
In [3]
```

이 숫자는 노트북 위에서부터 몇 번째 셀인지를 뜻하지 않습니다.

현재 커널 세션에서 코드 셀이 실제로 실행된 순서를 의미합니다.

예를 들어 아래쪽 셀을 가장 먼저 실행하면 아래쪽 셀에 `In [1]`이 표시될 수 있습니다.

### 실행 번호의 의미

| 표시 | 의미 |
|---|---|
| `In [ ]` | 아직 실행하지 않은 셀 |
| `In [1]` | 현재 커널에서 첫 번째로 실행한 셀 |
| `In [3]` | 현재 커널에서 세 번째로 실행한 셀 |
| `In [*]` | 현재 실행 중인 셀 |

---

## 6.2 셀 간 메모리 공유

하나의 주피터 커널에 연결된 코드 셀은 같은 메모리를 공유합니다.

첫 번째 셀에서 만든 변수를 다른 셀에서 사용할 수 있습니다.

### 첫 번째 셀

```python
price = 10000
```

### 두 번째 셀

```python
quantity = 3
```

### 세 번째 셀

```python
total_price = price * quantity

print(total_price)
```

### 실행 결과

```text
30000
```

세 번째 셀에는 `price`와 `quantity`를 선언하는 코드가 없습니다.

하지만 앞에서 실행한 셀의 변수가 커널 메모리에 저장되어 있기 때문에 사용할 수 있습니다.

---

## 6.3 비순차적 실행의 문제

다음과 같은 셀이 있다고 가정합니다.

### 첫 번째 셀

```python
price = 10000
```

### 두 번째 셀

```python
quantity = 3
```

### 세 번째 셀

```python
total_price = price * quantity

print(total_price)
```

위에서 아래로 실행하면 결과는 `30000`입니다.

```text
30000
```

그런데 첫 번째 셀의 값을 다음과 같이 변경한 후 첫 번째 셀만 다시 실행하면 메모리 속 `price` 값이 변경됩니다.

```python
price = 20000
```

세 번째 셀을 다시 실행하면 결과도 달라집니다.

```python
total_price = price * quantity

print(total_price)
```

### 실행 결과

```text
60000
```

노트북 화면에 표시된 셀의 위치는 그대로이지만 메모리에 저장된 변수의 값은 마지막으로 실행한 코드에 따라 달라집니다.

이것이 비순차적 실행으로 인해 발생하는 상태 변경입니다.

---

## 6.4 `NameError` 발생 예시

변수를 선언하는 셀을 실행하지 않고 변수를 사용하는 셀부터 실행하면 오류가 발생합니다.

```python
print(user_name)
```

### 오류 예시

```text
NameError: name 'user_name' is not defined
```

`user_name` 변수를 먼저 생성하지 않았기 때문에 발생하는 오류입니다.

먼저 다음 셀을 실행해야 합니다.

```python
user_name = "김지현"
```

그 후 아래 셀을 실행합니다.

```python
print(user_name)
```

### 실행 결과

```text
김지현
```

---

## 6.5 비순차적 실행 문제 해결

실행 순서가 꼬여 현재 메모리 상태를 확인하기 어려운 경우 커널을 다시 시작합니다.

### 해결 순서

1. 노트북 상단의 **Restart** 버튼을 클릭합니다.
2. 커널 메모리에 저장된 변수를 초기화합니다.
3. 첫 번째 셀부터 마지막 셀까지 위에서 아래로 다시 실행합니다.
4. 필요한 경우 **Run All**을 실행합니다.

```text
Restart → Run All
```

### Restart의 의미

Restart를 실행하면 다음 데이터가 초기화됩니다.

- 변수
- 함수
- 클래스
- 불러온 모듈
- 메모리에 저장된 계산 결과
- 실행 번호

단, 노트북 셀에 작성한 코드와 마크다운 내용이 삭제되는 것은 아닙니다.

| `.venv` | Virtual Environment Directory | 프로젝트의 가상환경 파일과 파이썬 인터프리터가 저장되는 폴더입니다. |
