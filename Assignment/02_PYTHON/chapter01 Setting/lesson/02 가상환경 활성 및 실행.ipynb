# 1장 2강 : 프로젝트 가상환경 활성화와 기본 실행 방식
## 학습 목표
- 프로젝트 가상환경을 활성화하여 내부 인터프리터의 구동 상태를 검증할 수 있다.
- 파이썬의 전통적인 두 가지 실행 방식(REPL, 스크립트)의 특징과 차이점을 설명할 수 있다.

### 1. 운영체제별 가상환경 활성화
---
- 가상환경 폴더를 만드는 것만으로는 부족합니다. 컴퓨터에게 "지금부터는 전역 공간이 아니라, 내가 만든 이 방의 규칙을 따르겠다"고 선언하는 활성화(Activate) 단계가 필요합니다.

-  가상환경 진입 명령어

```python
# Windows (PowerShell)
.venv\Scripts\activate

# macOS / Linux
source .venv/bin/activate

## 명령어를 실행하면 터미널 프롬프트 맨 앞에 프로젝트 폴더 이름이 괄호 형태로 표시됩니다.
## 프롬프트 앞에 (프로젝트명)이 표시된다면 가상환경이 성공적으로 활성화된 것입니다.
```

### 2. 실시간으로 소통하는 REPL(대화형 셸) 구동
---
- 파이썬과 가볍게 대화하며 소통할 수 있는 첫 번째 실행 방식, REPL을 만나봅니다.

-  REPL(Read-Eval-Print Loop)의 개념
    - 코드를 한 줄 입력(Read)하면, 파이썬이 즉시 계산(Eval)하여 결과를 화면에 출력(Print)하고, 다음 입력을 기다리는 루프(Loop) 상태를 의미합니다. 마치 파이썬과 실시간으로 주고받는 DM(채팅)과 같습니다.

- REPL 실습

```python
# REPL 실행 및 사칙연산 실습
## 가상환경이 활성화된 터미널창에 python을 입력하여 대화형 셸을 엽니다. 프롬프트가 >>> 모양으로 바뀌면 성공입니다.
python

## >>> 옆에 간단한 연산식을 입력하고 엔터를 눌러 실시간 피드백을 확인합니다.
>>> 1 + 1
2
>>> 10 * 5
50

# REPL 탈출하기
## 실습을 마친 후 대화형 셸을 종료하고 원래 터미널로 돌아가려면 아래 명령어를 입력합니다.
>>> exit()
```

- REPL 방식의 특징과 한계
    - 코드를 짜는 족족 결과를 볼 수 있어 간단한 문법 테스트나 수식 검증에는 최적입니다.
    - 하지만 터미널 창을 닫는 순간 내가 작성한 모든 코드가 메모리에서 사라지므로, 코드를 영구적으로 보존하거나 자동화 프로그램을 만들기에는 부적합합니다.

```python
print("Hello World")
```

- 터미널에서 스크립트 일괄 구동하기
    - 가상환경이 활성화된 터미널창에서 아래 명령어를 실행하여 수동 저장한 파일의 전체 코드를 실행합니다. 

- VS Code 내부에서 터미널 열기
    - 마우스 조작: VS Code 상단 메뉴에서 Terminal → New Terminal을 선택합니다.
    - 단축키 조작: Windows / MacOS: Ctrl + `

### 4. REPL 방식 vs 스크립트 방식
---
| **구분** | **REPL(대화형 셸)** | **스크립트 구동 방식(`.py`)** |
|---|---|---|
| **실행 단위** | 코드를 한 줄 입력하면 즉시 실행 | 파일 내 전체 코드를 위에서 아래로 차례대로 실행 |
| **코드 영속성** | **X** — 터미널이나 세션을 종료하면 입력한 코드가 사라짐 | **O** — 하드디스크에 파일로 저장되어 계속 보존됨 |
| **주요 목적** | 가벼운 문법 테스트, 실시간 수식 검증 | 실제 기능 구현, 서비스 배포 및 자동화 시스템 개발 |

# 개념 요약 키워드
---
- **가상환경 활성화 (Activate):** 전역 환경의 간섭을 차단하고 격리 공간의 규칙을 적용하는 단계로, 프롬프트 앞 (프로젝트명)으로 진입 상태를 검증합니다.
- **REPL (대화형 셸):** Read-Eval-Print Loop의 약자로, 실시간 소통이 가능하나 세션 종료 시 코드가 소멸하는 **휘발성** 특징이 있어 가벼운 테스트에 적합합니다.
- **스크립트 구동 방식 (.py):** 소스 코드를 파일에 저장하여 영구 보존(영속성)하고, python 파일명.py 명령어로 전체 로직을 일괄 실행하는 실무 표준 방식입니다.
- **VS Code 터미널 제어:** Ctrl + ` 단축키를 활용해 에디터 내부에서 터미널을 신속하게 열고 가상환경 명령어를 실행할 수 있습니다.


# 용어 사전
---
| **용어** | **영어 표기** | **설명** |
|---|---|---|
| **활성화** | Activation | 가상환경 폴더를 만든 후, 컴퓨터에게 “지금부터 시스템 전역 공간이 아니라 내가 만든 가상환경의 규칙을 따르겠다”고 선언하고 연결하는 단계입니다. |
| **터미널 프롬프트** | Terminal Prompt | 터미널에서 명령어 입력을 기다리는 입력줄의 앞부분입니다. 가상환경이 활성화되면 일반적으로 맨 앞에 **`(프로젝트명)`**이 표시됩니다. |
| **파워셸** | PowerShell | Windows 운영체제에서 사용하는 명령어 기반 셸이자 터미널 환경입니다. |
| **REPL(대화형 셸)** | REPL (Read-Eval-Print Loop) | 코드를 입력하면 컴퓨터가 즉시 읽고 실행하여 결과를 보여주는 방식입니다. 파이썬과 대화하듯 간단한 문법이나 수식을 테스트할 때 사용합니다. |
| **휘발성** | Volatility | 터미널이나 프로그램을 종료하면 파일로 저장하지 않은 코드와 실행 상태가 메모리에서 사라지는 성질입니다. |
| **스크립트 실행 방식** | Script Execution | 코드를 `.py` 파일에 저장한 후, 파일 전체를 파이썬으로 실행하는 방식입니다. 코드가 파일로 보존되므로 실제 프로그램 개발에 사용됩니다. |
| **소스 코드** | Source Code | 개발자가 프로그램을 만들기 위해 파이썬 같은 프로그래밍 언어의 문법에 맞춰 작성한 코드입니다. |
