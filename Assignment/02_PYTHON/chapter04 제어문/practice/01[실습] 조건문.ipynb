### 필수 1 : if-else 구조를 활용한 기상 경보 알림 시스템
--- 

#### 문제 1-1 :  기온에 따른 폭염 주의보 안내

```python
current_temp = 32

if current_temp >= 30 :
    print("폭염주의보가 발령되었습니다.")

else : 
    print("쾌적한 날씨입니다.")


current_temp = 25

if current_temp >= 30 :
    print("폭염주의보가 발령되었습니다.")

else : 
    print("쾌적한 날씨입니다.")

```

### 필수 2 : if-elif-else 다중 분기를 활용한 수하물 요금 분류기
---

#### 문제 2-1 :  수하물 무게별 서비스 유형 자동 분류

```python
weight = 20 

if weight > 23 : 
    print("대형 수하물 대상입니다.추가 요금이 부과됩니다.")
elif weight > 15 : 
    print("일반 위탁 수하물 대상입니다.")
else : 
    print("무료 수하물 대상입니다.")
    
```

### 심화 1 : 논리 연산자 결합을 통한 제어 최적화
---

#### 문제 3-1 : 놀이기구 탑승 요건 동시 만족 검사 시스템

```python
age = 14
height = 135

if age >= 12 and height >=  130 : 
    print("탑승 가능합니다. 관리자 확인을 진행하십시오." )

else : 
    print("탑승 요건을 만족하지 못했습니다.") 
```
