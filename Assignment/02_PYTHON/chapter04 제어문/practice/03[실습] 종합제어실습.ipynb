### 필수 2 : 농장 과일 당도 통계 분석
---

#### 문제 2-1 :  최댓값, 최솟값 및 평균 당도 산출

### 심화 1 : 가상의 보물찾기 숫자 게임 및 무한 루프 예방
---

```python
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]

min_sweetness = fruit_sweetness[0]
max_sweetness = fruit_sweetness[0]

total_sweetness = 0

for a in fruit_sweetness : 
    if a > max_sweetness : 
        max_sweetness = a
    if a < min_sweetness : 
        min_sweetness = a 
    total_sweetness += a 

count = len(fruit_sweetness)
average = total_sweetness / count

print(f"최고 과일 당도: {max_sweetness}")
print(f"최저 과일 당도:{min_sweetness}")
print(f"과일 평균 당도 :{average}")
print(f"과일 갯수:{count}")
```

#### 문제 3-1 : 무한 루프 기반의 보물 위치 맞추기 게임

```python
import random

computer_number = random.randint(1, 100) 
try_count = 0 

while True : 
    user_input = int(input("숫자를 맞춰보세요:"))
    try_count += 1 

    if user_input < computer_number :
        print("보물은 더 큰 숫자의 방에 있습니다.")
    elif user_input > computer_number : 
        print("보물은 더 작은 숫자의 방에 있습니다.")
    else : 
        print(f"축하합니다! {try_count}회 만에 보물을 찾았습니다.")
        break
```
