### 필수 1 : 정해진 횟수의 반복과 range 함수 활용
---

#### 문제 1-1 :  무인 계산대 번호표 발행기

```python
for i in range(1,5):
    print(f"{i}는 {i}번째 고객님 입니다.")
```

### 필수 2 : 조건 기반의 반복과 상태 변경
---

#### 문제 2-1 :  스마트폰 배터리 충전 시뮬레이션

```python
a = 80

while a <= 100 : 
    a += 5
    print(f"현재 배터리:{a}")

a = int(input())
```

```python
a = int(input("배터리 잔량을 입력하세요:"))

while True : 
    a += 5

    if a < 100 : 
        print(f"현재 배터리:{a}")

    else : 
        print("충전 완료.")
        break
```

### 심화 1 : 중첩 루프를 이용한 공간 정보 순회
---

#### 문제 3-1 : 2층 건물 호실 안내판 생성

```python

for a in range(1,4) :
    for b in range(1,4) :
        print(f"{a}층 {b}호")
```
