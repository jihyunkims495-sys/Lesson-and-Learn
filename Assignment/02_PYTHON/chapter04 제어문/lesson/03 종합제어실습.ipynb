### 1. 제어문과 반복문의 구조 및 데이터 필터링

#### 1.1 중첩 조건문 배치를 통한 복잡한 데이터 필터링
- 다단계 필터링의 필요성
    - 현실 세계의 정형·비정형 데이터는 단 한 줄의 조건식으로만 분류하기 어려울 만큼 복잡하게 얽혀 있습니다. for 문을 사용하여 데이터 집합을 순회하는 내부 트랙에 if-elif-else 구문을 다단계 계층형으로 중첩 배치하면, 세부 요건을 순차적으로 필터링하는 정밀한 제어 구조를 설계할 수 있습니다.

```python
# 가상의 회원 정보 데이터 집합 (리스트 내 딕셔너리 구조)
user_database = [
    {"name": "민수", "level": "VIP", "active": True, "point": 1500},
    {"name": "영희", "level": "Gold", "active": False, "point": 800},
    {"name": "철수", "level": "VIP", "active": True, "point": 2300},
    {"name": "지영", "level": "Silver", "active": True, "point": 400},
    {"name": "성민", "level": "Gold", "active": True, "point": 1200}
]

print("--- 마케팅 대상 활성 VIP 및 Gold 회원 식별 결과 ---")

for user in user_database :
    if user["active"] : # active 된 상태일 경우, 
        if user["level"] == "VIP" : #level이 VIP 일 경우, 
            print(f"[우수 등급] {user["name"]} 회원님에게 VIP 전용 쿠폰이 발송되었습니다.") # 이거를 print! 
        elif user["level"] == "GOLD" : #level이 GOLD 일 경우,
            print(f"[일반 등급] {user["name"]} 회원님에게 GOLD 전용 쿠폰이 발송되었습니다.") # 이거를 print! 
```

#### 1.2 조건별 데이터 필터링 및 누적 연산
- 누적 변수의 초기화
    - 대량의 데이터를 순회하며 특정 조건에 부합하는 항목의 개수나 특정 수치의 합계를 산출할 때는, 반드시 반복문 트랙에 진입하기 전에 수치를 저장할 누적 변수를 0으로 선언하고 초기화해 두어야 안전합니다.
- 순차적 데이터 필터링 및 조건별 누적 연산 방식
    - 반복문 내에서 조건을 만족하는 데이터만 필터링한 뒤, 복합 대입 연산자(+=)로 누적하여 최종 통계값을 산출합니다.

```python
# 누적 연산 시, 반복문의 구조
누적할_변수 = 0 #최초에 for 반복문에 들어가기 전, 값을 초기화 해주는 것! 만약, for 반복문 하위에 누적할 변수 = 0 을 지정하면, for 반복문이 돌면서 계속 0으로 리셋됨.

for 데이터 in 전체_데이터:
    if 조건:
        누적할_변수 += 더할_값
```

```python
# 누적 변수 : 반복할 때마다 기존 값에 새로운 값을 더해 나가기
total_vip_points = 0    # 계산한 결과를 저장할 두 개의 빈 상자를 준비하는 과정
active_user_count = 0   # 아직 회원을 확인하지 않았기 때문에 둘 다 0으로 시작

for user in user_database: #회원 한명씩 꺼내기
    if user["active"]: #활성회원인지 검사하기
        active_user_count += 1  # 활성 회원 수를 세기 active_user_count = active_user_count + 1

        if user["level"] == "VIP": #활성 회원 중 VIP인지 다시 검사하기
            total_vip_points += user["point"]  # VIP 포인트 합산하기 total_vip_points = total_vip_points + user["point"]

print(f"전체 활성 회원 수: {active_user_count}명")
print(f"활성 VIP 회원의 누적 포인트 합계: {total_vip_points}점")
```

### 2. 데이터 통계 분석 실습
---

#### 2.1 반복 순회를 통한 통계 지표 산출
- 첫 번째 데이터를 기준으로 하나씩 비교하는 방식
    - 데이터의 전체 크기를 모르는 상태에서 최댓값과 최솟값을 안전하게 구하려면, 데이터의 첫 번째 요소(data[0])를 초기 기준값으로 설정하는 것이 좋습니다.
- 한 번의 반복문으로 여러 지표를 동시에 산출하는 구조
    - 첫 번째 요소부터 순회를 시작하면 첫 요소는 자기 자신과 비교되므로 값이 바뀌지 않습니다. 이후 루프를 돌며 "현재 데이터가 기존 최댓값보다 크다"는 조건이 성립하면 최댓값을 즉시 갱신합니다. 이와 동시에 모든 요소를 누적 변수에 계속 더해가며 총합을 구하는 방식으로 동작합니다.

```python
# 학생 성적 데이터
student_scores = [85, 92, 78, 60, 99, 71, 88, 65]

# 통계 분석용 변수 초기화 (첫 번째 점수를 임시 기준으로 설정)
## 아직 모든 점수를 확인하지 않았기 때문에 일단 첫 번째 학생의 점수를 임시 최고점이자 임시 최저점으로 정한 것입니다.
## 이후 다른 점수를 확인하면서 더 높거나 더 낮은 점수가 발견되면 기준을 바꿉니다.
highest_score = student_scores[0] # 리스트의 첫번째 점수인 85를 가져옵니다. 최대 점수 변수 초기화
lowest_score = student_scores[0] # 리스트의 첫번째 점수인 85를 가져옵니다. 최저 점수 변수 초기화
score_sum = 0 # 합계 변수는 0으로 시작, 모든 점수를 더한 결과를 보관할 누적 변수

for score in student_scores:  # 선언된 리스트에서 점수를 하나씩 꺼냅니다.
    if score > highest_score: # 꺼내진 점수가 "최고 점수" 보다 높은가? 기준의 최고 점수는 "85"이며 85점이랑 비교했을 때, 참 or 거짓
        highest_score = score # 꺼내진 점수가 참일 때, 최고 점수가 score로 갱신! 예를 들어, 92이면 85보다 크므로 score 값이 됩니다!!

    if score < lowest_score:  # 꺼내진 점수가 "최저 점수" 보다 낮은가? 기준의 최저 점수는 "85"이며, 85점이랑 비교했을 때, 참 or 거짓
        lowest_score = score  # 꺼내진 점수가 참일 때, 최저 점수가 score로 갱신! 예를 들어, "78"이면 기준보다 낮으니 참이어서 score 값이 된다!

   
    score_sum += score  # 합계 지표 자동 누적


total_students = len(student_scores) # len()은 데이터 안에 요소가 몇 개 있는지 알려주는 내장 함수
score_average = score_sum / total_students #평균 산출

print("--- 학급 성적 종합 통계 분석 보고서 ---")
print(f"최고 성적 점수: {highest_score}점")
print(f"최저 성적 점수: {lowest_score}점")
print(f"학급 성적 총점: {score_sum}점")
print(f"학급 평균 성적: {score_average:.2f}점")
print(f"학생 수: {total_students}명")
```

### 3.  업다운(Up-Down) 숫자 맞추기 게임
---

#### 3.1 반복문과 조건문을 결합한 Up-Down 게임 구현
- 의도된 무한 루프 설계
    - 사용자가 정답을 맞히는 시점은 사전에 예측할 수 없으므로, while True: 구문을 통해 연산이 지속적으로 작동하는 루프 환경을 조성하여 제어 흐름을 설계합니다.
- 분기 제어와 조기 종료
    - 루프 내부에서는 사용자가 입력한 정수와 컴퓨터가 생성한 난수를 if-elif-else 다중 분기문으로 비교합니다. 
    - 입력값이 난수보다 작으면 'Up', 크면 'Down'을 출력하고, 두 값이 정확히 일치하는 순간 break 문을 실행하여 무한 루프를 안전하게 탈출합니다.

[설명.] 이 게임은요...
- random.randint(시작값, 끝값) -> randint()는 시작값과 끝값을 모두 포함!!!!!
- 이 게임은, 무작위로 고른 컴퓨터의 비밀 숫자를 맞추는 게임입니다. 
- while True로 숫자를 계속 입력받고, if-elif-else로 정답과 비교한 뒤, 숫자가 같아지면 break로 반복문을 끝내는 게임입니다. 
- 현재 코드는 사용자가 숫자가 아닌 값을 입력하면 오류가 발생!!

```python
# 난수 생성을 위한 표준 라이브러리 모듈 불러오기
## import는 다른 파일이나 모듈에 만들어진 기능을 현재 코드에서 사용하기 위해 불러오는 명령어
## random은 무작위 숫자를 만들 때 사용하는 파이썬 표준 모듈
import random  

## 컴퓨터가 1~100 중 숫자 하나를 무작위로 선택
### random.randint(1, 100)은 1부터 100 사이의 정수 하나를 무작위로 만듭니다.
### 이 숫자는 사용자에게 노출되지 않습니다. 컴퓨터가 숫자를 만든 거에요.
computer_number = random.randint(1, 100)  
try_count = 0    ## 사용자가 시도한 횟수 # 누적 변수

print("★ 컴퓨터가 1부터 100 사이의 비밀 숫자를 생성했습니다. ★")

while True:   # 무한 루프(=무한 반복문) 공식!!!!
    
    user_input = input("컴퓨터의 숫자를 맞춰보세요 (1~100): ") # 사용자에게 숫자를 입력 받고, (왜? : input 함수는 항상 문자열로 출력되므로)
    guess = int(user_input)   # 사용자 입력값을 정수로 변환합니다. 
    try_count += 1  #  입력할 때마다 시도 횟수 1씩 증가

    
    if guess < computer_number:   # 사용자의 숫자와 컴퓨터의 숫자를 비교했을 때,
        print("Up! 더 큰 숫자입니다.")
    elif guess > computer_number:
        print("Down! 더 작은 숫자입니다.")
    else:
        print(f"♠ 정답입니다! {try_count}회 만에 숫자를 맞추셨습니다.") # 조건 충족 시 무한 루프를 즉시 안전하게 종료하고 탈출!
        break  
```
