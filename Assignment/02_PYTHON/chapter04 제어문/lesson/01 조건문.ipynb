# 4장 1강 : 조건문
## 학습 목표
- 조건문이 프로그램의 순차적인 실행 흐름을 제어하는 원리를 설명할 수 있다.
- if, elif, else를 활용하여 다중 분기 및 중첩 조건문을 올바른 들여쓰기로 작성할 수 있다.
조건문이란?
- 판별식에 따라서 참(True)이면 참에 실행되는 코드를 정의하고, 거짓(False)이면 거짓 조건에 실행되는 코드를 정의할 수 있는 문장!
- 판별식 : 비교 연산자, 논리 연산자, 값

### 1. 단일 조건문 : if
---

'''
if 판별식 :
    판별식이 참일 때 실행되는 코드
'''

```python
# 기본 조건문 예시
number = 10
if number > 5 :
    print("5보다 크다") 
# tap키로 들여쓰기를 하지 않으면, "IndentationError:"가 뜹니다! -> 파이썬 코드의 들여쓰기 위치나 간격이 올바르지 않을 때 발생하는 오류
```

### 2. if-else 조건문

```python
# 기본 조건문 예시 : 참과 거짓
number = 12
if number == 10 : # if 조건문의 판별식이 참(True)일 때, 바로 들여쓰기 한 문장이 실행됩니다. 
    print("10입니다.")
else :   # if 조건문이 거짓(False)일 때, 하단 들여쓰기 한 문장이 실행됩니다. 
    print("10이 아닙니다.")
```

### 3. 다중 조건 분기 : if-elif-else

```python
age = input()  #값을 입력받을 때, input 함수를 씁니다. 무조건 str(문자열)로 반환합니다. 
print(age, type(age))

age = int(input())  #값을 정수int로 변환하면 변화됩니다.
print(age, type(age))
```

```python
age = int(input())  #15살일때, # 결과는 '중학생'이어야 하지만 결과는 '초등학생'이 나옵니다.
if age >= 8 :
    print("초등학생")
elif age >=14 : # elif는 다중 조건문에 사용하나,앞선 조건이 참(True)일 경우, elif는 실행되지 않아요~그래서 논리적 결과물을 추론하면서 짜야해요~
    print("중학생") 
```

```python
age = int(input()) #15살일때, # 결과는 '중학생'
if age >= 14 :
    print("중학생")
elif age >=8 : 
    print("초등학생") 
```

```python
age = int(input())
if age >= 20 : 
    print("성인")
elif age >= 17 : 
    print("고등학생")
elif age >= 14 : 
    print("중학생")
elif age >= 8 : 
    print("초등학생")
else :    #앞선 모든 조건문이 거짓(False)이면 실행됩니다.
    print("유치원생")
```

### 4. 복잡한 조건 처리 : 중첩 조건문과 논리 연산자 결합

- 일반적으로 '중첩 조건문'을 사용할 수 있으나, 가독성이 떨어져서 많이 사용하진 않습니다. (중첩 조건문 : 들여쓰기 여러번하는 거)

```python
# 예시
is_logged_in = True
has_permission = True

# 1. 들여쓰기가 깊어지는 중첩 조건문 구조 (권장 X)
if is_logged_in:
    if has_permission:
        print("관리자 대시보드 진입")

# 2. 논리 연산자를 결합한 클린 코드 최적화 구조 (권장 O)
if is_logged_in and has_permission:
    print("관리자 대시보드 진입")
```

```python
# 들여쓰기가 깊어지는 중첩 조건문 구조 예시 : 코드의 가독성이 떨어집니다. 
## 입력받은 숫자가 10 이상 100 미만인지 확인 

num =  int(input()) #입력값 : 20
if num >=10 : 
    if num < 100 : 
        print(f"{num}는 10이상 100미만입니다.")
```

```python
# 논리연산자를 통해서 코드 효율 올리기 예시_1 : 
num =  int(input()) #입력값 : 20
if num >=10 and num < 100 :  #논리연산자 "and" 사용 예시
    print(f"{num}는 10이상 100미만입니다.")
```

```python
num =  int(input()) 
if 10 =< num < 100 :  #이거는 안되! 논리 연산자가 안맞아서 인식을 못합니다!
    print(f"{num}는 10이상 100미만입니다.")
```
