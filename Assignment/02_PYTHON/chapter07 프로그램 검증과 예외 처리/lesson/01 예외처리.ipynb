## 문법이 틀림 → 구문 오류(Syntax Error)
## 문법은 맞지만 실행 중 문제 발생 → 예외(Exception)

1. 구문오류(Syntex Error), 런타임 예외(Exception)
- 구문 오류 : 문법 에러 / 파이썬 문법이 잘못되었다는 걸 알려주는 에러입니다. 
- 인터프리터가 해석하고 실행이 안되요~

2. 런타임 예외란? 
- 문법상에는 문제가 없어요~
- 논리적 오류가 있을 때
- 실행 과정 중에서 발생하는 문제에요~ 실행하는 과정에 발생하기 때문에 사전에 조심해야되요~
- 구문 에러를 제외한 나머지는 모두 "런타임 예외"입니다~

- 예외 처리를 하는 이유?
    - 서비스는 계속 유지하면서, 오류가 발생하면 개선하고 반영하고의 반복
    - 서비스의 안정적인 운영을 위해

- 예외 vs 오류?
    - 오류(Error) = 프로그램에 문제가 생긴 상황을 넓게 부르는 말 / 통제 불가능할 때
    - 예외(Exception) = 그중에서 프로그램 실행 중 발생하고, 코드로 처리할 수 있는 문제 / 통제 가능할 때
    - 즉, 오류는 문제 전체를 넓게 부르는 말이고, 예외는 그중 실행 중 발생해서 처리할 수 있는 오류라고 보면 된다.

3. try - except - else - finally

```python
# 예외 처리 구문_1
try:

    # 예외가 발생할 가능성이 있는 코드
except:

    # 예외가 발생하면 유입되는 구간, 예외에 대한 대안 처리
    # 대안 로직, 로깅 처리
```

- 로깅은 예외이든 아니든 전부 기록합니다. 로깅 처리는 원인 분석을 위해 항상 기록해야 합니다.

```python
# try - except 예시
try :
    number = int(input())
    result = 10 / number

    print(result)
    print("로깅 처리...")
except:
    print("0이 아닌 정수를 입력하세요.") # 예외를 알리는 문구 명시
    print("로깅 처리...")
```

## 모든 예외 클래스는 Exception의 자식 클래스
## 어떤 특정 예외를 특정하지 못하는 경우, Exception으로 처리가 가능합니다.
### 모든 예외 케이스들을 다 정리하지 못하기 때문에, Exception 처리를 합니다.

```python
# else
try :
    number = int(input())
    result = 10 / number

    print(result)
   
except:
    print("0이 아닌 정수를 입력하세요.") # 예외를 알리는 문구 명시

else :
    print("정상 처리")

finally:
    print("로깅 처리...") # 공통 처리
```

5. 다중 예외 처리(try - except - else - finally)

```python
# except를 여러개 쓸 수 있어요~
try :
    number = int(input())
    result = 10 / number

    print(result)
   
except ZeroDivisionError: #제로 에러가 발생할 때, 하단 내용이 프린트 됩니다. 
    print("0이 아닌 정수를 입력하세요.") 
```

```python
# except를 여러개 쓰는 예시 : (=다중 예외 처리)
try :
    number = int(input())
    result = 10 / number

    print(result)
   
except ZeroDivisionError: 
    print("0이 아닌 정수를 입력하세요.") 

except ValueError:
    print("정수만 입력하세요.")
```

```python
# 특정 예외를 그룹 지어서 튜플 형태로도 except 처리 할 수 있어요~
try :
    number = int(input())
    result = 10 / number

    print(result)
   
except (ZeroDivisionError,ValueError): 
    print("0이 아닌 정수를 입력하세요.") 
```

```python
# 발생한 문제 - 예외 객체를 기록할 수 있어요~
try :
    number = int(input())
    result = 10 / number

    print(result)
   
except ZeroDivisionError as e: # 관례적으로 "e"라는 별명을 써~ / except의 약자
    print(e) # "e"를 출력하면 "division by zero" 기록 가능
```

```python
try :
    number = int(input())
    result = 10 / number

    print(result)
   
except ZeroDivisionError: 
    print("0이 아닌 정수를 입력하세요.") 

except Exception as e:  #Exception은 가장 하단쪽 except로 정의합니다. / 맨 위에서부터 순서대로 점검하고 예외 케이스를 하단에 첨부하는 것!
    print(e)
```

6. raise(예약어)를 활용한 데이터 유효성 검증
- 예외를 발생 시키기
- 구조 : raise 예외클래스("메세지...")

```python
# 로그인 함수
def login(user_id,user_pass):
    if not (user_id == "user01"):
        print("아이디가 일치하지 않습니다.")
        return

    if not (user_pass == "1234"):
        print("비밀번호가 일치하지 않습니다.")
        return

    print("로그인 처리 로직...")
```

```python
# raise 처리
def login(user_id,user_pass):
    if user_id != "user01":
        raise ValueError("아이디가 일치하지 않습니다.") # 예외가 발생하고, 코드 실행 흐름은 중단 됩니다. 

    if user_pass != "1234":
        raise ValueError("비밀번호가 일치하지 않습니다.")

    print("로그인 처리 로직...")
```

```python
# 
def login(user_id,user_pass):
    if user_id != "user01":
        print("아이디가 일치하지 않습니다.")
        print("아이디를 확인하는 회원 아이디 찾기 페이지로 이동...")
        return

    if user_pass != "1234":
        print("비밀번호가 일치하지 않습니다.")
        return

    print("로그인 처리 로직...")
```

```python
login("user02","1234")
```

```python
# 회원 페이지 로직 처리
try: 
    login("user01","1234")
except ValueError as e:
    print(e)
    print("회원 페이지에서 추가적으로 할 로직...")
```

```python
# 관리자 페이지 로직 처리
try: 
    login("user02","1234")
except ValueError as e:
    print(e)
    print("회원 페이지에서 추가적으로 할 로직...")
```

7. 사용자 정의 예외
- Exception은 모든 예외의 부모 클래스
- Exception을 상속 받으면 or Exception을 상속받은 자식 클래스를 상속 받으면 그 클래스도 예외 클래스! (상속의 상속)

```python
# 클래스명만 보더라도 예외케이스를 누구나 판단 할 수 있도록 명명하는 것! 
class UserIdMismatchError(Exception):
    pass

class UserPassmismatchError(Exception):
    pass 
```

```python
# 사용자 정의 예외를 raise와 함께 쓰기!
def login(user_id,user_pass):
    if user_id != "user01":
        raise UserIdMismatchError("아이디가 일치하지 않습니다.") # 예외가 발생하고, 코드 실행 흐름은 중단 됩니다. 

    if user_pass != "1234":
        raise UserPassmismatchError("비밀번호가 일치하지 않습니다.")

    print("로그인 처리 로직...")
```

```python
try:

except UserIdMismatchError as e:

except UserPassmismatchError as e:
```

```python
try:
    login("user02","1234")
except (UserIdMismatchError, UserPassmismatchError) as e:
    print(e)
```

```python
try:
    login("user01","12345")
except (UserIdMismatchError, UserPassmismatchError) as e:
    print(e)
```

- [중요!]
- raise를 만나면, 흐름이 끊기고 최초 발생한 예외 케이스로 알림니다. (return하고 비슷함)

```python
# __str__사용하기!
class UserIdMismatchError(Exception):
    def __str__(self):
        return "아이디가 일치하지 않습니다."

class UserPassmismatchError(Exception):
    def __str__(self):
        return "비밀번호가 일치하지 않습니다."
```

8. traceback
- traceback.print_exc() : 에러 스택을 콘솔 출력
- traceback.format_exc() : 에러 스택을 문자열로 반환
- traceback.print_stack() : 에러 없이 현재 호출 흐름 확인
- traceback = 오류 로그
- 가장 하단에 있는 로그가 최신 로그

```python
import traceback

try:
    result = 10 / 0
    print(result)

except:
    print("에러 발생")
    traceback.print_exc()
```

```python
import traceback

try:
    result = 10 / 0
    print(result)

except:
    print("에러 발생")
    message = traceback.format_exc()
    print("message",message)
```
