### 필수 1 : 에러 트레이스백 역추적 및 원인 진단
---
#### 문제 1-1 :  ZeroDivisionError 확인과 소스코드 역추적

```python
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

### 필수 2 : VS Code 디버거와 중단점을 활용한 실행 일시 정지
---
#### 문제 2-1 :  변수 상태 모니터링을 위한 중단점 설정

```python
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```

#### 문제 3-1 : 반복문 내부의 누적 변수 데이터 변화 검증

```python
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```
