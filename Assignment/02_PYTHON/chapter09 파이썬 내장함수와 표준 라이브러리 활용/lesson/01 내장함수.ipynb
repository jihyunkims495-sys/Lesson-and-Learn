2. all(), any() 함수 
- 핵심 차이는 “어떤 조건에서 True가 되느냐”
    - all(iterable값) : 모든 요소가 참(True)으로 인식 될 때 = “전부 다 맞아?”
        - 예외 ) 빈 리스트all([]) 는 True
    - any(iterable값) : 요소중 하나라도 참(True)로 인식 될 때 = 하나라도 True가 있으면 True

```python
all([True,1," "])
```

```python
all([False,False])
```

```python
all([])  # 빈 리스트 예외
```

```python
any([False,1," "])
```

```python
any([])  # 빈 리스트는 거짓
```

3. len(),sum(),round() 함수
- len() 
    - 문자열에서 요소의 길이 (글자수)
    - 리스트, 튜플 -> 요소의 갯수
- sum(interable값) : 합계(요소는 숫자로만 구성)
- round() : 반올림
- ceil() : 올림
- floor() : 버림

```python
num = 10.56
print(round(num))
```

```python
import math # math 모듈에 포함 되어 있습니다

print(math.ceil(10.2))
print(math.floor(num))
```

4. mean(), median()
- statistics 모듈에 포함된 표준 라이브러리로 데이터 분석할 때 많이 씁니다
- mean(interable값) : 평균
- median(interable값) : 중앙값


```python
import statistics
nums =[10,15,20,8,7,100000]

print("평균:",statistics.mean(nums),"중앙값:",statistics.median(nums))
```
