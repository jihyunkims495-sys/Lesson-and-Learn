### 1. 항과 연산자
---
- 항 : 연산에 사용되는 값 /  1개이면 '단항, 2개이면 '이항', 3개이면 '3항'
    - -5 → 값 1개
    - 3 + 5 → 값 2개
    - A if 조건 else B → 값 3개
- 연산자 : 연산에 사용되는 기호 (eg.) +,-,=,*,/,//

### 2. 연산자의 종류
---
#### 2-1. 산술 연산자의 종류
-  "+": 더하기
- "-": 빼기
- "*": 곱하기
- "/": 나누기
- "//": 정수나누기, 몫
- "%": 나머지
- "**": 제곱 
#### 2-2. 산술 연산자의 우선순위
- ** , * , /,//,% > +,-
- 우선순위 강제 적용 :  ( ) 소괄호로 묶어주면, 연산의 우선순위가 높아집니다. 
    - 일반적으로 코드를 연산할 때, ()를 항상 작성해주는 것이 개발자 코드 소통 방법입니다!
    - 복잡한 복합 연산식에서 소괄호를 명시적으로 사용하면 우선순위를 따로 외우지 않아도 논리적 오류와 실수를 원천 차단 가능!
    - 코드를 읽는 동료 개발자에게도 연산의 의도를 명확히 전달할 수 있어 협업 시 가독성이 획기적으로 높아짐!


```python
# 대입 연산자 (=) 연산의 우선순위가 가장 낮으며, 가장 마지막에 연산됩니다. 
# 산술 연산자 (+, -, *, /, //, %, **) 연산의 우선순위가 대입 연산자보다 높으며, 가장 먼저 연산됩니다.
x = 10
x = 10 + 20 
print(x)
```

```python
# 산술 연산 예시
num1 = 10
num2 = 3

print(num1 // num2)
print(num1 % num2)
print(num1 ** num2)
print(num1**2)
print(num1**3)
```

#### 2-3. 비교 연산자
- "> , < ,>= , <=" 
- == : 같다 (동등성, 동일한 가치)
- != : 다르다
    - ! 는 부정 연산자로 반대로 출력하는 연산자
    - 비교 연산의 결과는 무조건 참(true)과 거짓(false)의 논리값이 나옵니다. 
    - 조건문(if,elif,else)과 반복문(for)과 같은 판별이 필요한 구문에서 주로 사용!

[별첨.] 동등성 vs 동일성
- 동등성 : 동일한 가치를 비교 (== 연산자)
- 동일성 : 동일한 대상인지 비교 (is 메서드)
    - is 연산자는 두 변수가 같은 객체(메모리 주소)를 가리키는지 확인하는 연산자
    - 값이 같은지를 비교하는 것이 아니라, "같은 객체(object)인가?"를 비교한다.
        -  a is b : a와 b가 같은 객체 인가?
        - a is not b : a와 b가 다른 객체 인가? 
    

```python
#비교 연산자의 예시
a = 10 > 5
print(a)
```

```python
# 1. 괄호가 없는 복합 연산: 산술 연산자 '*'가 '+'보다 먼저 계산된 후 비교 연산이 수행됨
# 2 * 5(=10) -> 10 + 10(=20) -> 20 > 25 순서로 비교되어 최종 False 결과가 도출됩니다.
result1 = 10 + 2 * 5 > 25 
print(result1)

# 2. 소괄호를 활용한 연산 제어: 괄호 내부의 덧셈 연산이 최우선으로 계산됨
# (10 + 2)(=12) -> 12 * 5(=60) -> 60 > 25 순서로 비교되어 최종 True 결과가 도출됩니다.
result2 = (10 + 2) * 5 > 25  
print(result2)
```

```python
#동등성과 동일성 예시
num1 = 1000
num2 = 1000

print(id(num1))
print(id(num2))
```

```python
#동등성 비교 예시
num1 == num2 
```

```python
#동일성 비교 예시 (주소를 비교합니다.)
num1 is num2 
num1 is not num2
```

#### [별첨.] 코드 단축을 위한 복합 대입 연산자
- 복합 대입의 작동 방식
    - 변수의 현재 값에 연산을 가한 뒤, 그 결과를 다시 동일한 변수에 저장할 때 코드를 간결하게 줄여주는 표현식
    - 대입연산자(=)와 함께 다른 연산자(+,-,*,/,//,%,**)와 결합
    - 형태 : +=, -=, *=, /=, //=, %= 등
- 실무적 이점
    - 반복적인 변수 작성을 배제하여 코드의 길이를 줄이고 가독성을 높이며, 변수명을 여러 번 적는 과정에서 발생하는 오타를 원천적으로 방지할 수 있음

```python
# 'num' 변수가 계속 반복되는 이항식의 코드는 비효율적이므로 복합 대입 연산으로 코드 효율을 높입니다. 
# 반복적인 이항연산을 단항으로 효율화 해주는 것이 개발자의 역할 입니다. 
num = 1
num = num + 1
num = num + 1
num = num + 1

print(num)
```

```python
# 위 코드에 복합 대입 연산자를 사용하면,
num = 1
num += 1
num += 1
num += 1
print(num)

num *= 2
print(num)
```

```python
# 복합 대입 연산의 예시
a = 100

a += 50 # a = a + 50 와 같다.
print(a)

a -= 50 # a = a - 50 와 같다.
print(a)

a *= 50 # a = a * 50 와 같다.
print(a)

a /= 50 # a = a / 50 와 같다. 단, 나누기를 하면 무조건, 실수형(float)으로 변환됨!
print(a)

a %= 50 # a = a % 50 와 같다. 무조건, 실수형(float)으로 변환됨!
print(a)
```

#### 2-4. 논리 연산자 (and, or, not)
---
- 연산자의 종류
    - and(논리곱): 각 항이 모두 True 조건, 하나라도 False라면 False가 된다.
    - or(논리합) : 각 항의 하나라도 True 이면, True가 된다. 
    - not(논리부정) : 조건을 반대로 바꿈(반전), True 조건이면 False로 반환, False 조건이면 True로 반환.

- 논리 연산자의 우선순위
    - 비교 연산자가 우선된 후에 논리 연산이 진행됩니다. 비교 연산이 된 후에 결과들을 가지고 논리를 따져야 하기 때문입니다.

```python
#논리 연산의 예시
num1 = 20

result1 = num1 > 10
result2 = num1 < 100

print(result1)
print(result2)
```

```python
# and 연산 예시
num1 = 20

result1 = num1 > 10
result2 = num1 < 100

print(result1 and result2)

# not 연산 예시
print(not(result1 and result2))

# or 연산 예시
fruit1 = "Apple"
fruit2 = "Pear"

result1 = fruit1 == "Apple"
fesult2 = fruit2 == "Orange"

print(result1)
print(result2)
print(result1 or result2)
```

### 3. 단락 회로 평가 (Short-Circuit Evaluation)
---
논리 연산 시 왼쪽 항의 결과만으로 전체 식의 참/거짓 결과가 이미 확정되었다면, 오른쪽 항은 평가(실행)하지 않고 건너뛰는 파이썬의 연산 최적화 방식
- and 연산은 좌측이 False이면 우측 항이 무엇이든 최종 결과가 무조건 False이므로 오른쪽 연산을 생략합니다.
- or 연산은 좌측이 True이면 우측 항이 무엇이든 최종 결과가 무조건 True이므로 오른쪽 연산을 생략합니다.

```python
# or 연산의 단락 회로 평가 예시
## 변수의 값이 첫 항에서 True라는 결과가 참으로 결정이 되어 or 연산의 결과가 이미 결정되었으므로, or 연산 뒷쪽의 산술 연산을 진행하지 않습니다. 
result = True or 10**2 
print(result)

## 변수의 값이 첫 항에서 False라는 결과가 거짓이어서 or의 연산 결과가 결정되지 않았으므로, or 연산 뒷쪽의 산술 연산을 진행합니다. 
result = False or 10**2
print(result)

# or 연산을 이용한 기본값 (없을 때 대체되는 값) 설정할 때 
message = "" #값이 비어있으면 False 값이 되므로, or의 연산이 뒷쪽의 값으로 넘깁니다. 
result = message or "기본 메세지"
print(result)
```

```python
# and 연산의 단락 회로 평가 예시
## 변수의 값이 첫 항에서 False라는 결과가 참으로 결정이 되어 and 연산의 결과가 이미 결정되었으므로, and 연산 뒷쪽의 산술 연산을 진행하지 않습니다. 
result = False and 10**2 
print(result)

## 변수의 값이 첫 항에서 True라는 결과가 거짓이어서 and의 연산 결과가 결정되지 않았으므로, and 연산 뒷쪽의 산술 연산을 진행합니다. 
result = True and 10**2 
print(result)
```

```python
# and 연산을 이용한 간단한 조건 처리
height = 165

result = height >= 160 and "크다" #변수의 값이 160 보다 크거나 같고 (and) "크다"
print(result) 
```
