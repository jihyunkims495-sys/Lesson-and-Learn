# 1. 변수의 본질과 파이썬의 참조 메커니즘

## 1.1 변수의 정의와 참조 메커니즘

### 메모리에 객체를 만들고 이름 붙이기

파이썬에서 다음 코드를 실행한다고 가정해 보겠습니다.

```python
x = 10
```

이 코드가 실행되면 다음과 같은 일이 일어납니다.

1. 파이썬이 `10`이라는 정수 객체를 준비합니다.
2. 변수 이름 `x`가 정수 객체 `10`을 가리키도록 연결됩니다.
3. 이후 `x`를 사용하면 `x`가 가리키는 객체를 가져옵니다.

```text
변수 이름 x ─────→ 정수 객체 10
```

변수는 데이터를 직접 담고 있는 상자라기보다 메모리에 존재하는 객체에 붙인 **이름표**라고 이해할 수 있습니다.

### 변수 선언과 출력

```python
x = 10

print(x)
```

### 실행 결과

```text
10
```

### 변수의 참조 대상 변경

```python
x = 10
print(x)

x = 20
print(x)
```

### 실행 결과

```text
10
20
```

처음에는 `x`가 `10`을 가리켰지만, 재할당 이후에는 `20`을 가리킵니다.

```text
재할당 전: x ─────→ 10
재할당 후: x ─────→ 20
```

---

## 1.2 참조 방식의 특징

파이썬 변수는 객체 자체가 아니라 객체를 가리키는 이름입니다.

따라서 여러 변수가 같은 객체를 가리킬 수도 있습니다.

```python
a = 100
b = a

print(a)
print(b)
```

### 실행 결과

```text
100
100
```

`b = a`를 실행하면 `b`도 `a`가 가리키는 객체를 가리키게 됩니다.

```text
a ─────┐
       ├────→ 정수 객체 100
b ─────┘
```

`id()` 함수로 두 변수가 같은 객체를 가리키는지 확인할 수 있습니다.

```python
a = 100
b = a

print(id(a))
print(id(b))
print(id(a) == id(b))
```

### 실행 결과 예시

```text
140724000000000
140724000000000
True
```

> `id()`의 실제 출력값은 컴퓨터와 실행 환경마다 다를 수 있습니다.

## 1.3 올바른 변수 이름을 위한 명명 규칙

변수 이름은 원하는 대로 작성할 수 있는 것이 아니라 파이썬의 문법 규칙을 지켜야 합니다.

### 사용할 수 있는 문자

파이썬 변수 이름에는 다음 문자를 사용할 수 있습니다.

- 영문 대문자
- 영문 소문자
- 숫자
- 언더바 `_`
- 한글 등 일부 유니코드 문자

일반적인 실무 코드에서는 주로 영문 소문자, 숫자, 언더바를 사용합니다.

```python
user_name = "김지현"
age2 = 30
_total_price = 50000
```

### 숫자로 시작할 수 없음

다음 변수 이름은 올바르지 않습니다.

```python
# 잘못된 변수명입니다.
# 1st_value = 100
# 3rd_step = "완료"
```

다음과 같이 숫자를 첫 글자가 아닌 위치에 사용해야 합니다.

```python
value_1st = 100
step_3rd = "완료"

print(value_1st)
print(step_3rd)
```

### 공백을 사용할 수 없음

```python
# 잘못된 코드입니다.
# user name = "김지현"
# total price = 50000
```

```python
user_name = "김지현"
total_price = 50000
```

### 특수문자를 사용할 수 없음

언더바를 제외한 대부분의 특수문자는 변수 이름에 사용할 수 없습니다.

```python
# 잘못된 코드입니다.
# user-name = "김지현"
# total@price = 50000
# item!count = 3
```

올바른 변수 이름은 다음과 같습니다.

```python
user_name = "김지현"
total_price = 50000
item_count = 3
```

### 파이썬 키워드를 사용할 수 없음

파이썬이 문법을 위해 미리 사용하고 있는 예약어는 변수 이름으로 사용할 수 없습니다.

```python
# 잘못된 코드입니다.
# if = 10
# for = "반복"
# while = True
# True = 1
```

파이썬 키워드는 `keyword` 모듈로 확인할 수 있습니다.

```python
import keyword

print(keyword.kwlist)
```

특정 단어가 파이썬 키워드인지 확인할 수도 있습니다.

```python
import keyword

print(keyword.iskeyword("if"))
print(keyword.iskeyword("user_name"))
```

### 실행 결과

```text
True
False
```

# 2. 파이썬의 핵심 특성: 동적 타이핑

## 2.1 동적 타이핑의 개념

### 자료형 선언 생략

일부 프로그래밍 언어에서는 변수를 만들 때 자료형을 직접 지정해야 합니다.

다음은 개념을 설명하기 위한 예시이며 파이썬 코드가 아닙니다.

```text
int age = 20
String name = "김지현"
```

파이썬에서는 자료형을 미리 지정하지 않고 변수에 값을 대입합니다.

```python
age = 20
name = "김지현"
```

파이썬은 실행 시점에 대입된 객체를 확인하여 자료형을 결정합니다.

```python
age = 20
name = "김지현"

print(type(age))
print(type(name))
```

### 실행 결과

```text
<class 'int'>
<class 'str'>
```

---

## 2.2 실시간 타입 결정

하나의 변수에 서로 다른 자료형의 값을 재할당할 수 있습니다.

```python
data = 100
print(data)
print(type(data))

data = "Python"
print(data)
print(type(data))

data = 3.14
print(data)
print(type(data))

data = True
print(data)
print(type(data))
```

### 실행 결과

```text
100
<class 'int'>
Python
<class 'str'>
3.14
<class 'float'>
True
<class 'bool'>
```

변수 `data`의 자료형이 고정된 것이 아닙니다.

`data`가 현재 어떤 객체를 가리키고 있는지에 따라 `type(data)`의 결과가 달라집니다.

정확히 표현하면 변수 자체의 타입이 바뀐다기보다 변수가 **서로 다른 타입의 객체를 새롭게 가리키게 되는 것**입니다.

---

## 2.3 동적 타이핑의 장점

### 코드가 간결함

자료형을 일일이 선언할 필요가 없어 코드가 간결합니다.

```python
student_name = "김지현"
student_age = 30
average_score = 92.5
is_passed = True
```

### 빠른 개발과 실험

변수의 자료형을 미리 고정하지 않아도 되므로 코드를 빠르게 작성하고 실험할 수 있습니다.

이러한 특징은 다음 분야에서 유용하게 사용됩니다.

- 데이터 분석
- 데이터 과학
- 머신러닝
- 인공지능
- 자동화
- 웹 서비스 프로토타입

---

## 2.4 동적 타이핑의 트레이드오프

**트레이드오프(Trade-off)**는 어떤 장점을 얻는 대신 다른 단점이나 위험을 감수해야 하는 관계를 의미합니다.

동적 타이핑은 코드를 유연하고 빠르게 작성할 수 있다는 장점이 있지만 실행하기 전까지 타입 관련 문제를 발견하기 어려울 수 있습니다.

### 정상적인 연산

```python
price = 10000
quantity = 3

total_price = price * quantity

print(total_price)
```

### 실행 결과

```text
30000
```

### 타입이 잘못 들어온 경우

```python
price = 10000
quantity = "세 개"

# 다음 코드는 예상한 숫자 계산이 되지 않습니다.
total_price = price * quantity

print(total_price)
```

문자열에 정수를 곱하면 문자열이 반복되므로 의도와 다른 결과가 나올 수 있습니다.

다른 연산에서는 `TypeError`가 발생할 수 있습니다.

```python
price = 10000
discount = "1000"

# 정수와 문자열을 뺄 수 없으므로 TypeError가 발생합니다.
# final_price = price - discount
```

예상되는 오류는 다음과 같습니다.

```text
TypeError: unsupported operand type(s) for -: 'int' and 'str'
```

### 타입 확인 후 연산하기

```python
price = 10000
discount = "1000"

print(type(price))
print(type(discount))
```

### 문자열을 정수로 변환하기

```python
price = 10000
discount = "1000"

discount = int(discount)
final_price = price - discount

print(final_price)
```

### 실행 결과

```text
9000
```

---

## 2.5 동적 타이핑의 장단점

| 구분 | 설명 |
|---|---|
| 장점 | 자료형을 미리 선언하지 않아 코드가 간결합니다. |
| 장점 | 다양한 값을 유연하게 처리할 수 있습니다. |
| 장점 | 아이디어를 빠르게 코드로 구현할 수 있습니다. |
| 단점 | 실행 전에는 타입 관련 문제를 발견하기 어려울 수 있습니다. |
| 단점 | 변수에 어떤 타입의 값이 들어 있는지 추적하기 어려울 수 있습니다. |
| 단점 | 잘못된 타입이 들어오면 실행 중 `TypeError`가 발생할 수 있습니다. |

# 3. 변수 재할당과 메모리 전환 구조

## 3.1 변수 재할당

이미 값이 연결되어 있는 변수에 새로운 값을 대입하는 것을 **재할당(Reassignment)**이라고 합니다.

```python
age = 20
age = 25

print(age)
```

### 실행 결과

```text
25
```

### 내부 작동 과정

첫 번째 코드가 실행됩니다.

```python
age = 20
```

```text
age ─────→ 정수 객체 20
```

다음 코드가 실행됩니다.

```python
age = 25
```

```text
정수 객체 20     age ─────→ 정수 객체 25
```

`age`라는 변수 이름이 기존 객체 `20`과의 연결을 끊고 새로운 객체 `25`를 가리키게 됩니다.

---

## 3.2 자료형이 다른 객체로 재할당

동적 타이핑을 사용하는 파이썬에서는 자료형이 다른 객체도 같은 변수에 재할당할 수 있습니다.

```python
value = 100

print(value)
print(type(value))

value = "백"

print(value)
print(type(value))
```

### 실행 결과

```text
100
<class 'int'>
백
<class 'str'>
```

---

## 3.3 가비지 컬렉터

변수가 다른 객체를 가리키게 되면 기존 객체를 가리키는 참조가 사라질 수 있습니다.

어떤 객체를 가리키는 참조가 더 이상 없고 파이썬이 해당 객체가 필요하지 않다고 판단하면 자동 메모리 관리 기능이 객체가 사용하던 메모리를 정리할 수 있습니다.

이러한 자동 메모리 관리에 관여하는 기능을 **가비지 컬렉터(Garbage Collector)**라고 합니다.

```python
message = "첫 번째 문자열"
message = "두 번째 문자열"
```

첫 번째 줄이 실행된 상태는 다음과 같습니다.

```text
message ─────→ "첫 번째 문자열"
```

두 번째 줄이 실행되면 `message`의 참조 대상이 바뀝니다.

```text
"첫 번째 문자열"     message ─────→ "두 번째 문자열"
```

개발자가 일반적인 파이썬 프로그램에서 메모리를 직접 해제하지 않아도 파이썬이 자동으로 관리합니다.

> 실제 메모리 정리 시점과 방식은 파이썬 구현과 객체 상태에 따라 달라질 수 있습니다.

## 3.4 `type()` 함수

`type()`은 객체의 자료형을 확인하는 내장 함수입니다.

```python
number = 10
text = "Hello"
decimal = 3.14
is_active = True

print(type(number))
print(type(text))
print(type(decimal))
print(type(is_active))
```

### 실행 결과

### 주요 자료형

| 값 | 자료형 | 의미 |
|---|---|---|
| `10` | `int` | 정수 |
| `3.14` | `float` | 실수 |
| `"Hello"` | `str` | 문자열 |
| `True` | `bool` | 불리언 |
| `[1, 2, 3]` | `list` | 리스트 |
| `{"name": "김지현"}` | `dict` | 딕셔너리 |

## 3.5 `id()` 함수

`id()`는 객체를 구분하기 위한 고유한 정수 식별값을 반환하는 내장 함수입니다.

```python
x = 10

print(type(x))
print(id(x))
```

### 실행 결과 예시

```text
<class 'int'>
140724000000000
```

출력되는 `id` 값은 실행 환경마다 다를 수 있습니다.

> CPython에서는 `id()` 값이 객체의 메모리 주소와 관련된 값으로 구현되는 경우가 많지만, 파이썬 언어 차원에서는 객체를 구분하는 고유 식별값으로 이해하는 것이 정확합니다.

## 3.6 재할당 전후의 `type()`과 `id()` 비교

```python
x = 10

print("첫 번째 값:", x)
print("첫 번째 자료형:", type(x))
print("첫 번째 객체 식별값:", id(x))

x = "Hello"

print("두 번째 값:", x)
print("두 번째 자료형:", type(x))
print("두 번째 객체 식별값:", id(x))
```

### 실행 결과 예시

```text
첫 번째 값: 10
첫 번째 자료형: <class 'int'>
첫 번째 객체 식별값: 140724000000000

두 번째 값: Hello
두 번째 자료형: <class 'str'>
두 번째 객체 식별값: 2200000000000
```

`x`에 새로운 객체를 재할당하면 다음 내용이 달라질 수 있습니다.

- `x`가 가리키는 값
- `x`가 가리키는 객체의 자료형
- `x`가 가리키는 객체의 `id`

## 3.7 같은 객체를 참조하는 변수 확인

```python
x = 100
y = x

print("x의 값:", x)
print("y의 값:", y)

print("x의 id:", id(x))
print("y의 id:", id(y))

print("같은 객체인가?", x is y)
```

### 실행 결과 예시

```text
x의 값: 100
y의 값: 100
x의 id: 140724000000000
y의 id: 140724000000000
같은 객체인가? True
```

`is` 연산자는 두 변수가 같은 객체를 가리키는지 확인합니다.

값이 같은지 비교할 때는 `==`를 사용하고, 동일한 객체인지 비교할 때는 `is`를 사용합니다.

```python
print(x == y)
print(x is y)
```

# 7. 용어 사전

| 용어 | 영어 표기 | 설명 |
|---|---|---|
| 변수 | Variable | 메모리에 존재하는 객체를 가리키는 이름입니다. 데이터를 직접 담는 상자보다 객체에 붙인 이름표로 이해할 수 있습니다. |
| 명명 규칙 | Naming Rules | 변수 이름을 만들 때 반드시 지켜야 하는 문법 규칙입니다. 숫자로 시작할 수 없고 공백이나 파이썬 키워드를 사용할 수 없습니다. |
| 명명 관례 | Naming Convention | 문법 오류는 아니지만 코드의 가독성과 일관성을 위해 함께 지키는 이름 작성 방식입니다. |
| 스네이크 케이스 | Snake Case | 여러 단어를 언더바로 연결하는 파이썬의 권장 변수 명명 방식입니다. |
| 참조 | Reference | 변수 이름과 메모리에 존재하는 객체가 연결된 관계입니다. |
| 참조 메커니즘 | Reference Mechanism | 변수가 실제 데이터를 직접 소유하는 것이 아니라 객체를 가리키도록 연결되는 동작 방식입니다. |
| 동적 타이핑 | Dynamic Typing | 실행 시점에 변수가 가리키는 객체에 따라 자료형이 결정되는 특성입니다. |
| 런타임 | Runtime | 프로그램이 실제로 실행되고 있는 시점입니다. |
| 트레이드오프 | Trade-off | 어떤 장점을 얻는 대신 다른 단점이나 위험을 감수해야 하는 관계입니다. |
| 변수 재할당 | Variable Reassignment | 기존 변수에 새로운 값을 대입하여 변수가 새로운 객체를 가리키게 하는 것입니다. |
| 객체 | Object | 파이썬 메모리에 실제로 존재하는 숫자, 문자열, 리스트 등의 데이터 실체입니다. |
| 자료형 | Data Type | 객체가 어떤 종류의 데이터인지 나타내는 분류입니다. `int`, `str`, `float`, `bool` 등이 있습니다. |
| `type()` | Type Function | 객체의 자료형을 확인하는 파이썬 내장 함수입니다. |
| `id()` | Identity Function | 객체를 구분하기 위한 고유 정수 식별값을 반환하는 파이썬 내장 함수입니다. |
| 가비지 컬렉터 | Garbage Collector | 더 이상 필요하지 않은 객체가 사용하던 메모리를 자동으로 관리하고 정리하는 기능입니다. |
| 세션 유지 | Session Persistence | 주피터 커널을 종료하거나 재시작하기 전까지 실행된 변수와 함수 등의 메모리 상태가 유지되는 특성입니다. |
| 자동 출력 | Automatic Output | 주피터 코드 셀의 마지막 표현식을 `print()` 없이 화면에 표시하는 기능입니다. |
| 커널 | Kernel | 주피터 노트북의 코드를 실행하고 변수, 함수, 객체 등의 상태를 메모리에 유지하는 실행 엔진입니다. |
| Restart | Kernel Restart | 현재 커널을 다시 시작하여 메모리에 저장된 변수와 실행 상태를 초기화하는 기능입니다. |
| Run All | Run All Cells | 노트북의 모든 셀을 위에서 아래 순서로 실행하는 기능입니다. |
