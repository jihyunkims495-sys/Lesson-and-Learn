## 0. 문자열 형식화(Formatting)
---
변수에 담긴 데이터를 고정된 문자열 틀 사이에 자연스럽게 끼워 넣어 정형화된 텍스트 양식으로 출력하는 기술

## 1. 1세대 : % 연산자 방식
---
- C언어 등 초기 프로그래밍 언어에서부터 사용되던 전통적인 방식
- 문자열 안에 "%s(문자열), %d(정수), %f(실수)" 등의 기호로 자리(=변경되는 자리)를 만들어 두고, 우측에서 변수를 순서대로 대입
- 변수의 자료형과 대입할 기호의 타입이 일치하지 않으면 에러가 발생

    - 자릿수(너비) 지정
        - %와 타입 기호 사이에 숫자를 넣어 출력할 글자 총 너비를 지정할 수 있습니다. 양수는 오른쪽 정렬, 음수는 왼쪽 정렬을 수행합니다. (예: %5d는 5칸을 확보하고 오른쪽 정렬)

    - 실수 소수점 지정
        - %f 기호 앞에 .과 숫자를 조합하여 소수점 아래 몇 번째 자리까지 출력할지 제어합니다. (예: %.2f는 소수점 둘째 자리까지 표기)

- [%숫자s][%숫자d][%숫자f] : 최소 숫자 자리수 만큼 공간을 차지하며 남는 공간은 공백
- [%0숫자d][%0숫자f] : 남는 공백은 0으로 대체되며, 문자열에는 반영되지 X
- [%.숫자f] : 숫자만큼 소수점을 표시 (eg.) %.2f는 소숫점 2자리까지 표기함
- [%숫자.숫자f] : 소수점 앞은 "정수" 소수점 뒤는"소수점"이며 값이 없으면 공백
- [%0숫자.숫자f] : 공백에 0 채우기


```python
# 1세대 예시로 가장 오래된 방식 입니다.
message = "나의 이름은 %s이고, 나이는 %d입니다." 
print(message %("철수",20))

# [%숫자s][%숫자d][%숫자f] : 숫자만큼 공백이 생깁니다.
message = "나의 이름은 %10s이고, 나이는 %10d입니다." 
print(message %("철수",20))

# [%0숫자d][%0숫자f] : 숫자만큼 0이 생깁니다.
message = "나의 이름은 %010s이고, 나이는 %010d입니다." 
print(message %("철수",20))

#[%.숫자f] : 소수점 자릿수 지정하기
message = "숫자:%f" % 10.123456789
message = "숫자:%.2f" % 10.123456789
print(message)

#[%숫자.숫자f] : 소수점 앞에 "10"자리 넣기
message = "숫자:%10.2f" % 10.123456789
print(message)

#정수 앞 공백에 "0" 채워넣기
message = "숫자:%010.2f" % 10.123456789
print(message)
```

```python
# 정수 표현하기
message = "숫자 : %d" % 10.12345
print(message)

# 정수보다 실수를 더 많이 씁니다!!!
message = "숫자 : %s" % 10.12345
print(message)
```

## 2. 2세대 : str.format() 메서드 방식
---
- 순서 지정 교체 : {0},{1},{2} ...
- 단어 지정 교체 : {단어},{단어}...
- 파이썬 3 버전부터 도입된 방식으로, 기호 매칭의 엄격함을 줄이기 위해 중괄호(***{}***) 대입 방식을 사용
- 중괄호 안에 인덱스 번호(***{0}***)나 임의의 이름(***{price}***)을 지정할 수 있어 타입 에러에서 자유로움

```python
순서 지정 교체
# 출력의 차이 비교 예시
message = "내 이름은 {0}이고, 나이는 {1}입니다."
print(message)
message.format("철수",20)

# 순서는 index로 지정 되므로, 자리를 switch 할 수  있습니다.
message = "내 이름은 {1}이고, 나이는 {0}입니다."
message.format("철수",20)
```

```python
# 단어 지정 교체 예시
message = "내 이름은 {name}이고, 나이는 {age}입니다."
message.format(name="철수",age=20)

# 순서도 switch 할 수  있습니다.
message = "내 이름은 {age}이고, 나이는 {name}입니다."
message.format(name="철수",age=20)
```

```python
# format 메서드 안에는 관례적으로 숫자(정수,실수)를 먼저 써줍니다.
message = "내 이름은 {name}이고, 나이는{0}입니다."
message.format(20,name="철수")
```

## 3. 3세대 : f-string 방식
---
- 파이썬 3.6부터 도입되어 현재 업계 표준으로 정착한 직관적인 문자열 대입법
- 문자열 시작 부분에 문자 f를 붙이고, 중괄호 {변수명} 안에 변수나 가벼운 수식을 직접 적어요

### 정렬 및 서식 지정 지시자
- 중괄호 내부에서 콜론(:) 뒤에 지시자를 붙여 출력 형식을 정밀하게 제어하는 기능
- 최근 추가 된 기능 
    - 정렬 및 서식 지정 지시자
        - :<는 왼쪽 정렬, :>는 오른쪽 정렬을 수행하며, 지정한 숫자만큼 너비(공백)를 확보. :,를 추가하면 숫자에 천 단위 쉼표를 자동으로 표시
        - 뾰족한 쪽의 방향이에요!
    - 디버깅 조절자 (=) 버그를 제거하기 위해 
        - 중괄호 내부의 변수명 뒤에 등호(=)를 붙여 출력하는 기능
        - [중요!] 출력 시 변수의 이름과 현재 값이 동시에 표시!
        - 별도의 라벨 텍스트를 적지 않아도 데이터의 상태를 직관적으로 추적할 수 있어 로그 확인과 디버깅에 유용
        

### 공백 넣기!
- 정렬 및 서식 지정 지시자 뒷쪽에 "숫자"를 넣으면, "숫자"만큼 공백이 지정됩니다.

```python
# 공백 넣기!
price = 1234567

print(f"금액:{price:<15,}원")  # 왼쪽 정렬 → 오른쪽에 공백
print(f"금액:{price:>15,}원")  # 오른쪽 정렬 → 왼쪽에 공백
print(f"금액:{price:^15,}원")  # 가운데 정렬 → 양쪽에 공백
```

```python
# 공백에 0 넣기!
price = 1234567

print(f"금액:{price:<015,}원")  # 왼쪽 정렬 → 오른쪽에 공백
print(f"금액:{price:>015,}원")  # 오른쪽 정렬 → 왼쪽에 공백
print(f"금액:{price:^015,}원")  # 가운데 정렬 → 양쪽에 공백
```

```python
# 변수에 값을 지정, 한 줄로 직접 입력 가능!
name = "영희"
age = 20

message = f"내 이름은 {name}이고, 나이는 {age}입니다."
```

```python
# :,를 추가해 천 단위 쉼표를 자동으로 표시
price = 10000
f"가격 : {price:,}"

# 정렬 예시 / 오른쪽 정렬
price = 10000
f"가격 : {price:>10,}원"
```

```python
# 정렬 예시 / 왼쪽 정렬
price = 10000
f"가격 : {price:<10,}원"
```

```python
# 디버깅 조절자 예시
price = 10000
f"{price=}"
```

```python
# 디버깅 조절자가 없다면, 값만 출력됩니다.
price = 10000
f"{price}"
```
