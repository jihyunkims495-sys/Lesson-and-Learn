### 1. 파이썬의 기본 자료형 (Primitive Data Types)
---
- 숫자형 : 정수형 (int) 과 실수형 (flaot)
- 문자형 : 문자와 문장, ''와 "" (한줄) """..."""(여러줄) 
- 논리형 : 참(true)과 거짓(false)


[별첨.] 값 자체로 참과 거짓을 인식하는 특징

    - 거짓(False)으로 인식하는 값

    - 참(True)으로 인식하는 값
        - `0`이 아닌 숫자, 글자가 포함된 문자열, 요소가 채워진 리스트 등 (값이 존재하는 상태)
        - 이 특징 덕분에 데이터 분석이나 조건문을 작성할 때 코드를 훨씬 간결하게 줄일 수 있습니다.

### 2. 명시적 형변환(Explicit Type Conversion)
---
- 연산은 같은 자료형끼리 가능

````python
#다른 자료형의 연산시 error가 발생하는 예시
num = "100"
print(type(num))
````

````python
num + 200
print(num)
````

- 형변환 함수
    - int() : 정수로 변환
    - float() : 실수로 변환
    - str() : 문자로 변환
    - bool() : 불릿형태로 변환

````python
#동일한 자료형으로 맞춰주면, typeError가 발생하지 않습니다. 
int(num) + 200

print(num)
````

````python
"현재 시간은" + 10 + "시 입니다." # 문자형 + 숫자형 + 문자형 구조로, typeerror가 발생합니다. 
"현재 시간은 " + str(10) + "시 입니다." # 문자형 + 문자형 + 문자형 구조로, 실행이 됩니다. 
````

### 3. 묵시적 형변환(Implicit Type Conversion)
---

````python
#정수형과 실수형의 자료형 확인 예시
num1 = 100
num2 = 3.123

print(type(num1))
print(type(num2))
````

````python
#그런데, 하단처럼 "정수 + 실수"연산 시, 자동으로 실수(foat)형으로 변환되어 연산이 됩니다.
num3 = num1 + num2 #num1이 int > float 자동 형변환(묵시적 형변환)
print(num3)
````

````python
num3 = num1 + int(num2) #num2가 정수를 포함하며 소수점이 유실됩니다. 
print(num3)
````

##### [개념 정리 ] 묵시적 형변환(Implicit Type Conversion)
-  데이터의 성격이 다른 두 값을 연산할 때, 개발자가 매번 형변환을 명시하지 않아도 파이썬이 내부적으로 타입을 맞춰주는 경우, 내부에서 자동으로 변환
    - 정수와 실수의 연산시 자동 형변환 원리
        - **자동 스케일 업**
            - 정수(int) 자료형과 실수(float) 자료형이 사칙연산을 하게 되면, 파이썬이 자동으로 정수를 실수 형태로 임시 변환하여 연산을 수행
            
            ```jsx
            result = 10 + 3.5
            print(result)     # 출력 결과: 13.5
            ```
            
    - 데이터 손실 방지를 위한 업캐스팅(Upcasting)
        - **표현 범위가 넓은 쪽으로 변환**
            - 정수(10)를 실수(10.0)로 바꾸는 것은 데이터 손실이 없지만, 실수(3.5)를 정수(3)로 바꾸면 소수점이 날아가는 데이터 손실이 발생
        - **안전제일 메커니즘**
            - 파이썬은 원본 데이터의 가치를 보존하기 위해, **항상 데이터 표현 범위가 더 넓고 정밀한 실수형(float) 쪽으로 타입을 맞추도록 묵시적 업캐스팅 규칙이 설계**

### 4. 형변환 시 주의해야 할 예외와 에러(ValueError)
---
- 숫자로 바꿀 수 없는 일반 글자를 강제로 변환하려고 하면 프로그램은 오류를 발생시키며 즉시 멈춤
- 데이터 수집이나 외부 API 통신을 할 때 가장 빈번하게 발생

````python
#valueerror 예시
time = "10시"
print(int(time) + 1)
````

````python
#str.isdigit() 메서드 : 위 예시와 같이 valueerror를 해결하려면, 같은 자료형으로 맞춰주면 실행됩니다. 
time = "10시"
if time.isdigit(): #변수.메서드 구조
    print(int(time) + 1)
````

### 5. 거짓으로 인식하는 값
---
- [중요!] 0,"",[],None 은 거짓(False)로 인식
