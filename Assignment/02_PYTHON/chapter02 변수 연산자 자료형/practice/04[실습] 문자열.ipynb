### 필수 1 : 문자열 포매팅 기법과 특징 이해하기
#### 문제 1-1 :  기존 방식과 f-string 출력 비교

```python
# % 연산자 방식
menu_name = 'Americano'
menu_price = 45000

menu = "상품: %s | 가격: %d" 
print(menu %('Americano',45000))
```

```python
# fomat 방식
menu_name = 'Americano'
menu_price = 45000

menu = "상품: {menu_name} | 가격: {menu_price}"
menu.format(menu_name ='Americano',menu_price = 45000)
```

```python
# f-string 방식
menu_name =  'Americano'
menu_price = 45000

print(f"상품:{menu_name} | 가격:{menu_price}")
```

### 필수 2 : 정렬 지시자와 천 단위 쉼표 활용하기

### 심화 1 : 디버깅용 등호 조절자(=) 활용하기
#### 문제 3-1 : 등호(=)를 활용한 데이터 추적

```python
current_status = '정산 진행중'
price =  6500

print(f"디버깅 로그->{current_status=},{price=}")
```

# 문자열 형식화 실습
---
#### 1세대 % 연산자 실습

```python
# 방법
## %s: 문자열, %d: 정수, %.1f: 소수점 첫째 자리 실수
## %15s: 15칸의 넓이를 확보한 후 오른쪽 정렬
# 결과
## 상품 :   Americano | 가격: 4500원 | 할인율: 10.0%

menu_name = "Americano"
menu_price = 4500
discount_rate = 0.1

menu = "상품: %15s | 가격: %d원 | 할인율: %.1f%%" % (menu_name,menu_price,discount_rate)
print(menu)
```

#### 3세대 f-string 및 최신 기능 실습

```python
name = "Cheese Cake"
price = 6500
qty = 1

# 문자열 안에 변수와 가벼운 연산식을 직접 삽입
print(f"주문하신 {name} {qty}개의 최종 청구액은 {price}원 입니다.")

# 3자리 단위 콤마(,) 및 오른쪽 정렬 제어
# :>10, -> 10칸의 자리를 확보하고 오른쪽 정렬한 후 천 단위 쉼표 추가
print(f"포맷팅 적용: {name} | 가격: {price:>10,}원")

# 디버깅 전용 조절자 (=) 사용
# 변수 이름과 현재 값을 동시에 출력하여 로그 확인 시 유용
current_status = "정산 진행중"
print(f"디버깅 로그 -> {current_status=},{price=}")

# 최신 Python 3.12/3.13 기능 (PEP 701)
# 바깥쪽과 안쪽 모두 큰따옴표("")를 중복 사용해도 에러가 나지 않음
menu_dict = {"name": "Cheese Cake", "price": 6500} #딕셔너리{}의 값(calue)가 아닌 변수 부분 출력도 가능!
print(f"딕셔너리 직접 추출: {menu_dict["name"]}의 가격은 {menu_dict["price"]:,}원")
```
